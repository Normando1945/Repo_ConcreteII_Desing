## Scope of the Lecture

A beam does not resist a moment because a formula says it can. It resists because concrete and reinforcing steel develop compatible strains, those strains create stresses, the stresses form internal resultants, and the resultants create an internal couple. This class follows that chain before we introduce the design shortcuts used in the next class.

The governing question is: **what sequence of physical events occurs as moment increases, and which event determines whether the member is safe, usable, and ductile?**

### Fundamental Objectives

This is a mechanics class, not a calendar written as a table. By the end, a section should be readable in four mutually consistent languages: **geometry** (where the material is), **strain** (how it deforms), **stress** (what each material develops), and **resultants** (the forces and moment that equilibrate the applied action). The calculation in the next class is merely a compact way of solving these four descriptions together.

The discussion moves from sectional equilibrium to the linear kinematics of bending, then from constitutive laws to numerical integration of stress. It follows the response through first cracking, yield, and compression deterioration. At every stage, we will distinguish three questions that are often confused: **does the section have strength, is it stiff enough for service, and does its deformation sequence provide ductility?**

### Required and Available Flexural Strength

The member demand is the factored effect obtained from the structural model and applicable load combinations. For flexure, the basic strength check is expressed as:

$$\phi M_n \geq M_u$$

Here, $M_u$ is not a property of the beam: it comes from loading, span, continuity, and analysis. $M_n$ is not the same as the factored moment: it is the section's nominal internal capacity. The factor $\phi$ converts that nominal claim into design strength by recognizing uncertainty and the desired failure mode.

Keeping demand and resistance separate prevents a common mistake: changing reinforcement cannot fix a wrong tributary load, and changing a load combination cannot repair a brittle section.

For a simply supported member under a uniform factored line load $w_u$, the maximum factored moment is $M_u = w_uL^2/8$; support continuity, concentrated loads, frame action, second-order effects, and lateral combinations change that demand model but not the distinction. The structural analysis supplies a distribution of $M_u(x)$ along the member. The section analysis asks whether each critical section has compatible internal resultants that can resist its local value.

Keep dimensions visible. If force is in kN and length in m, moment is kN-m; if force is in N and length in mm, moment is N-mm. A factor of one thousand quietly misplaced between analysis and section design can make a plausible-looking but useless answer.

### Sign Convention

Positive bending normally compresses the top fibres and places the bottom reinforcement in tension; negative bending reverses those roles. The sign is not cosmetic. It identifies where cracking begins, which bars must be developed, which face needs confinement or continuity reinforcement, and which construction joint or support zone becomes critical.

When reviewing any moment diagram, ask four questions: Which face is in compression? Which face is in tension? Where must force enter or leave the reinforcement? Is the assumed detail physically available at that location?

## 1. Flexural Equilibrium

External loads produce shear and bending moment along a member. At a section, the external moment must be balanced by an internal couple. Concrete is efficient in compression; reinforcement carries the tensile force after cracking. The essential equilibrium statement is therefore:

$$C = T, \qquad M = Tz = Cz$$

The lever arm $z$ is not automatically equal to $d$. It depends on the location of both resultants, which in turn depends on the strain and stress fields. This is why section behavior must precede the familiar design equation.

The distinction also separates bending from shear. Bending changes the magnitude and relative location of normal-force resultants over a section; shear is associated with the change of bending moment along the member, $V = dM/dx$, and produces a different stress field. A flexural calculation is not a waiver for shear design, but a correct moment diagram tells us where both mechanisms must be examined.

### Section Geometry and Notation

| Quantity | Physical meaning | Why it matters |
|---|---|---|
| $b, h, d$ | Width, total depth, and effective depth | Geometry controls compression area and lever arm. |
| $A_s, A'_s$ | Tension and compression reinforcement area | Steel amount changes force capacity and ductility. |
| $c, x$ | Neutral-axis depth | Connects strain compatibility to the compression zone. |
| $f'_c, f_y$ | Concrete strength and steel yield strength | Material strengths are not interchangeable with stresses. |
| $\varepsilon_c, \varepsilon_s$ | Concrete and steel strains | Strain identifies yield, crushing, and deformation control. |
| $\rho$ | Reinforcement ratio | A compact measure of how heavily the section is reinforced. |

Every symbol should be introduced through a sketch of the section. A notation list without its physical location is a dictionary with the pages shuffled.

Two ratios will recur. The reinforcement ratio is $\rho = A_s/(bd)$; it compares tensile steel to the available concrete compression geometry. The mechanical ratio $\omega = \rho f_y/f'_c$ further compares steel yield stress to concrete compressive strength. They are useful because different-sized beams can have similar sectional behavior when their geometry and material ratios are similar. They do not replace a compatibility calculation, especially with multiple steel layers or nonrectangular sections.

## 2. Strain Compatibility

Three assumptions form the classical flexure model:

1. Plane sections remain plane after bending. The strain profile is linear through the depth.
2. Reinforcing steel and surrounding concrete have the same strain at their shared level; bond is sufficient to enforce compatibility.
3. Stresses are obtained from the constitutive response of each material.

The first assumption creates a linear strain diagram. The second locates the steel strain from similar triangles. The third converts strain into stress and then into force. If any of these links fails - for example through bond failure, local crushing, or inadequate development length - the simple section model stops describing the real member.

### Material Response

Concrete has a nonlinear compression response, very limited tensile strength, and time-dependent deformation. Reinforcing steel is commonly represented by an elastic branch followed by yielding. The important classroom point is not to memorize one curve: it is to know which part of the curve is active at a given stage and what that implies for stiffness, crack width, force redistribution, and warning before failure.

For preliminary response studies, an elastic-perfectly-plastic steel model and a calibrated concrete compression model are useful. For design strength, the next notebook will introduce the code-compatible equivalent compression block.

## 3. Moment-Curvature Response

Moment is a sectional force effect; curvature measures how rapidly the beam bends. Curvature is associated with the slope of the strain diagram. A moment-curvature curve reveals changes that a single nominal moment cannot show: initial stiffness, crack formation, steel yield, strain hardening or concrete softening, and usable deformation capacity.

A practical interpretation follows five stations:

1. **Uncracked elastic response**: concrete carries both compression and limited tension; stiffness is high.
2. **Cracking**: the extreme tension fibre reaches its tensile limit; stiffness drops abruptly.
3. **Cracked elastic response**: steel increasingly carries tension while concrete carries compression.
4. **Steel yield**: curvature increases rapidly with comparatively modest moment increase; ductility becomes visible.
5. **Compression-controlled deterioration**: crushing or strength loss limits usable curvature.

### Cracking and Section Stiffness

The cracking moment is the flexural effect at which the extreme concrete tension fibre reaches its modulus of rupture. Before this point, an uncracked gross-section model can be useful. After it, the tensile contribution of concrete across the main crack cannot be treated as a durable flexural mechanism.

Cracking matters because stiffness changes. Deflection, redistribution in indeterminate systems, crack width, vibration, and the apparent demand on adjacent elements can all be influenced by that transition. A member may be far from strength failure and already be a serviceability problem.

### Yielding, Crushing, and Failure Mode

A tension-controlled section allows the reinforcement to yield before the concrete compression zone crushes. This produces visible deformation and energy dissipation. A heavily reinforced section can gain nominal strength while losing the warning and rotational capacity that make a structure reliable in reality.

The balanced condition is a reference point where tensile steel reaches yield as concrete reaches its limiting compression strain. It is not the target for ordinary beam design. The desirable side of the boundary is one with adequate tensile strain and a stable, ductile sequence of response.

## 5. Sectional Equilibrium

The familiar pair $C = T$ is the special result for a section under pure bending with no net axial force. Its origin is more general. On any cross-section, the normal stress field $\sigma(y)$ produces an axial resultant and a bending resultant:

$$N = \int_A \sigma(y)\,dA, \qquad M = \int_A y\,\sigma(y)\,dA$$

The coordinate $y$ is measured from a stated reference axis. The first integral asks whether the compression and tension forces balance; the second asks whether their locations create exactly the moment demanded at that section. Discrete reinforcement enters the same equations as concentrated areas: a bar at level $y_i$ contributes $f_{s,i}A_{s,i}$ to axial force and $f_{s,i}A_{s,i}y_i$ to moment.

For a nonprestressed beam with negligible axial load, the signed equilibrium equation becomes $C_c + C_s - T_s = 0$. A reported capacity is credible only if that residual is essentially zero at the proposed neutral-axis depth. Moment is then calculated about any convenient axis; changing the axis changes the individual terms but not the final resultant moment when force equilibrium is satisfied.

### Kinematics of Plane Sections

Let $u(x)$ be the axial displacement of a reference fibre and let $v(x)$ be the transverse displacement. Under the Euler-Bernoulli description, a cross-section rotates but does not warp. The axial displacement at depth $y$ can therefore be written as $u(x,y) = u_0(x) - y\theta(x)$. Differentiating with respect to $x$ gives:

$$\varepsilon(x,y) = \varepsilon_0(x) - y\,\kappa(x)$$

where $\varepsilon_0$ is the strain of the reference fibre and $\kappa = d\theta/dx$ is curvature. Thus the strain diagram is linear because it is a kinematic consequence of plane sections, not because concrete happens to have a linear stress law. In pure bending of a homogeneous elastic beam, the neutral axis is where $\varepsilon = 0$ and the diagram can be written as $\varepsilon(y) = \kappa y$ after choosing that axis.

In a reinforced concrete section, the neutral axis must be found rather than presumed. It migrates as cracking removes concrete tension, as steel yields, and as the compression law becomes nonlinear. The straight strain line remains the compatibility hypothesis; the stress distribution below it may be curved, stepped, or partly zero.

### Strain Compatibility Relationships

Take the neutral axis at depth $c$ from the extreme compression fibre and let the extreme concrete strain be $\varepsilon_c$. For a tension layer at effective depth $d$, compatibility gives:

$$\varepsilon_s = \varepsilon_c\frac{d-c}{c}, \qquad \kappa = \frac{\varepsilon_c}{c}$$

For compression reinforcement at depth $d'$, use $\varepsilon'_s = \varepsilon_c(c-d')/c$ when that layer lies inside the compression zone. These are not design formulas; they are geometry. Once $c$ and one endpoint strain are specified, the strain at every bar follows immediately.

This is the central diagnostic of flexure: a shallow neutral axis produces a large tension-steel strain and usually a ductile sequence; a deep neutral axis produces small tensile strain and moves the section toward compression-controlled behavior. The code later translates that mechanical observation into strength-reduction provisions.

## 6. Constitutive Relationships

For reinforcing steel, the first useful model is elastic-perfectly-plastic. With elastic modulus $E_s$, yield strain is $\varepsilon_y = f_y/E_s$. The stress rule is $f_s = E_s\varepsilon_s$ while the absolute stress remains below $f_y$; after idealized yield, its magnitude is limited to $f_y$. A more refined model may include strain hardening, but it cannot be selected merely because it produces a larger answer.

Concrete is fundamentally different. In compression, its stress rises nonlinearly, reaches a peak, and then softens as damage localizes. In tension, its tensile strength is small relative to its compressive strength; after a dominant flexural crack forms, tension is transmitted mainly by steel and by local tension-stiffening mechanisms between cracks. Therefore the exact stress field for a service or research analysis is obtained from $f_c(\varepsilon)$ and integrated over the compressed region, not assumed rectangular from the outset.

A crucial separation: a **constitutive model** describes material response, while the equivalent compression block introduced later is a calibrated strength-design representation of the resultant compression force. Treating the latter as the actual stress distribution is convenient, but physically false.

### Solution of Sectional Equilibrium

For a specified curvature or a specified extreme concrete strain, the unknown controlling the section is often $c$. Choose a trial $c$; obtain all strains from compatibility; convert them to stresses with the material laws; integrate concrete stress and sum bar forces; then evaluate the axial residual:

$$R(c) = C_c(c) + \sum_i C_{s,i}(c) - \sum_j T_{s,j}(c) - N_{\mathrm{applied}}$$

The compatible section state occurs where $R(c) = 0$. Only after locating that state should moment be assembled from the same force set. This is the mathematical engine behind a fibre-section moment-curvature analysis. It also exposes bad results: if a spreadsheet reports a moment without a force residual, it has not demonstrated equilibrium.

A robust hand calculation mirrors the later algorithm: bracket a plausible neutral axis, calculate the residual at both ends, narrow the interval, and retain a table of $c$, strains, stresses, forces, residual, and moment. The computer will accelerate the iteration; it must not conceal the mechanics.

## 7. Uncracked Elastic Behavior

At very low moment, if concrete tension is retained in the model, both materials are approximately elastic and curvature follows the familiar relation $M = EI\kappa$. The complication is that steel and concrete have different elastic moduli. A transformed section replaces steel area by an equivalent concrete area using the modular ratio $n = E_s/E_c$. For a tension bar, the transformed contribution is commonly $nA_s$ when the displaced concrete is neglected at this level of approximation.

From the transformed geometry, calculate the centroidal neutral axis and transformed second moment of area $I_{tr}$. The elastic strain and stress follow from $\varepsilon = \kappa y$, $f_c = E_c\kappa y$, and $f_s = E_s\kappa y$. The steel stress is therefore approximately $n$ times the concrete stress at the same level, which is precisely why the transformed area works.

This stage is useful for physical interpretation, but it should not be extended past its domain. Once a flexural crack crosses the tensile concrete, gross-section stiffness ceases to represent the local cracked section.

### Cracking Moment, $M_{cr}$

For an uncracked elastic section, the extreme-fibre tension is $f_t = My/I_g$. If the tensile limit is represented by the modulus of rupture $f_r$, the corresponding first-cracking estimate is:

$$M_{cr} = \frac{f_r I_g}{y_t}$$

Here $I_g$ is the gross concrete second moment of area and $y_t$ is the distance from its centroid to the tensile extreme fibre. This expression estimates the onset of a tensile crack under the ideal elastic stress field. It does not mean that all tension disappears everywhere at that instant, nor that the member has lost safety.

After cracking, stiffness drops because the tensile side no longer contributes as a continuous concrete area across the governing crack. Between cracks, bond causes concrete to share some tension with steel; this tension-stiffening effect matters in refined serviceability work, but it does not restore the gross-section stiffness.

### Cracked-Section Analysis

In the simplest cracked-section idealization, tensile concrete is ignored. Let $x$ be the cracked elastic neutral-axis depth from the compression face. For a rectangular section with one tension layer, force equilibrium of the elastic stress triangles leads to the transformed-area condition:

$$\frac{b x^2}{2} = n A_s(d-x)$$

The left term is the first moment of the triangular concrete compression region about its neutral axis; the right term is the transformed steel first moment. Solving it locates $x$. Then the cracked transformed inertia is $I_{cr} = bx^3/3 + nA_s(d-x)^2$. Since $I_{cr}$ is usually far below $I_g$, curvature and deflection rise sharply after cracking for the same moment.

The formula is not a substitute for a code serviceability procedure. Its value here is diagnostic: it shows algebraically why reinforcement location matters so much. Moving the same steel farther from the compression face increases both its lever arm and its contribution to cracked inertia.

## 8. Curvature and Member Deformation

Curvature is not an abstract graph axis. For small rotations, $\kappa = d\theta/dx$ and approximately $\kappa = d^2v/dx^2$. Member rotation is the area under the curvature diagram, $\theta = \int \kappa\,dx$; transverse displacement follows a second integration. Thus a local change in sectional stiffness propagates into deflection, drift, redistribution, and plastic rotation.

In an elastic prismatic region, $\kappa = M/(EI)$. Once cracking or yielding occurs, neither $E$ nor $I$ is globally constant. A moment-curvature analysis replaces one constant stiffness with a sequence of compatible section states. Its tangent slope, $dM/d\kappa$, is the instantaneous flexural stiffness. A steep curve is stiff; a long nearly horizontal post-yield branch is a source of rotation capacity, provided degradation and instability do not intervene.

For a member with nonuniform moment, it is wrong to multiply one curvature by the full span. The curvature must be associated with its location and integrated along the actual moment diagram. This becomes indispensable when estimating rotations in plastic-hinge regions.

### Yield Curvature and Ductility

Steel yield begins when $\varepsilon_s = \varepsilon_y$, not when a chart looks flat. The corresponding section state has a yield curvature $\kappa_y$. Later, under increasing curvature, a section may reach an ultimate curvature $\kappa_u$ when concrete crushing, bar fracture, confinement loss, or another limiting mechanism is reached. A basic curvature-ductility index is $\mu_{\kappa} = \kappa_u/\kappa_y$.

This ratio is useful but incomplete. Structures dissipate energy and accommodate drift through member and connection rotations. Plastic curvature must occur over a finite length, and the effective plastic-hinge length depends on moment gradient, bar development, shear response, confinement, cyclic deterioration, and boundary conditions. A section can have an attractive $\mu_{\kappa}$ but produce poor member ductility if bar slip, shear failure, or instability terminates the response first.

The safe conceptual sequence is therefore: material strain capacity -> sectional curvature capacity -> rotational capacity -> system displacement demand. Skipping any arrow is how a beautiful moment capacity becomes an ugly failure mechanism.

## 9. Dimensionless Section Parameters

The reinforcement ratio $\rho = A_s/(bd)$ is more revealing than steel area alone because it compares steel to the concrete compression geometry available to balance it. Increasing $\rho$ generally deepens the neutral axis. From compatibility, a deeper $c$ reduces $\varepsilon_s$ for a given compression strain, which is the mechanical reason heavily reinforced sections tend to lose tension-controlled behavior.

Effective depth has a particularly strong role. At a fixed reinforcement ratio and comparable material response, internal force scales roughly with $bd$, while lever arm scales with $d$; flexural resistance therefore grows approximately with $bd^2$. This squared depth effect explains why modest increases in effective depth can be more efficient than adding steel, while also benefiting stiffness. It is a scaling insight, not permission to ignore cover, shear, anchorage, or architectural constraints.

Increasing concrete strength changes both the compression resultant and the compression strain behavior. Increasing yield strength raises the stress that steel may develop, but if the same layout is used it can also alter the neutral axis and curvature sequence. The engineer should ask which variable controls the problem: force, stiffness, crack width, rotation capacity, reinforcement congestion, or constructibility. There is no universal direction called 'better'.

## 10. General Section Analysis

Consider a rectangular beam section with one tension layer, no axial force, and a trial compression strain $\varepsilon_c$. The unknown neutral-axis depth $c$ defines the entire strain diagram. The tension-steel strain follows from compatibility, $\varepsilon_s = \varepsilon_c(d-c)/c$. A steel constitutive rule returns $f_s$; a concrete constitutive rule gives the compression stress over the interval from the top fibre to $c$.

The concrete resultant is not guessed. For width $b$, it is $C_c = b\int_0^c f_c(\varepsilon(y))\,dy$. Its line of action is located by the first moment of that same stress field: $y_C = [b\int_0^c y f_c(\varepsilon(y))\,dy]/C_c$. The steel force is $T_s = A_sf_s$. The compatible equilibrium state satisfies $C_c - T_s = 0$.

Once equilibrium closes, the internal moment about the compression resultant is $M = T_s(d-y_C)$. This derivation shows exactly what later code equations compress into a few symbols. It also clarifies why a compression block has two separate jobs: estimate the magnitude of $C_c$ and locate its resultant.

If the steel has not yielded, the solution is coupled because $f_s$ depends on $c$. If it has yielded, its stress may be nearly constant but the concrete resultant still changes with $c$. In neither case is it defensible to select the lever arm before determining the strain state.

### Moment-Curvature Analysis Procedure

A hand-built curve is the best audit trail for a later numerical implementation. Select a sequence of increasing controlling strains or curvatures. At each station, solve force equilibrium for $c$, then record the resulting moment. The table below is the actual state record; it is more important than a smooth plot.

| Step | Prescribed quantity | Solve for | Record | Mechanical interpretation |
|---|---|---|---|---|
| 1 | Small $\varepsilon_c$ or $\kappa$ | $c$ | $M$, steel stress, residual | Nearly elastic response |
| 2 | Tension reaches $f_r$ | $c$ | $M_{cr}$, stiffness change | First flexural crack |
| 3 | Larger $\kappa$ | $c$ | $M$, $I_{sec} = M/\kappa$ | Cracked elastic response |
| 4 | $\varepsilon_s = \varepsilon_y$ | $c$ | $M_y$, $\kappa_y$ | Steel yield |
| 5 | Limiting compression strain | $c$ | $M_u$, $\kappa_u$ | Compression limit or deterioration onset |

At each row, check units explicitly: strains and curvature have dimensions of 1/length; stress times area gives force; force times lever arm gives moment. Also record the force residual, not merely its rounded display value. The curve becomes trustworthy when the transition points have a mechanical explanation and the residual remains acceptably small.

### Scope and Limitations of the Model

Plane sections and perfect bond are powerful idealizations, but they do not certify a member. Near supports, point loads, lap splices, openings, abrupt depth changes, disturbed regions, and anchorage zones, the strain field can depart materially from the one-dimensional flexure assumption. A section calculation also does not prove that shear strength, development length, bar buckling restraint, confinement, durability, construction tolerances, or connection behavior are adequate.

Cyclic loading adds further complications: crack opening and closing, bond deterioration, stiffness and strength degradation, pinching, low-cycle fatigue, and the interaction of axial load with flexure. The monotonic moment-curvature curve remains a foundation, not a complete seismic member model. The correct habit is to name the assumptions beside the result; hiding them is not rigor.

This boundary is especially important when teaching. The goal is not to make the beam appear more complicated than it is. The goal is to know exactly which physical mechanism each equation represents and which mechanisms are being excluded from the equation.

## Summary

Flexural design begins with a sequence, not a coefficient: **moment creates strain; strain creates stress; stress creates internal force; internal force creates resistance; and the deformation sequence determines ductility.**

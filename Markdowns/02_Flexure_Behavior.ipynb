## Class question

A beam does not resist a moment because a formula says it can. It resists because concrete and reinforcing steel develop compatible strains, those strains create stresses, the stresses form internal resultants, and the resultants create an internal couple. This class follows that chain before we introduce the design shortcuts used in the next class.

The governing question is: **what sequence of physical events occurs as moment increases, and which event determines whether the member is safe, usable, and ductile?**

### Required strength and available strength

The member demand is the factored effect obtained from the structural model and applicable load combinations. For flexure, the basic strength check is expressed as:

<div style="margin:18px 0; padding:14px; background:#f8fafc; color:#1e293b; text-align:center; font-family:Georgia,serif; font-size:1.15em;">&phi;M<sub>n</sub> &ge; M<sub>u</sub></div>

Here, <i>M<sub>u</sub></i> is not a property of the beam: it comes from loading, span, continuity, and analysis. <i>M<sub>n</sub></i> is not the same as the factored moment: it is the section's nominal internal capacity. The factor <i>&phi;</i> converts that nominal claim into design strength by recognizing uncertainty and the desired failure mode.

Keeping demand and resistance separate prevents a common mistake: changing reinforcement cannot fix a wrong tributary load, and changing a load combination cannot repair a brittle section.

### Sign convention is a physical statement

Positive bending normally compresses the top fibres and places the bottom reinforcement in tension; negative bending reverses those roles. The sign is not cosmetic. It identifies where cracking begins, which bars must be developed, which face needs confinement or continuity reinforcement, and which construction joint or support zone becomes critical.

When reviewing any moment diagram, ask four questions: Which face is in compression? Which face is in tension? Where must force enter or leave the reinforcement? Is the assumed detail physically available at that location?

## 1. The internal flexural mechanism

External loads produce shear and bending moment along a member. At a section, the external moment must be balanced by an internal couple. Concrete is efficient in compression; reinforcement carries the tensile force after cracking. The essential equilibrium statement is therefore:

<div style="margin:18px 0; text-align:center; font-family:Georgia,serif; font-size:1.12em; color:#1e293b;">C = T &nbsp;&nbsp;&middot;&nbsp;&nbsp; M = Tz = Cz</div>

The lever arm <i>z</i> is not automatically equal to <i>d</i>. It depends on the location of both resultants, which in turn depends on the strain and stress fields. This is why section behavior must precede the familiar design equation.

## 2. Compatibility turns geometry into mechanics

Three assumptions form the classical flexure model:

1. Plane sections remain plane after bending. The strain profile is linear through the depth.
2. Reinforcing steel and surrounding concrete have the same strain at their shared level; bond is sufficient to enforce compatibility.
3. Stresses are obtained from the constitutive response of each material.

The first assumption creates a linear strain diagram. The second locates the steel strain from similar triangles. The third converts strain into stress and then into force. If any of these links fails—for example through bond failure, local crushing, or inadequate development length—the simple section model stops describing the real member.

### Constitutive response is the bridge

Concrete has a nonlinear compression response, very limited tensile strength, and time-dependent deformation. Reinforcing steel is commonly represented by an elastic branch followed by yielding. The important classroom point is not to memorize one curve: it is to know which part of the curve is active at a given stage and what that implies for stiffness, crack width, force redistribution, and warning before failure.

For preliminary response studies, an elastic-perfectly-plastic steel model and a calibrated concrete compression model are useful. For design strength, the next notebook will introduce the code-compatible equivalent compression block.

## 3. The moment-curvature story

Moment is a sectional force effect; curvature measures how rapidly the beam bends. Curvature is associated with the slope of the strain diagram. A moment-curvature curve reveals changes that a single nominal moment cannot show: initial stiffness, crack formation, steel yield, strain hardening or concrete softening, and usable deformation capacity.

A practical interpretation follows five stations:

1. **Uncracked elastic response**: concrete carries both compression and limited tension; stiffness is high.
2. **Cracking**: the extreme tension fibre reaches its tensile limit; stiffness drops abruptly.
3. **Cracked elastic response**: steel increasingly carries tension while concrete carries compression.
4. **Steel yield**: curvature increases rapidly with comparatively modest moment increase; ductility becomes visible.
5. **Compression-controlled deterioration**: crushing or strength loss limits usable curvature.

### Cracking is a transition, not collapse

The cracking moment is the flexural effect at which the extreme concrete tension fibre reaches its modulus of rupture. Before this point, an uncracked gross-section model can be useful. After it, the tensile contribution of concrete across the main crack cannot be treated as a durable flexural mechanism.

Cracking matters because stiffness changes. Deflection, redistribution in indeterminate systems, crack width, vibration, and the apparent demand on adjacent elements can all be influenced by that transition. A member may be far from strength failure and already be a serviceability problem.

### Yield, crushing, and preferred failure

A tension-controlled section allows the reinforcement to yield before the concrete compression zone crushes. This produces visible deformation and energy dissipation. A heavily reinforced section can gain nominal strength while losing the warning and rotational capacity that make a structure reliable in reality.

The balanced condition is a reference point where tensile steel reaches yield as concrete reaches its limiting compression strain. It is not the target for ordinary beam design. The desirable side of the boundary is one with adequate tensile strain and a stable, ductile sequence of response.

## 4. Parameter effects: strength, stiffness, and ductility are not the same variable

| Change | Nominal strength | Cracked stiffness | Deformation capacity | Design interpretation |
|---|---|---|---|---|
| Increase <i>A<sub>s</sub></i> | Usually increases | Often increases | Usually decreases | More steel is not automatically better. |
| Increase <i>d</i> | Strong increase | Strong increase | Often maintained or improved | Depth is an efficient flexural resource. |
| Increase <i>b</i> | Increases | Increases | May improve | Helpful when architectural width is available. |
| Increase <i>f'<sub>c</sub></i> | Moderate increase | Increases | Must be checked | Stronger concrete can be less forgiving in compression. |
| Increase <i>f<sub>y</sub></i> | Increases | Limited direct effect | Can reduce curvature capacity | Strength gain must be read through strain compatibility. |
| Add A'<sub>s</sub> | Can increase | Can improve long-term response | Can improve | Useful for ductility, long-term effects, and congestion-sensitive design. |

## Closing synthesis

Flexural design begins with a sequence, not a coefficient: **moment creates strain; strain creates stress; stress creates internal force; internal force creates resistance; and the deformation sequence determines ductility.**

The next class converts this mechanics into ACI design decisions: the equivalent compression block, strength-reduction factor, reinforcement limits, singly and doubly reinforced beams, and sections with flanges.

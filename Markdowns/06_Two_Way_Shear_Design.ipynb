## Closing the Punching Shear Mechanism

The previous notebook closed with two pieces of the two-way slab design sequence deliberately left open: the shear-triggered minimum reinforcement of 8.6.1.2, which depends on a permissible shear stress not yet developed, and the quantitative punching shear check itself, introduced only conceptually in the behavior notebook two chapters back. This notebook closes both.

Unlike the direct design method, every provision developed here is fully normative in ACI CODE-318-25 as published: the critical section and its geometry (8.4.4, 22.6.4), the concrete two-way shear strength $v_c$ (22.6.5), the eccentric shear stress produced by an unbalanced moment (8.4.2.2, 8.4.4.2), shear reinforcement (22.6.6 through 22.6.8), and the shear-triggered minimum flexural reinforcement (8.6.1.2). No 318-14 cross-reference is needed anywhere in this notebook, a useful contrast with the previous one: the direct design method survives in the Code only by permission and by inheritance, but two-way shear strength is examined, tested, and revised in every new edition, this one included (the size-effect factor $\lambda_s$ used throughout this notebook is itself a product of the ACI 318-19 revision cycle, not a decades-old holdover).

## The Critical Section

Two-way shear is checked not at the face of the column but at a critical section offset a distance $d/2$ from the loaded area, where $d$ is the average effective depth in the two orthogonal directions (22.6.2.1). ACI CODE-318-25 22.6.4.1 places that section so that its perimeter $b_o$ is a minimum, no closer than $d/2$ to the edges or corners of the column, concentrated load, or reaction area, and no closer than $d/2$ to changes in slab thickness such as the edge of a drop panel or shear cap. For an interior column this produces a rectangular perimeter offset $d/2$ outward on all four sides; for an edge column, three sides; for a corner column, two sides, exactly the same interior/edge/corner distinction already used in the previous notebook for the exterior negative-moment tables.

ACI CODE-318-25 Fig. R8.4.4.2.3, critical section geometry (left) and assumed shear stress distribution (right) at an interior column (a) and an edge column (b).</figcaption></figure>

The four corners of the critical section, labelled $A$, $B$, $C$, $D$ in the figure, and the centroidal axis $c$-$c$ running through it, are not decorative labels: they are the reference frame the eccentric shear stress calculation later in this notebook is built on. $c_{AB}$ and $c_{CD}$ are the distances from that centroidal axis to the two faces of the critical section, and they are generally unequal, even for an interior column, whenever the critical section itself is not symmetric about the axis the moment acts on.

ACI CODE-318-25 22.6.4.3 adds a further reduction where an opening in the slab, or a free edge, lies within $4h$ of the critical section: the portion of the perimeter enclosed by straight lines projecting from the centroid of the column toward the boundaries of the opening is treated as ineffective, simply removed from $b_o$, because that part of the critical perimeter can no longer mobilize concrete on both sides of an inclined crack.

## Two-Way Shear Strength Provided by Concrete

For a member without shear reinforcement, nominal two-way shear strength equals $v_c$ alone (22.6.1.2), and $v_c$ itself is the least of three expressions in ACI CODE-318-25 Table 22.6.5.2:

$$v_c=\text{least of}\begin{cases}4\lambda_s\lambda\sqrt{f'_c}&(a)\\\left(2+\dfrac{4}{\beta}\right)\lambda_s\lambda\sqrt{f'_c}&(b)\\\left(\alpha_s\dfrac{d}{b_o}+2\right)\lambda_s\lambda\sqrt{f'_c}&(c)\end{cases}$$

where $\lambda$ is the lightweight-concrete modification factor already used throughout this course (1.0 for normalweight concrete), $\beta$ is the ratio of the long to the short side of the column, concentrated load, or reaction area, $\alpha_s$ is 40 for interior columns, 30 for edge columns, and 20 for corner columns (22.6.5.3), and $\lambda_s$ is the size effect factor developed next.

### The Size Effect Factor

Two-way shear strength does not grow in direct proportion with member depth: a deeper slab, tested at the same $\sqrt{f'_c}$, fails at a lower unit shear stress than a shallow one. ACI CODE-318-25 22.5.5.1.3 captures this size effect with

$$\lambda_s=\sqrt{\frac{2}{1+\dfrac{d}{10}}}\leq1\qquad(d\text{ in inches})$$

This is a genuinely new provision, introduced in the 2019 edition and carried into the current Code, and its constant of 10 is calibrated to $d$ in inches exactly the way the Table 7.3.1.1 thickness modifier was calibrated to $f_y$ in psi in an earlier notebook. ACI CODE-318-25 does not print an SI or MKS form of this expression directly, but ACI 318-19 Appendix C, which tabulates the equivalent forms of every non-homogeneous equation in the Code across the three unit systems, gives the mks-metric ($\mathrm{kgf/cm^2}$, cm) form as

$$\lambda_s=\sqrt{\frac{2}{1+0.04\,d}}\leq1\qquad(d\text{ in centimeters})$$

Note that $0.04=1/25$, not $1/25.4$: ACI's own published conversion rounds the inch-to-centimeter factor to a clean $25$ rather than carrying the exact $2.54$ through, the same kind of deliberate rounding already seen in the Table 7.3.1.1.1 $f_y$ modifier. For a modest slab, $d=20$ cm, this gives $\lambda_s=\sqrt{2/(1+0.04\times20)}=\sqrt{2/1.8}=1.05$, capped at $1.0$: no size-effect penalty. For a thick transfer slab, $d=40$ cm, $\lambda_s=\sqrt{2/(1+0.04\times40)}=\sqrt{2/2.6}=0.88$: a real 12 percent reduction in $v_c$ that a designer working directly in centimeters would miss entirely by plugging $d=40$ into the psi-calibrated $d/10$ term as if the 10 already meant centimeters, and would miss a second time by dropping the outer square root altogether.

### Converting the Strength Coefficients

The coefficients 4, $(2+4/\beta)$, and $(\alpha_sd/b_o+2)$ multiplying $\sqrt{f'_c}$ are not unit-free ratios: $\sqrt{f'_c}$ itself does not carry consistent stress units the way $f'_c$ does, so the whole expression $v_c=C\sqrt{f'_c}$ is an empirical shorthand calibrated to $f'_c$ and $v_c$ both in psi. Converting it requires tracking the conversion through the square root, not just substituting the constant $1\text{ psi}=0.070307\text{ kgf/cm}^2$ directly. Writing $f'_c\text{[psi]}=f'_c\text{[kgf/cm}^2\text{]}/0.070307$,

$$v_c\text{[psi]}=C\sqrt{f'_c\text{[psi]}}=\frac{C}{\sqrt{0.070307}}\sqrt{f'_c\text{[kgf/cm}^2\text{]}}$$

and multiplying both sides by $0.070307$ to convert $v_c$ itself back to $\mathrm{kgf/cm^2}$,

$$v_c\text{[kgf/cm}^2\text{]}=\left(C\times\sqrt{0.070307}\right)\sqrt{f'_c\text{[kgf/cm}^2\text{]}}=0.2652\,C\,\sqrt{f'_c\text{[kgf/cm}^2\text{]}}$$

which is the same mechanism, and the same factor $0.2652$, that ACI 318-19 Appendix C itself used to build the mks-metric column of Table 22.6.5.2 (there labelled by its pre-2019 provision number, but algebraically identical to the current Table 22.6.5.2). Quoting that appendix directly rather than re-deriving and rounding independently:

$$v_c=\text{least of}\begin{cases}1.1\,\lambda_s\lambda\sqrt{f'_c}&(a)\\0.53\left(1+\dfrac{2}{\beta}\right)\lambda_s\lambda\sqrt{f'_c}&(b)\\0.27\left(2+\alpha_s\dfrac{d}{b_o}\right)\lambda_s\lambda\sqrt{f'_c}&(c)\end{cases}\qquad(f'_c,\ v_c\text{ in kgf/cm}^2)$$

with $\alpha_s$ unchanged at 40, 30, and 20 for interior, edge, and corner columns: the appendix keeps $\alpha_s$ inside the same bracket that gets scaled by $0.27$, rather than pre-multiplying $\alpha_s$ itself, and this notebook follows that same published form rather than distributing the factor into three separate rounded constants.

This is the most consequential unit conversion in this course so far, because getting it wrong does not merely mis-size a member, it overstates shear capacity on the brittle failure mode this entire unit has been building toward. For a typical $f'_c=280$ $\mathrm{kgf/cm^2}$, expression (a) correctly gives $v_c=1.1\sqrt{280}=18.4$ $\mathrm{kgf/cm^2}$. A designer who instead substitutes $f'_c=280$ directly into the psi-calibrated "$4\sqrt{f'_c}$" without converting the coefficient obtains $4\sqrt{280}=66.9$ $\mathrm{kgf/cm^2}$, nearly 3.6 times too high. That is not a conservative rounding error to be caught later; it is an unsafe overestimate of the strength governing the single most abrupt, least ductile failure mode a two-way slab can exhibit.

## Eccentric Shear Transfer

Whenever a slab-column connection must transfer an unbalanced moment $M_{sc}$, gravity pattern loading at an interior column, or the geometry itself at every edge and corner column, ACI CODE-318-25 8.4.2.2.1 requires that fraction of $M_{sc}$ to be transferred partly by flexure and partly by eccentricity of shear, in fixed proportions set by the geometry of the critical section alone.

### Splitting the Moment: Flexure and Shear

The fraction transferred by flexure, $\gamma_f$, is (8.4.2.2.2)

$$\gamma_f=\frac{1}{1+\left(\dfrac{2}{3}\right)\sqrt{b_1/b_2}}$$

where $b_1$ is the width of the critical section measured in the direction of the span for which moments are being determined, and $b_2$ is the width measured perpendicular to it. This ratio is unit-free, so it needs no conversion in any consistent system. The remaining fraction is transferred by eccentricity of shear (8.4.4.2.2):

$$\gamma_v=1-\gamma_f$$

Physically, $\gamma_f$ grows as the critical section becomes wide in the direction resisting the moment ($b_1$ large relative to $b_2$): a wide section can develop flexural reinforcement effectively across $b_{slab}$ (8.4.2.2.3, the column or capital width plus $1.5h$ on each side), leaving less of the moment to be carried by the less ductile shear-eccentricity mechanism. ACI CODE-318-25 8.4.2.2.4 permits $\gamma_f$ to be increased beyond this base value at connections with sufficient ductility and shear headroom below the Table 8.4.2.2.4 limits, precisely because shifting more of $M_{sc}$ onto flexural reinforcement is the safer direction to move in.

### The Shear Stress Produced by $\gamma_vM_{sc}$

ACI CODE-318-25 8.4.4.2.3 assumes the shear stress produced by $\gamma_vM_{sc}$ varies linearly about the centroidal axis $c$-$c$ of the critical section, exactly the linear stress distribution already used for combined axial load and bending on a cross section, $\sigma=P/A\pm Mc/I$, except that here a shear stress varies around a perimeter rather than a normal stress varying across an area:

$$v_{u,AB}=\frac{V_u}{A_c}+\frac{\gamma_vM_{sc}\,c_{AB}}{J_c}\qquad v_{u,CD}=\frac{V_u}{A_c}-\frac{\gamma_vM_{sc}\,c_{CD}}{J_c}$$

where $A_c=b_od$ is the area of the critical section, and $J_c$ plays the role of a polar moment of inertia for that section. For an interior column, with $c_1$ and $c_2$ the column dimensions in the two directions,

$$J_c=\frac{d(c_1+d)^3}{6}+\frac{d^3(c_1+d)}{6}+\frac{d(c_2+d)(c_1+d)^2}{2}$$

Because $c_{AB}$ and $c_{CD}$ measured from the centroidal axis are generally unequal, $v_{u,AB}$ and $v_{u,CD}$ are unequal too, and the larger of the two, always the face where the direct and eccentric contributions add rather than subtract, governs the punching check. $J_c$ itself is built entirely from lengths, so, like $\gamma_f$ and $\gamma_v$, it requires no unit-calibrated conversion constant: computed with $d$, $c_1$, and $c_2$ in cm throughout, it returns a $J_c$ in $\mathrm{cm^4}$ that combines directly with $M_{sc}$ in $\mathrm{kgf\cdot cm}$ and $c_{AB}$/$c_{CD}$ in cm to give a shear stress in $\mathrm{kgf/cm^2}$, exactly as it would in inches and pounds.

## Design Strength

The two-way shear design check itself is stated compactly in ACI CODE-318-25 8.5.1.1(d): $\phi v_n\geq v_u$ at the critical section defined in 8.4.4.1, where $v_u$ is the larger of $v_{u,AB}$ and $v_{u,CD}$ from the eccentric shear calculation above (or simply $V_u/A_c$ where no moment transfer governs), and $v_n=v_c$ for a member without shear reinforcement. The strength reduction factor $\phi$ for shear follows 21.2, the same $\phi$ family already used throughout this course, evaluated at the tension-controlled or shear value rather than the flexural one.

If $\phi v_c<v_u$ at this stage, the designer has three physical options, not one: increase the slab thickness (which increases both $d$ and, through $\lambda_s$, the size-effect factor), enlarge the column or add a capital or drop panel (which enlarges $b_o$ directly), or add shear reinforcement. The next section develops that third option.

## Shear Reinforcement

Where $v_u$ exceeds $\phi v_c$ at the critical section adjacent to the column, ACI CODE-318-25 permits two families of two-way shear reinforcement: single- or multi-leg stirrups (22.6.7, detailed per 8.7.6) and headed shear stud reinforcement (22.6.8, detailed per 8.7.7). Adding shear reinforcement does two things at once: it raises the permissible $v_c$ at the critical section immediately adjacent to the column, and it introduces a second, larger critical section further out, located $d/2$ beyond the outermost line of shear reinforcement, where the slab must again be checked without help from that reinforcement.

The inner critical section, through the first line of stirrup legs or stud heads, stays the same rectangular (or three- or two-sided) perimeter already developed above. The outer critical section, required by 22.6.4.2, is not simply that same rectangle shifted outward: ACI CODE-318-25 requires it to be a polygon chosen to minimize $b_o$, which is why the figures show an octagonal shape at an interior column rather than a larger rectangle. A rectangular perimeter offset $d/2$ beyond the reinforcement would overstate $b_o$ at the corners, understating the average shear stress precisely where the slab is most likely to punch just outside the reinforced zone.

### Permissible Stress at Each Section

At the inner critical section, ACI CODE-318-25 Table 22.6.6.1 raises the permissible $v_c$ above the unreinforced value: $2\lambda_s\lambda\sqrt{f'_c}$ with stirrups, $3\lambda_s\lambda\sqrt{f'_c}$ with headed shear studs, reflecting the higher effectiveness headed studs have shown at anchoring the compression strut around the column. Converting with the same $0.2652$ factor derived above, these become $0.53\lambda_s\lambda\sqrt{f'_c}$ (stirrups) and $0.80\lambda_s\lambda\sqrt{f'_c}$ (studs), in $\mathrm{kgf/cm^2}$. At the outer critical section, $v_c$ reverts to essentially the unreinforced expressions of Table 22.6.5.2, capped at $2\lambda_s\lambda\sqrt{f'_c}$ regardless of reinforcement type, because outside the reinforced zone the slab must again rely on concrete alone.

Section 22.6.6.2 allows $\lambda_s$ to be taken as 1.0, recovering the full size-effect-free strength, where the shear reinforcement itself is detailed with at least a minimum ratio $A_v/s\geq2\sqrt{f'_c}\,b_o/f_{yt}$ (converted the same way: $2\times0.2652=0.53$, so $A_v/s\geq0.53\sqrt{f'_c}\,b_o/f_{yt}$ in $\mathrm{kgf/cm^2}$). Physically, a slab reinforced generously enough crosses the inclined crack with enough steel that the size effect governing lightly reinforced, unreinforced sections no longer controls.

## The Shear-Triggered Minimum Reinforcement

The previous notebook developed the ordinary flexural minimum reinforcement, $A_{s,min}=0.0018A_g$ (8.6.1.1), and deferred a second, larger minimum that ACI CODE-318-25 8.6.1.2 triggers whenever

$$v_{uv}>\phi\,2\sqrt{f'_c}\,\lambda_s\lambda$$

on the critical section for two-way shear, that is, whenever the direct shear stress alone (excluding the eccentric contribution) already exceeds half of the maximum unreinforced strength $4\lambda_s\lambda\sqrt{f'_c}$. Where this condition governs, the minimum flexural reinforcement over the width $b_{slab}$ must satisfy

$$A_{s,min}=\frac{5\,v_{uv}\,b_{slab}\,b_o}{\phi\,\alpha_s\,f_y}$$

This is not a second, independent shear check; it is a ductility safeguard. The commentary explains the mechanism directly: in lightly reinforced slabs, the flexural tension reinforcement near the column can yield under load, and once it yields, local rotation increases and any inclined crack already present within the slab tends to open further, allowing the two crack faces to slide rather than stay locked in aggregate interlock. That sliding is what triggers a flexure-driven punching failure at a shear demand below the level $4\lambda_s\lambda\sqrt{f'_c}$ would otherwise suggest is safe. Requiring more flexural reinforcement across $b_{slab}$ when $v_{uv}$ already runs high keeps the section from reaching that yielding condition in the first place, which is why $A_{s,min}$ here is a flexural quantity governed by a shear trigger rather than the other way around.

Every quantity in this expression is already in the units this course uses, kgf, cm, $\mathrm{kgf/cm^2}$, once $v_{uv}$ and $f_y$ are expressed consistently; no additional conversion constant is needed beyond the $v_c$ conversion already carried through the trigger condition itself.

## Summary

Two-way shear is checked at a critical section offset $d/2$ from the loaded area, a rectangular, three-sided, or two-sided perimeter for interior, edge, and corner columns respectively, and reduced further wherever an opening or free edge lies within $4h$. Concrete strength alone, $v_c$, is the least of three expressions in Table 22.6.5.2, all calibrated to psi and requiring the derived factor $0.2652$ to convert correctly to $\mathrm{kgf/cm^2}$, a conversion whose omission overstates capacity by nearly a factor of four on the most brittle failure mode this course has covered. The size effect factor $\lambda_s=2/(1+d/25.4)$ (d in cm) further reduces that strength for thick slabs. Wherever a connection must also transfer an unbalanced moment, that moment splits between flexure ($\gamma_fM_{sc}$) and eccentric shear ($\gamma_vM_{sc}$, producing a linearly varying stress through $J_c$), and the larger of the two resulting face stresses governs. Where $v_u$ exceeds $\phi v_c$, stirrups or headed shear studs raise the permissible stress at the column face and introduce a second, polygonal critical section further out; and wherever the direct shear stress alone exceeds half the unreinforced concrete strength, a shear-triggered minimum flexural reinforcement, $A_{s,min}=5v_{uv}b_{slab}b_o/(\phi\alpha_sf_y)$, guards against a brittle, flexure-driven punching failure.

## ACI CODE-318-25 Reading Plan: Slabs, Columns, and Beam-Column Connections

This plan is a self-study companion to three topics of this course: slabs, columns, and beam-column (and slab-column) connections. It is not a summary of ACI CODE-318-25, and it does not replace the course notebooks or lectures. It is a schedule for reading the Code and Commentary text itself, directly from your own licensed or library copy, so that by the end of the semester you have gone through, in your own words and at your own pace, every chapter and every specific numbered provision that governs these three topics.

This plan does **not** cover the entire ACI CODE-318-25. It is intentionally narrow: only Chapter 7 (One-Way Slabs), Chapter 8 (Two-Way Slabs), Chapter 10 (Columns), and Chapter 15 (Cast-in-Place Joints, which is where ACI places beam-column and slab-column connections) are read in full. Every other numbered provision that appears in this plan is there because one of those four chapters explicitly sends you to it: a required-strength clause that says "calculated in accordance with 22.4," a detailing clause that says "refer to 25.7.2," and so on.

### How to Use This Plan

Plan for 3 hours per week, in a single sitting or split across two. A reasonable way to spend that time:

- About half the time on a first, careful pass through the assigned Code text and its facing Commentary (the "R" sections). Read the Code requirement first, then immediately read the Commentary that explains why the requirement exists, before moving to the next provision.
- About a third of the time following the specific cross-references assigned for that week (a clause that points to a numeral in another chapter) and reading that numeral in its own chapter, not just taking the pointer on faith.
- The remainder reconciling what you read with the corresponding course notebook and, where one already exists, with the manual example worked in class for that topic. If a symbol, a limit, or a derivation in the notebook does not match what the Code says, that mismatch is worth resolving before the next class, not after.

Several weeks in this plan are marked as consolidation weeks. On those weeks there is little or no new material; the three hours go into rereading, cross-checking, and reconciling what earlier weeks already covered. This is deliberate. A code provision that is only read once is rarely retained; the plan is paced so that Slabs, Columns, and Connections are each read at least twice before the semester ends.

### Legend

- **Ch.** refers to a full chapter; a plain number such as **22.4** refers to a specific numeral (Code plus its Commentary, printed together).
- Page numbers below refer to the specific ACI CODE-318-25 printing kept for this course. If your own copy is a different printing or a different edition, use the numeral (for example, 22.4) to locate the section; do not rely on the page number alone.
- "Supporting numeral" means a provision from a chapter other than 7, 8, 10, or 15, read only because one of those four chapters cites it directly.

### Scope Note on Seismic Provisions

Chapter 15 explicitly distinguishes an ordinary joint from one that is part of a designated seismic-force-resisting system, and for the latter it points to 18.8 (Joints of Special Moment Frames) in Chapter 18. Chapter 10 similarly notes, in general terms, that columns required to resist earthquake effects are designed in accordance with Chapter 18 as a whole. This course keeps "capacity design and ductile detailing" as its own later topic, so this plan does **not** open Chapter 18 as a chapter. The single numeral 18.8.4.3, which Chapter 15 cites directly for joint shear strength, is the only piece of Chapter 18 read here, and it is read strictly as a forward pointer: enough to see that a more demanding version of joint design exists, not to master it yet. Everything else in Chapter 18, including 18.7 (Columns of Special Moment Frames), is left for the seismic unit.

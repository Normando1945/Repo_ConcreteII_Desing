## Introduction

Reinforced Concrete Design II begins at the point where a member is no longer treated as an isolated calculation. A column carries the accumulated decisions of the slabs and beams above it; a slab depends on supports, continuity, cracking, deflection, and construction sequence; a wall that is strong on paper can still fail the building if its load path, boundary zones, or connections are not resolved.

The central question is therefore not *which equation should be used first?* It is: **what physical mechanism is expected to carry the action, and can that mechanism actually be built and sustained throughout the life of the structure?**

This notebook establishes the language for answering that question. It connects structural behavior, idealized analysis, code requirements, detailing, and construction into one design process.

### A first design lens

A building is never designed for one isolated number. It is designed for a sequence of events: loads are applied, forces are transferred, materials deform, cracks form, reinforcement engages, and the structural system either remains stable or reveals a weakness. The task of the engineer is to anticipate that sequence before the building exists.

For every member studied in this course, keep four questions visible:

1. **What action reaches the member?**
2. **Which internal mechanism resists that action?**
3. **Which failure mode must be prevented or controlled?**
4. **What reinforcement and detailing make the intended mechanism physically possible?**

These questions are more useful than starting with an equation, because they identify the physical claim that every equation must support.

## A short history: from empirical construction to design codes

For most of construction history, builders learned through observation, craft, precedent, and occasional failure. Masonry, lime mortars, timber, stone arches, and early cementitious materials made remarkable structures possible, but the design rules were largely empirical: they were refined because they had worked before, not because every internal force could be calculated.

The nineteenth century changed the scale of the question. The industrial production of Portland cement created a more consistent binding material, and the deliberate combination of concrete with iron and then steel reinforcement made it possible to use a material that could work in both compression and tension. Reinforced concrete began to move from a construction technique toward an engineered structural system.

That transition matters. A reinforced-concrete member is not simply concrete with steel inside it. It is a composite mechanism in which geometry, bond, cracking, strain, confinement, and detailing determine whether the materials can work together.

### From analysis to design

As structural mechanics developed, engineers gained increasingly useful tools to connect loads, internal forces, stress, strain, and displacement. Equilibrium alone was not enough; a credible solution also needed compatible deformations and realistic material behavior.

In the classroom, it is tempting to treat analysis as the final goal because equations and software return precise-looking values. In practice, analysis is a means to a design decision. The engineer must still decide whether the force path is credible, whether the model represents the intended behavior, whether the failure mechanism is acceptable, and whether the reinforcement can be detailed and constructed.

This distinction is particularly important for reinforced concrete. We often analyze structures with linear-elastic models under service-level behavior, but we design critical regions so that their inelastic response remains controlled when demands become severe.

### What a structural model is allowed to claim

Structural analysis is powerful because it deliberately leaves things out. A model may represent a building with line elements, restrained nodes, load patterns, elastic stiffnesses, and a finite set of combinations. None of those objects is the building itself. They are a controlled claim about its equilibrium, deformation, and force transfer.

For reinforced concrete, that distinction matters immediately. The analysis may be elastic while a critical section is expected to crack, yield locally, redistribute force, or dissipate energy. The design is credible only when the idealization, the expected mechanism, and the detailing tell the same story.

Equilibrium asks whether the forces can balance. Compatibility asks whether the assumed deformations can occur. Constitutive relations ask what stresses the materials develop at those strains. Every later model in this course is an intentionally limited combination of those three ideas.

### Concrete, steel, and an unfinished experiment

Reinforced concrete was revolutionary because it made a difficult partnership useful. Concrete is strong in compression but weak and brittle in tension. Steel is strong and ductile in tension, but it needs protection against corrosion and fire and must be anchored so that its force can be transferred. Their similar thermal expansion and their mechanical bond make composite action possible, but only when the reinforcement is properly located and the concrete surrounding it can perform.

This means every reinforced-concrete member is, in a sense, an experiment in compatibility. When load increases, concrete may crack; after cracking, force redistributes toward the steel; when steel yields, stiffness changes; when compression strains become large, confinement and section geometry become decisive. The member is not behaving incorrectly when these phenomena occur. They are the behavior that our design models are trying to anticipate and control.

A material model cannot by itself guarantee a safe building. A system analysis cannot by itself guarantee that a lap splice is placed correctly. The course will repeatedly move across these scales, because each one supplies assumptions for the next.

### Why codes change

A code is often experienced by students as a dense collection of restrictions. A better interpretation is that it is a negotiated technical memory. Research reveals mechanisms that were not represented well by older equations. Field performance reveals what happens when construction conditions differ from laboratory assumptions. Earthquakes, fires, durability problems, and failures expose weak links that may not be obvious in a clean calculation.

For this reason, the history of ACI 318 is not administrative trivia. Its evolution from early building regulations to a modern code with detailed provisions for strength, serviceability, durability, anchorage, construction, and seismic design reflects the expanding range of questions that structural concrete must answer. Reading a provision should therefore lead to two questions: **what behavior is this trying to control, and what unsafe outcome is it trying to avoid?**

This course will use ACI CODE-318-25 as its principal reference, while always recognizing that the governing local building code and project conditions control real design practice.

## Reinforced concrete as a structural material

Concrete and reinforcing steel form a composite material because their individual limitations complement one another. Concrete provides stiffness, mass, durability, fire resistance, and strong compressive capacity. Reinforcing steel provides the tensile strength and ductility that plain concrete does not possess.

That complement is powerful, but it is not automatic. A reinforced-concrete member only behaves as intended when concrete and steel can transfer force through bond, reinforcement is located correctly, cracking is anticipated, compression zones remain stable, and the bars have adequate cover, spacing, anchorage, and development length.

> **Design is the act of turning a structural idea into a safe, ductile, durable, and constructible physical system.** Analysis is indispensable, but it is only one part of that act.

## The structural system sets the demand

Before a section is proportioned, the engineer must identify the complete mechanism that delivers action to that section. Tributary area, support conditions, continuity, diaphragm action, member stiffness, and the structural system determine the internal effects that the section will be asked to resist. A beam does not receive a moment from nowhere; a column does not receive an axial load without the accumulated decisions of the floors above it.

The same reasoning applies to lateral actions. Wind and earthquake demand depend on mass, stiffness, configuration, boundary conditions, and the selected lateral-force-resisting system. The purpose of analysis is to make those transfers explicit enough to test them—not to hide them behind a final value of $M_u$, $V_u$, or $P_u$.

### Questions that precede a calculation

- What is the structural system and what is the load path?
- Which elements collect, transfer, and resist each action?
- Where are the critical regions expected to form?
- Which members are controlled by flexure, axial force, shear, serviceability, or detailing?
- Under severe lateral loading, where should inelastic deformation occur, and where must it be prevented?

Only after these questions have a reasonable answer does a section calculation become meaningful.

## Behavior, idealization, and design

Real reinforced concrete is nonlinear. Concrete cracks in tension, its compression response changes with strain, and it may lose capacity after reaching its peak stress. Steel yields and can sustain significant inelastic deformation before fracture when it is properly selected and detailed. Bond, confinement, creep, shrinkage, and load history also influence the response of a real member.

A complete structural model cannot represent every aggregate particle, crack, bar rib, and construction imperfection. Engineering design therefore uses idealizations: models simple enough to calculate and conservative enough to support safe decisions.

| Principle | Question answered | Reinforced-concrete interpretation |
| --- | --- | --- |
| **Equilibrium** | Are all forces and moments balanced? | Internal compression, tension, shear, and external actions must balance. |
| **Compatibility** | Can the assumed deformation pattern occur? | Strain varies continuously through a bonded section, except across modeled discontinuities such as cracks. |
| **Constitutive behavior** | How do materials transform strain into stress? | Concrete and steel develop stress according to their idealized material response. |

The rectangular stress block, strain-compatibility diagrams, interaction diagrams, shear models, and development-length expressions are all tools built from these principles. They are not arbitrary formulas: their assumptions define when they can be used and when a different model is needed.

## Strength design and uncertainty

Structural loads, material properties, construction quality, and future use are never known with perfect certainty. Strength design addresses this uncertainty by increasing the actions used for design and reducing the nominal resistance credited to a member.

$$
U \leq \phi R_n
$$

where $U$ is the required strength from factored load combinations, $R_n$ is the nominal resistance predicted by the selected model, and $\phi$ is the strength-reduction factor appropriate to the failure mode and deformation condition.

This expression is concise, but the design work behind it is not. The engineer must define the relevant load combination, identify the governing limit state, use compatible units, choose an appropriate section model, and verify that the detailing assumptions used by the model are satisfied.

## Limit states are performance boundaries

A limit state is reached when a structure, a member, or a detail no longer meets the performance expected from it. The phrase does not mean only collapse. It asks a sharper question: *which unacceptable outcome is being prevented, at which scale, and under which demand?*

### Strength limit states

Strength limit states concern loss of equilibrium, instability, fracture, crushing, shear failure, loss of anchorage, punching, or formation of an uncontrolled mechanism. Their probability must be very small because the consequence may involve a local failure, a partial collapse, or loss of the entire load path. In reinforced concrete, the preferred response is often one that gives visible warning and deformation capacity before a brittle mode can govern.

### Serviceability limit states

A structure can remain standing and still fail its intended use. Excessive deflection can damage partitions and façades; cracking can compromise durability or appearance; vibration can make a floor unacceptable; water leakage can reveal an inadequate crack-control or detailing decision. These effects are evaluated at service-level demands because the building is occupied and judged there every day, not only at a factored extreme.

### Exceptional and durability-related limit states

Fire, impact, earthquake, corrosion, construction-stage loading, and accidental actions may change both the demand and the mechanisms that matter. The correct response is not a universal detail added at the end of the calculation. It is a design strategy that identifies the exposure, preserves continuity, and assigns deformation or protection where it is actually needed.

## From external action to internal resistance

Loads do not act on a section as a ready-made moment, shear, or axial force. They enter the structural system through tributary areas, diaphragms, supports, connections, and geometry. Analysis transforms those actions into **load effects**; section mechanics transforms strain and stress into **internal resultants**. Design is the disciplined comparison between the two.

For a flexural member, the external moment is resisted by an internal couple: compression in concrete and tension in reinforcement separated by a lever arm. For a column, axial compression and moment produce a strain field whose neutral axis, steel response, and compression block move together. For shear, the resisting mechanism involves cracked concrete, aggregate interlock, longitudinal reinforcement, transverse reinforcement, and the geometry of the member. Each check is therefore a statement about a mechanism, not a detached algebraic ritual.

The compact strength-design inequality is useful only after that chain is explicit:

$$\phi R_n \geq U$$

Here, $U$ is the factored effect obtained from the relevant load combinations. $R_n$ is a nominal resistance associated with a particular failure mode and a particular model. The factor $\phi$ recognizes that resistance is not known exactly and that some modes are less forgiving than others. A check is incomplete when any one of those terms is borrowed from an incompatible model, unit system, or detailing assumption.

### Why load factors and strength-reduction factors exist

Material strengths vary between batches; member dimensions vary from drawings; construction tolerances, curing, placement, and local congestion influence the realized member. Loads also vary in magnitude, duration, combination, and future use. Finally, the analytical model itself is an approximation. The safety format does not eliminate these uncertainties; it makes them explicit enough to calibrate a consistent level of reliability.

That is why a brittle failure mode deserves less credited resistance than a ductile one, even when their calculated nominal strengths appear similar. Ductility offers deformation, redistribution, and warning. Brittle mechanisms can lose capacity with little reserve. The design factors therefore encode both statistical variability and the engineering consequences of how a member fails.

The practical lesson is simple but demanding: never interpret $\phi$ as a decorative coefficient. It is attached to a specific response, strain condition, and code model. Changing the failure mode changes the meaning of the check.

## A member must do more than avoid collapse

### Strength
The member must resist factored actions associated with flexure, axial force, shear, torsion, bearing, or their interaction. Ductile mechanisms are preferred whenever practical; brittle modes related to shear, anchorage, or inadequate confinement require special attention.

### Serviceability
Structures are used long before they approach nominal strength. Deflection, cracking, vibration, and other service-level effects can govern the size or reinforcement of a member even when its strength check is satisfactory.

### Durability
Concrete cover, exposure conditions, mixture requirements, crack control, and construction quality affect whether a structure can perform throughout its intended life.

### Detailing and constructibility
Bars need space. Stirrups, ties, hooks, splices, anchorage zones, and congested joints need to be drawn and reviewed. The best reinforcement arrangement is not necessarily the one with the smallest calculated steel area; it is the one that safely develops the intended forces and can be placed, consolidated, inspected, and maintained.

### System behavior
Members do not act alone. A ductile beam attached to an inadequately detailed joint, a well-designed column on an insufficient foundation, or a wall with an interrupted load path cannot deliver the performance expected from an individual calculation.

## The role of detailing in seismic behavior

Under severe earthquake loading, the design objective is not always to prevent every crack or every inelastic deformation. Instead, the structure should develop predictable and stable mechanisms while protecting critical regions from premature or brittle failure.

This is why seismic design gives such importance to confinement, transverse reinforcement, anchorage, lap-splice locations, joint behavior, shear strength, and capacity-based checks. Detailing is not an administrative appendix to analysis. It is the physical mechanism that allows the assumed behavior to occur.

> The same lesson applies outside seismic design: an equation is only as reliable as the member that can actually realize the assumptions behind that equation.

## Course objectives

By the end of the course, students should be able to:

- interpret the behavior of common reinforced-concrete members under gravity and lateral actions;
- identify credible load paths and critical regions within a structural system;
- apply strength-design principles to members subjected to flexure, axial load, shear, and their interaction;
- design and interpret moment-axial-force interaction diagrams for reinforced-concrete columns;
- design and detail common one-way and two-way slab systems;
- evaluate shear, serviceability, reinforcement limits, and detailing requirements;
- explain how confinement and reinforcement detailing affect ductility and failure mode;
- organize a transparent hand-calculation procedure; and
- use Python notebooks to reproduce, inspect, and extend calculations without treating them as black boxes.

## A design agreement

Throughout this course, a result is not complete merely because it satisfies a numerical inequality. Each design must make its assumptions visible, identify the governing limit state, preserve a continuous force path, and show reinforcement that can transfer the intended forces without impossible congestion or unsupported detailing.

This is the standard used to read hand calculations, models, drawings, and notebooks: **a numerical result must remain physically credible when translated into a member and then into a building.**

## Learning through hand calculations and notebooks

Hand calculations and computational routines have different roles in this repository. Hand calculations make engineering logic visible: they force the designer to state assumptions, select the governing section, trace units, and interpret intermediate results. Notebooks make repetitive calculations reproducible, enable parameter studies, and allow the behavior of a section or member to be explored beyond a single numerical case.

Neither replaces the other. A notebook should reproduce a procedure that the student can explain. A hand calculation should be checked with enough discipline that it can be implemented safely in code.

For each topic, the intended sequence is:

1. study the physical behavior and relevant limit states;
2. identify code requirements and design assumptions;
3. complete a clear hand procedure;
4. review reinforcement layout and constructibility;
5. execute the corresponding notebook step by step; and
6. interpret the result, including the conditions under which it is no longer valid.

## Normative framework

A design procedure is never detached from its regulatory context. This course uses **ACI CODE-318-25** as its principal concrete-design reference. In Ecuador, the applicable provisions of the **Norma Ecuatoriana de la Construcción (NEC)** and the requirements of the authority having jurisdiction must be checked for each project. Standards such as ASCE 7 and ASCE 41 are valuable companions when the problem involves load definition, existing structures, or seismic evaluation, but they do not replace the governing local code.

The visual references below identify the families of documents that shape the design conversation. Their displayed editions are resources from the source course page; a real design must always verify the edition adopted by the project and jurisdiction.

## Reference standard and responsible use

The principal concrete-design reference for this repository is **ACI CODE-318-25, _Building Code Requirements for Structural Concrete and Commentary_**. The applicable building code and local regulations always govern a real project. Course notes will identify their assumptions and unit system, but every calculation must remain internally consistent.

> **Educational use only.** These notes, examples, and notebooks are intended for learning, discussion, and verification of instructional problems. They do not replace professional judgment, project-specific analysis, applicable regulations, construction documents, or independent design review.

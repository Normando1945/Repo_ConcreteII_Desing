## Shear in the Column: A Second Look at the Same Section

Columns I (notebook 07) answered one question about a cross section: which pairs $(P_n, M_n)$ it can develop. A column in a frame carries a third internal force that the interaction diagram does not see. Wherever the moment changes along the height of the column, equilibrium requires a shear force:

$$V=\frac{dM}{dy}$$

In a column bent in double curvature by end moments $M_{top}$ and $M_{bot}$, the moment varies linearly over the clear height $\ell$ and the shear is constant:

$$V=\frac{M_{top}+M_{bot}}{\ell}$$

Under gravity loads the end moments of an interior column are small and nearly cancel, and the shear is negligible. Under lateral loads (wind, earthquake) the story drift bends every column in double curvature: the two end moments act on the column in the same rotational sense, their effects on the shear add instead of cancelling, and the shear can become large. The formula also shows the danger hidden in the denominator: for the same end moments, a shorter column carries more shear.

Shear failure is the one a designer most wants to avoid. Flexural failure of a properly reinforced section is announced by yielding steel, wide cracks, and large deflections. Shear failure is announced by an inclined crack and, if the section is not reinforced for it, follows almost immediately. In a column, which carries the floors above it, a brittle failure can mean progressive collapse.

This notebook designs the shear reinforcement of the 30 x 50 cm tied column of notebook 07 under the same axial load $P_u=120$ tf used in its design check. Since the shear reinforcement of a column is its ties, the same notebook also covers the detailing of transverse reinforcement: tie size and spacing, lateral support of the longitudinal bars, spirals, and splices. The effect of length on moments (slenderness) and bending about two axes are left for Columns III (notebook 09).

## A Short History of Shear in Reinforced Concrete Columns

### The Truss Analogy

The first rational model of shear in reinforced concrete is due to the Swiss engineer Wilhelm Ritter, who in 1899 proposed that a cracked beam behaves like a truss: the longitudinal reinforcement is the tension chord, the concrete above the neutral axis is the compression chord, the stirrups are vertical tension members, and the concrete between inclined cracks forms diagonal struts at 45 degrees. Emil M&ouml;rsch developed the model in the following decade into a design method. In its classical form the truss assigns all the shear to the stirrups, and tests soon showed it to be conservative: beams carried more shear than the stirrups alone could explain. The difference was attributed to the concrete itself (the uncracked compression zone, the interlock of aggregate across the crack, and the dowel action of the longitudinal bars), and for most of the twentieth century codes wrote the shear strength as the sum of a concrete term and a steel term, $V_n=V_c+V_s$. That form is still the one used by ACI CODE-318-25.

### Wilkins, 1955: When Axial Tension Removes $V_c$

In 1955 the roof beams of a warehouse at the Wilkins Air Force Depot in Shelby, Ohio, failed in shear. The beams had little or no shear reinforcement in the regions that failed, and restrained shrinkage and temperature changes had induced axial tension in them. The tension opened inclined cracks that the concrete term of the design had assumed would not form, and the beams failed at loads below those they were designed for. The failure, studied in the following years, led to the report of Joint ACI-ASCE Committee 326 on shear and diagonal tension (1962), which shaped the shear provisions of ACI 318 for the following decades. Its lesson is still written in the Code: 22.5.1.8 requires the effect of axial tension due to creep and shrinkage to be considered in calculating $V_c$, and Note 2 of Table 22.5.5.1 sets $V_c=0$ when tension is large enough.

### The Axial-Load Term

The same research showed the opposite effect of axial compression: it delays the opening of inclined cracks and increases the shear carried by the concrete. For several decades, until ACI 318-14, the Code expressed it as a multiplier on the concrete term, $V_c=2\left(1+N_u/2000A_g\right)\lambda\sqrt{f'_c}\,b_wd$ in psi units. ACI 318-19 rewrote the expression in the additive form used today, with the term $N_u/6A_g$ added to the concrete stress.

### San Fernando, 1971: Columns That Failed in Shear

The San Fernando earthquake of 1971 gave the profession its most vivid images of column shear failure. At the new Olive View Hospital, tied columns were shattered by shear, while spirally reinforced columns, although severely deformed, kept carrying load. Post-earthquake surveys, there and in every major earthquake since, have also shown the **short column** or **captive column** failure: a column whose free length is shortened by a partial-height masonry wall, a mezzanine, or a band of windows attracts a shear force far larger than its neighbors, because its end moments act over a shorter length. The captive column is solved by hand in the worked example of this notebook. The seismic design of columns that grew out of these observations (capacity design and confinement) belongs to Chapter 18 and to the seismic unit of the course.

### ACI 318-19 and ACI CODE-318-25

Large, lightly reinforced members tested at the end of the twentieth century failed at shear stresses lower than those of the small specimens on which the old expressions had been calibrated. ACI 318-19 (Kuchma et al. 2019) introduced the size effect factor $\lambda_s$ and the longitudinal reinforcement ratio $\rho_w$ into $V_c$, through the three expressions of Table 22.5.5.1, together with a cap on the axial-load term. The same edition already contains the linear interaction for biaxial shear of 22.5.1.11 (Umehara and Jirsa 1984). ACI CODE-318-25 added two items that matter for columns: 22.5.2.1 permits $d=0.8h$ for rectangular columns, a value that tests show to be accurate (Sezen et al. 2021); and 22.5.5.1.1 sets a lower bound on $V_c$, so that the compounded effect of $\lambda_s$ and $\rho_w$ cannot drive it toward zero in members that are not in net axial tension.

## Behavior: How a Column Resists Shear

### Diagonal Tension and the Truss

In a region where shear and flexure act together, the principal tensile stresses are inclined. When they exceed the tensile strength of concrete, an inclined crack forms, roughly at 45 degrees to the axis of the member. If no transverse reinforcement crosses the crack, nothing holds its two faces together, and the member fails by diagonal tension.

Ties crossing the crack restore equilibrium. Each tie that crosses it carries a tensile force $A_vf_{yt}$ at yield, where $A_v$ is the area of all the tie legs parallel to the shear. A crack at 45 degrees has a projection along the member approximately equal to $d$, so the number of ties that cross it is about $d/s$, with $s$ the tie spacing. The shear carried by the ties is then

$$V_s=\frac{A_vf_{yt}\,d}{s}$$

which is Eq. (22.5.8.5.3) of ACI CODE-318-25. The commentary (R22.5.8.5.1) describes it as a modified truss analogy: the ties carry the shear in excess of the shear that causes inclined cracking, and the concrete contributes, through the compression zone, aggregate interlock, and dowel action, an amount equivalent to that cracking shear, $V_c$.

### Why Axial Compression Helps, and Tension Hurts

A column differs from a beam because it carries axial force. Axial compression reduces the principal tensile stress at a given shear, delays inclined cracking, keeps the cracks narrow when they form, and enlarges the compression zone that carries shear across the crack. All of this raises $V_c$. Axial tension does the opposite: it lowers the cracking shear and opens the cracks, and if it is large enough it removes the concrete contribution entirely. ACI CODE-318-25 reflects both effects through the term $N_u/6A_g$ inside $V_c$, positive for compression and negative for tension.

The beneficial effect of compression has a practical warning attached. The axial load of a column is not constant: under lateral loads an exterior column of a frame gains compression on one side of the building and loses it, or goes into tension, on the other. The shear design must use the axial force that acts **simultaneously** with the shear in each load combination, and the combination with the least compression often governs.

### The Short Column

The shear in a column bent in double curvature is $V=(M_{top}+M_{bot})/\ell$. The moments at the ends are limited by the flexural strength of the column (or of the beams framing into it), so the shear grows as $\ell$ shrinks. A column with a partial-height wall on both sides has its free length reduced to the part above the wall, and the full end moments must change sign over that short length. A column that would be controlled by flexure over its full height can be controlled by shear when it is made captive.

### Brittle and Ductile Failure

The design objective, made explicit in the seismic chapter of the Code and implicit everywhere else, is that a column should never fail in shear before it reaches its flexural strength. A flexural failure gives warning; a shear failure does not. Shear reinforcement serves this purpose twice: it provides the strength $V_s$, and, by crossing every potential crack at close spacing, it confines the core and holds the longitudinal bars in place.

## Shear Strength According to ACI CODE-318-25

Shear in columns is governed by 10.5.3.1, which sends the calculation of $V_n$ to the one-way shear provisions of 22.5, the same section used for beams. What follows collects the provisions that apply to a nonprestressed rectangular column.

### Design Condition

At every section, for every load combination (10.5.1.1, 22.5.1.1):

$$\phi V_n\ge V_u,\qquad V_n=V_c+V_s$$

with $\phi=0.75$ for shear (Table 21.2.1(b)).

### Concrete Contribution, Table 22.5.5.1

The choice of expression depends on whether the column has at least the minimum shear reinforcement $A_{v,min}$ of 10.6.2.2:

| Criterion | $V_c$ (psi units, as in the Code) | |
|---|---|---|
| $A_v\ge A_{v,min}$, either of | $\left[2\lambda\sqrt{f'_c}+\dfrac{N_u}{6A_g}\right]b_wd$ | (a) |
| | $\left[8\lambda\left(\rho_w\right)^{1/3}\sqrt{f'_c}+\dfrac{N_u}{6A_g}\right]b_wd$ | (b) |
| $A_v<A_{v,min}$ | $\left[8\lambda_s\lambda\left(\rho_w\right)^{1/3}\sqrt{f'_c}+\dfrac{N_u}{6A_g}\right]b_wd$ | (c) |

$N_u$ is positive for compression and negative for tension, and $V_c$ shall not be taken less than zero (Notes 1 and 2). For $A_v\ge A_{v,min}$ either (a) or (b) may be used; the commentary describes (a) as the simpler option. $\rho_w=A_s/(b_wd)$ is the ratio of longitudinal tension reinforcement; in a column with bars on several faces, $A_s$ is the area of the bars on the tension face.

### Limits on $V_c$

- **Upper limit** (22.5.5.1.1): $V_c\le5\lambda\sqrt{f'_c}\,b_wd$.
- **Lower bound, new in ACI CODE-318-25** (22.5.5.1.1): $V_c$ need not be taken less than $\lambda\sqrt{f'_c}\,b_wd$, except for elements in net axial tension and the seismic cases of 18.6.5.2 and 18.7.6.2.1.
- **Cap on the axial term** (22.5.5.1.2): $N_u/6A_g\le0.05f'_c$.
- **Size effect factor** (22.5.5.1.3): $\lambda_s=\sqrt{2/(1+d/10)}\le1$, with $d$ in inches.
- **Concrete strength** (22.5.3.1): $\sqrt{f'_c}\le100$ psi in $V_c$, that is, $f'_c$ up to 10,000 psi (about 700 $\mathrm{kgf/cm^2}$).

### Effective Depth

22.5.2.1(a) permits $d=0.8h$ for rectangular columns. The actual depth to the tension steel may be used instead. For the example column, $d=43.90$ cm against $0.8h=40$ cm; the worked example uses the actual value and shows the difference.

### Shear Reinforcement

Where $V_u>\phi V_c$, ties satisfying $\phi V_s\ge V_u-\phi V_c$ are required (22.5.8.1), with $V_s=A_vf_{yt}d/s$ (22.5.8.5.3). $A_v$ is the area of all the tie legs parallel to the shear within the spacing $s$ (22.5.8.5.5), and $f_{yt}$ shall not exceed 60,000 psi in $V_s$ (22.5.3.3, 20.2.2.4).

### Limit on the Cross Section

To control diagonal compression and service cracking (22.5.1.2):

$$V_u\le\phi\left(V_c+8\sqrt{f'_c}\,b_wd\right)$$

If this is not satisfied, no amount of ties is enough: the section must be enlarged.

### Minimum Shear Reinforcement

A minimum area $A_{v,min}$ is required wherever $V_u>0.5\phi V_c$ (10.6.2.1), equal to the greater of (10.6.2.2):

$$A_{v,min}=0.75\sqrt{f'_c}\,\frac{b_ws}{f_{yt}}\qquad\text{and}\qquad A_{v,min}=50\,\frac{b_ws}{f_{yt}}$$

### Maximum Spacing of Shear Reinforcement, Table 10.7.6.5.2

| $V_s$ | Maximum $s$, nonprestressed column |
|---|---|
| $\le4\sqrt{f'_c}\,b_wd$ | lesser of $d/2$ and 24 in. |
| $>4\sqrt{f'_c}\,b_wd$ | lesser of $d/4$ and 12 in. |

These limits ensure that every potential inclined crack, whose projection is about $d$, is crossed by at least two ties (four when the shear is high).

### Biaxial Shear

Columns are often loaded in shear along both principal axes at once. The commentary (R22.5.1.10) cites tests showing that the shear strength follows an elliptical interaction, so that checking each direction separately can be unconservative. The Code adopts a linear interaction. It may be neglected if either ratio is small (22.5.1.10):

$$\frac{V_{u,x}}{\phi V_{n,x}}\le0.5\qquad\text{or}\qquad\frac{V_{u,y}}{\phi V_{n,y}}\le0.5$$

and otherwise (22.5.1.11):

$$\frac{V_{u,x}}{\phi V_{n,x}}+\frac{V_{u,y}}{\phi V_{n,y}}\le1.5$$

The printed condition of 22.5.1.11 in the Code writes $V_{n,y}$ in the numerator of the second ratio; from 22.5.1.10 and Eq. (22.5.1.11) the intended quantity is $V_{u,y}$.

Every coefficient that multiplies $\sqrt{f'_c}$ is calibrated to psi and must be converted for $f'_c$ in $\mathrm{kgf/cm^2}$. The values below are those published in ACI 318-19 Appendix C (mks-metric column), which tabulates the same equations. The lower bound of 22.5.5.1.1 does not exist in ACI 318-19; its coefficient is the Appendix C equivalence of $\lambda\sqrt{f'_c}$ in psi, $0.27\lambda\sqrt{f'_c}$ in $\mathrm{kgf/cm^2}$ (the same conversion used, for example, in 9.6.3.1).

| Provision | psi form | $\mathrm{kgf/cm^2}$ form |

| Table 22.5.5.1(a) | $2\lambda\sqrt{f'_c}$ | $0.53\lambda\sqrt{f'_c}$ |
| Table 22.5.5.1(b), (c) | $8\lambda(\rho_w)^{1/3}\sqrt{f'_c}$ | $2.1\lambda(\rho_w)^{1/3}\sqrt{f'_c}$ |
| 22.5.5.1.1, upper limit | $5\lambda\sqrt{f'_c}\,b_wd$ | $1.33\lambda\sqrt{f'_c}\,b_wd$ |
| 22.5.5.1.1, lower bound | $\lambda\sqrt{f'_c}\,b_wd$ | $0.27\lambda\sqrt{f'_c}\,b_wd$ |
| 22.5.5.1.3 | $\lambda_s=\sqrt{2/(1+d/10)}$, $d$ in in. | $\lambda_s=\sqrt{2/(1+0.04d)}$, $d$ in cm |

The plain lengths of the Code (24 in., 12 in., 6 in., 3 in., 1 in.) are taken in this course as 60, 30, 15, 7.5, and 2.5 cm. These are rounded conversions (the exact values are 61.0, 30.5, 15.2, 7.6, and 2.54 cm), not values published in Appendix C.

Two terms need no conversion. $N_u/6A_g$ is a stress in whatever units $N_u$ and $A_g$ are given, and the cap $0.05f'_c$ is proportional to $f'_c$. Note also that Appendix C publishes 2.2, not 2.1, for 22.5.1.2, although the psi coefficient (8) is the same as in (b) and (c): the appendix rounds each equation separately, and its published value is the one used here. A naive substitution shows why the conversion matters: writing $2\sqrt{280}=33.5\ \mathrm{kgf/cm^2}$ in expression (a) instead of $0.53\sqrt{280}=8.87\ \mathrm{kgf/cm^2}$ would overestimate the concrete stress term 3.8 times.

## Transverse Reinforcement and Detailing

The ties of a column do three jobs at once: they are its shear reinforcement, they keep the longitudinal bars from buckling outward once the cover spalls, and they confine the core. The detailing rules of 10.7.6 and 25.7 serve the second and third jobs, and the column must satisfy whichever of all the spacing limits is most restrictive (10.7.6.1.1).

### Ties, 25.7.2

- **Spacing** (25.7.2.1): clear spacing at least $(4/3)d_{agg}$, and center-to-center spacing not greater than the least of $16d_b$ of the longitudinal bar, $48d_b$ of the tie bar, and the smallest dimension of the member.
- **Tie diameter** (25.7.2.2): at least No. 3 enclosing No. 10 or smaller longitudinal bars, and No. 4 enclosing No. 11 or larger or bundled bars (inch-pound bar designations; No. 10 is about 32 mm). In metric practice, 10 mm ties for longitudinal bars up to 32 mm, and 12 mm ties for larger bars.
- **Lateral support** (25.7.2.3): (a) every corner bar and alternate longitudinal bar shall be laterally supported by the corner of a tie with an included angle of not more than 135 degrees, and (b) no unsupported bar shall be farther than 6 in. (15 cm) clear on each side along the tie from a laterally supported bar. The limit comes from tests of axially loaded tied columns (Pfister 1964).
- **Anchorage** (25.7.2.3.1): rectilinear ties are anchored by standard hooks conforming to 25.3.2 that engage a longitudinal bar.

### Location of the First and Last Tie, 10.7.6.2

In every story, the bottom tie shall be not more than half the tie spacing above the top of the footing or slab (10.7.6.2.1), and the top tie not more than half the tie spacing below the lowest horizontal reinforcement of the slab, drop panel, or shear cap; if beams frame into all four sides, not more than 3 in. (7.5 cm) below the lowest reinforcement of the shallowest beam (10.7.6.2.2).

### Spirals, 25.7.3

A spiral is a continuous helix of bar or wire. Its clear spacing shall be at least the greater of 1 in. (2.5 cm) and $(4/3)d_{agg}$, and not more than 3 in. (7.5 cm) (25.7.3.1); its diameter at least 3/8 in. (10 mm) in cast-in-place construction (25.7.3.2); and it is anchored with 1-1/2 extra turns at each end (25.7.3.4). The amount of spiral is fixed by the volumetric ratio of 25.7.3.3:

$$\rho_s\ge0.45\left(\frac{A_g}{A_{ch}}-1\right)\frac{f'_c}{f_{yt}}$$

where $A_{ch}$ is the area of the core measured to the outside of the spiral, and $f_{yt}$ is not taken greater than 100,000 psi (about 7000 $\mathrm{kgf/cm^2}$). The expression is dimensionless and needs no unit conversion. Its origin is Richart's tests of 1929 to 1933: the spiral is sized so that the strength gained by confining the core at least replaces the strength lost when the shell spalls. Spiral columns also receive a higher $\phi$ and a higher axial-load cap than tied columns (notebook 07), consistent with the greater toughness that this confinement gives them.

### Splices of Column Bars, 10.7.5

Longitudinal bars are usually spliced by lapping just above each floor. If the bar force under factored loads is compressive, the compression lap length of 25.5.5.1 applies: for $f_y\le60{,}000$ psi, the longer of $0.0005f_yd_b$ and 12 in., which in technical-metric units (ACI 318-19 Appendix C) is the longer of $0.0073f_yd_b$ and 30 cm. If the bar force is tensile, Table 10.7.5.2.2 requires a Class A or Class B tension lap splice. In the commentary the Code also recommends ties at each end of lap-spliced bars (R10.7.6.1.4).

## Worked Example: Section, Loads, and Two Cases

The section is the tied column of notebook 07:

| Property | Value |
|---|---|
| Section | $b=30$ cm, $h=50$ cm |
| Concrete | $f'_c=280\ \mathrm{kgf/cm^2}$, normalweight ($\lambda=1.0$) |
| Longitudinal bars | 3 bars of 22 mm on each 30 cm face, $A_s=11.40\ \mathrm{cm^2}$ per face, $A_g=1500\ \mathrm{cm^2}$ |
| Ties | 10 mm, $A_b=0.79\ \mathrm{cm^2}$, $f_{yt}=4200\ \mathrm{kgf/cm^2}$ |
| Depth to the tension steel | $d=43.90$ cm (shear along the 50 cm side) |

The perimeter tie has two legs parallel to each side, so for shear along the 50 cm side:

$$A_v=2\left(0.79\right)=1.58\ \mathrm{cm^2}$$

**Loads.** The column carries $P_u=N_u=120$ tf, the axial load of the design check of notebook 07, where the moment $M_u=20\ \mathrm{tf\cdot m}$ was found to be inside the design interaction diagram. Under a lateral load combination the column is bent in double curvature with $M_u=20\ \mathrm{tf\cdot m}$ at each end. Two cases are compared:

- **Case A, free column.** The column is free over its whole clear height, $\ell=3.0$ m.
- **Case B, captive column.** Masonry infill walls 1.8 m high are built against both sides of the same column. The free length is reduced to $\ell=1.2$ m.

In case B the same end moments are kept, which is a simplification on the safe side for the argument: the shorter free length also makes the column stiffer, and in the real frame it would attract larger moments, and therefore even more shear.

**Shear demand.** From $V=(M_{top}+M_{bot})/\ell$:

$$\text{Case A:}\qquad V_u=\frac{20+20}{3.0}=13.33\ \mathrm{tf}=13{,}333\ \mathrm{kgf}$$

$$\text{Case B:}\qquad V_u=\frac{20+20}{1.2}=33.33\ \mathrm{tf}=33{,}333\ \mathrm{kgf}$$

The section, the reinforcement, the axial load, and the moments are identical. Only the free length changed, and the shear grew 2.5 times.

**A ceiling on the demand.** These shears come from the analysis moments. The end moments cannot exceed the flexural strength of the column, so the shear is also bounded by it: with $\phi M_n=24.43\ \mathrm{tf\cdot m}$ at $P_u=120$ tf (notebook 07), the captive column could be asked for up to $2\left(24.43\right)/1.2=40.72$ tf if the lateral load grew. That figure uses the design strength $\phi M_n$; the moments the column can actually develop are larger, its nominal strength $M_n=24.43/0.65=37.58\ \mathrm{tf\cdot m}$ or, with strain hardening and real material strengths, its probable strength, so the true ceiling is higher still: $2\left(37.58\right)/1.2=62.6$ tf with $M_n$. For the ordinary frames of this notebook the Code designs for the analysis shear; for frames that resist earthquakes, Chapter 18 requires the shear to be derived from the probable flexural strength instead (capacity design), a topic of the seismic unit. The final design of case B is compared with this ceiling at the end of the biaxial check.

Both cases go through the same checks: concrete contribution, comparison of $V_u$ with $0.5\phi V_c$ and $\phi V_c$, limit on the cross section, required $V_s$ when there is one, spacing limits, minimum reinforcement, and final strength. Case A has no required $V_s$, so it needs fewer steps and its step numbers differ from those of case B. As in notebook 07, each step is written as the general equation, then the substitution, then the result with its units.

## Step 1: Concrete Contribution $V_c$

The concrete term is common to both cases, because it depends only on the section and the axial load.

**Square root of $f'_c$ and $b_wd$**

$$\sqrt{f'_c}=\sqrt{280}=16.73\ \mathrm{kgf/cm^2},\qquad b_wd=30\left(43.90\right)=1317\ \mathrm{cm^2}$$

**Axial term, with its cap (22.5.5.1.2)**

$$\frac{N_u}{6A_g}=\frac{120{,}000}{6\left(1500\right)}=13.33\ \mathrm{kgf/cm^2}\le0.05f'_c=0.05\left(280\right)=14.00\ \mathrm{kgf/cm^2}\ \checkmark$$

The axial term is close to its cap: this column already receives almost all the benefit of compression that the Code allows.

**Expression (a) of Table 22.5.5.1**, valid if $A_v\ge A_{v,min}$, which is verified in each case below:

$$0.53\lambda\sqrt{f'_c}=0.53\left(1.0\right)\left(16.73\right)=8.87\ \mathrm{kgf/cm^2}$$

$$V_c=\left[0.53\lambda\sqrt{f'_c}+\frac{N_u}{6A_g}\right]b_wd=\left(8.87+13.33\right)\left(1317\right)=22.20\left(1317\right)=29{,}237\ \mathrm{kgf}$$

**Expression (b)**, the alternative for $A_v\ge A_{v,min}$:

$$\rho_w=\frac{A_s}{b_wd}=\frac{11.40}{1317}=0.00866,\qquad \left(\rho_w\right)^{1/3}=0.2054$$

$$2.1\lambda\left(\rho_w\right)^{1/3}\sqrt{f'_c}=2.1\left(1.0\right)\left(0.2054\right)\left(16.73\right)=7.22\ \mathrm{kgf/cm^2}$$

$$V_c=\left(7.22+13.33\right)\left(1317\right)=20.55\left(1317\right)=27{,}064\ \mathrm{kgf}$$

Either may be used; expression (a) gives the larger value and is adopted.

**Expression (c)**, for reference: it would govern if the ties provided less than $A_{v,min}$.

$$\lambda_s=\sqrt{\frac{2}{1+0.04d}}=\sqrt{\frac{2}{1+0.04\left(43.90\right)}}=\sqrt{\frac{2}{2.756}}=0.8519\le1$$

$$2.1\lambda_s\lambda\left(\rho_w\right)^{1/3}\sqrt{f'_c}=2.1\left(0.8519\right)\left(1.0\right)\left(0.2054\right)\left(16.73\right)=6.15\ \mathrm{kgf/cm^2}$$

$$V_c=\left(6.15+13.33\right)\left(1317\right)=19.48\left(1317\right)=25{,}655\ \mathrm{kgf}$$

12 percent less than (a). Providing at least the minimum ties is rewarded with the simpler and larger expression.

**Limits of 22.5.5.1.1**

$$V_{c,max}=1.33\lambda\sqrt{f'_c}\,b_wd=1.33\left(1.0\right)\left(16.73\right)\left(1317\right)=29{,}304\ \mathrm{kgf}$$

$$V_{c,min}=0.27\lambda\sqrt{f'_c}\,b_wd=0.27\left(1.0\right)\left(16.73\right)\left(1317\right)=5949\ \mathrm{kgf}$$

$$5949\le29{,}237\le29{,}304\ \checkmark\qquad\Rightarrow\qquad V_c=29{,}237\ \mathrm{kgf}=29.24\ \mathrm{tf}$$

$V_c$ is at 99.8 percent of the upper limit, and the axial term at 95 percent of its own cap: for this section expression (a) reaches the upper limit at $N_u=120.4$ tf, a little before the axial term is capped at 126 tf (Figure 2).

**Design value and thresholds**

$$\phi V_c=0.75\left(29{,}237\right)=21{,}928\ \mathrm{kgf}=21.93\ \mathrm{tf},\qquad 0.5\phi V_c=10{,}964\ \mathrm{kgf}=10.96\ \mathrm{tf}$$

**Two comparisons.** With no axial load the concrete term would be $V_c=8.87\left(1317\right)=11{,}682$ kgf, so the axial compression multiplies $V_c$ by 2.5. With the simplified depth of 22.5.2.1(a), $d=0.8h=40$ cm, expression (a) would give $V_c=22.20\left(30\right)\left(40\right)=26{,}640$ kgf, 8.9 percent less than with the actual $d$. The simplification is permitted and conservative here; the actual $d$ is used in the rest of the example.

**Quantities used in both cases**

$$A_vf_{yt}d=1.58\left(4200\right)\left(43.90\right)=291{,}320\ \mathrm{kgf\cdot cm}$$

$$1.1\sqrt{f'_c}\,b_wd=1.1\left(16.73\right)\left(1317\right)=24{,}237\ \mathrm{kgf}\qquad\text{(Table 10.7.6.5.2)}$$

$$2.2\sqrt{f'_c}\,b_wd=2.2\left(16.73\right)\left(1317\right)=48{,}474\ \mathrm{kgf}\qquad\text{(22.5.1.2)}$$

## Case A: The Free Column

**Step 2: Demand against the thresholds**

$$0.5\phi V_c=10{,}964<V_u=13{,}333\le\phi V_c=21{,}928\ \mathrm{kgf}$$

The concrete alone is enough for strength, so no calculated $V_s$ is needed (22.5.8.1). But $V_u>0.5\phi V_c$, so the minimum shear reinforcement of 10.6.2.1 is required. Because the ties are now shear reinforcement, their spacing must also satisfy Table 10.7.6.5.2 (10.7.6.5.1 and 10.7.6.5.2), in addition to the tie rules of 25.7.2.

The limit on the cross section of 22.5.1.2 does not depend on the ties and is satisfied with a wide margin: $0.75\left(29{,}237+48{,}474\right)=58{,}283\ \mathrm{kgf}\ge13{,}333$ kgf. $\checkmark$

**Step 3: Spacing limits**

Ties, 25.7.2.1(b):

$$s\le\min\left(16d_b,\ 48d_{b,tie},\ b\right)=\min\left(16\left(2.2\right),\ 48\left(1.0\right),\ 30\right)=\min\left(35.2,\ 48,\ 30\right)=30\ \mathrm{cm}$$

Shear reinforcement, Table 10.7.6.5.2: the provided $V_s$ will be below $1.1\sqrt{f'_c}\,b_wd=24{,}237$ kgf (checked in step 5), so

$$s\le\min\left(\frac{d}{2},\ 60\right)=\min\left(\frac{43.90}{2},\ 60\right)=21.95\ \mathrm{cm}$$

The shear limit governs over the tie rule. **Ties of 10 mm at $s=20$ cm** are chosen.

**Step 4: Minimum shear reinforcement (10.6.2.2)**

$$A_{v,min}=0.2\sqrt{f'_c}\,\frac{b_ws}{f_{yt}}=0.2\left(16.73\right)\frac{30\left(20\right)}{4200}=0.478\ \mathrm{cm^2}$$

$$A_{v,min}=3.5\,\frac{b_ws}{f_{yt}}=3.5\,\frac{30\left(20\right)}{4200}=0.500\ \mathrm{cm^2}$$

$$A_v=1.58\ \mathrm{cm^2}\ge A_{v,min}=0.500\ \mathrm{cm^2}\ \checkmark$$

This also confirms that expression (a) of Table 22.5.5.1 was applicable.

**Step 5: Design strength**

$$V_s=\frac{A_vf_{yt}d}{s}=\frac{291{,}320}{20}=14{,}566\ \mathrm{kgf}\le24{,}237\ \mathrm{kgf}\ \checkmark$$

$$\phi V_n=\phi\left(V_c+V_s\right)=0.75\left(29{,}237+14{,}566\right)=0.75\left(43{,}803\right)=32{,}852\ \mathrm{kgf}=32.85\ \mathrm{tf}$$

$$\frac{V_u}{\phi V_n}=\frac{13{,}333}{32{,}852}=0.41\ \checkmark$$

The free column is far from critical in shear. Its ties are governed by the maximum spacing for shear, not by strength.

## Case B: The Captive Column

**Step 2: Demand against the thresholds**

$$V_u=33{,}333\ \mathrm{kgf}>\phi V_c=21{,}928\ \mathrm{kgf}$$

Shear reinforcement is required for strength.

**Step 3: Limit on the cross section (22.5.1.2)**

$$\phi\left(V_c+2.2\sqrt{f'_c}\,b_wd\right)=0.75\left(29{,}237+48{,}474\right)=0.75\left(77{,}711\right)=58{,}283\ \mathrm{kgf}\ge V_u=33{,}333\ \mathrm{kgf}\ \checkmark$$

The section is large enough; the problem can be solved with ties.

**Step 4: Required $V_s$ and spacing**

$$V_{s,req}=\frac{V_u}{\phi}-V_c=\frac{33{,}333}{0.75}-29{,}237=44{,}444-29{,}237=15{,}207\ \mathrm{kgf}$$

$$s_{req}=\frac{A_vf_{yt}d}{V_{s,req}}=\frac{291{,}320}{15{,}207}=19.16\ \mathrm{cm}$$

**Step 5: Spacing limits.** $V_{s,req}=15{,}207\le1.1\sqrt{f'_c}\,b_wd=24{,}237$ kgf, so Table 10.7.6.5.2 gives $s\le d/2=21.95$ cm, and 25.7.2.1 gives $s\le30$ cm. Strength governs.

Ties at 20 cm, enough for case A, are not enough here: they give $\phi V_n=32{,}852<33{,}333$ kgf, 1.4 percent short. **Ties of 10 mm at $s=15$ cm** are chosen, a practical spacing below $s_{req}$. A spacing of 17.5 cm would also satisfy this direction; 15 cm is taken as a round trial value, and the biaxial check below will reduce it anyway.

**Step 6: Minimum shear reinforcement**

$$A_{v,min}=\max\left[0.2\left(16.73\right)\frac{30\left(15\right)}{4200},\ 3.5\,\frac{30\left(15\right)}{4200}\right]=\max\left(0.358,\ 0.375\right)=0.375\ \mathrm{cm^2}\le1.58\ \mathrm{cm^2}\ \checkmark$$

**Step 7: Design strength**

$$V_s=\frac{291{,}320}{15}=19{,}421\ \mathrm{kgf}$$

$$\phi V_n=0.75\left(29{,}237+19{,}421\right)=0.75\left(48{,}658\right)=36{,}494\ \mathrm{kgf}=36.49\ \mathrm{tf}$$

$$\frac{V_u}{\phi V_n}=\frac{33{,}333}{36{,}494}=0.913\ \checkmark$$

This spacing is revised in the next section, when the shear in the other direction is added.

### What If the Axial Load Disappears?

Suppose the same captive column belonged to an exterior frame line where, in the governing combination, the overturning effect of the lateral load cancels the gravity compression, $N_u\approx0$. Then

$$V_c=0.53\left(16.73\right)\left(1317\right)=8.87\left(1317\right)=11{,}682\ \mathrm{kgf}$$

$$V_{s,req}=44{,}444-11{,}682=32{,}762\ \mathrm{kgf}>1.1\sqrt{f'_c}\,b_wd=24{,}237\ \mathrm{kgf}$$

The higher row of Table 10.7.6.5.2 now applies, $s\le\min\left(d/4,\ 30\right)=\min\left(10.98,\ 30\right)=10.98$ cm, and strength requires

$$s_{req}=\frac{291{,}320}{32{,}762}=8.89\ \mathrm{cm}\quad\Rightarrow\quad s=8.5\ \mathrm{cm}$$

The spacing is rounded down, never up, to a practical value: 8.5 cm. The ties must almost halve their spacing. The section still passes 22.5.1.2: $0.75\left(11{,}682+48{,}474\right)=45{,}117\ \mathrm{kgf}\ge33{,}333$ kgf. The lesson is the one stated in the behavior section: $V_c$ must be computed with the axial force that acts simultaneously with the shear, and the combination with the least compression may govern.

In net tension the concrete term vanishes faster still. With $N_u=-30$ tf the axial term is $-30{,}000/\left[6\left(1500\right)\right]=-3.33\ \mathrm{kgf/cm^2}$, and the lower bound of 22.5.5.1.1 does not apply, because the member is in net axial tension:

$$V_c=\left(8.87-3.33\right)\left(1317\right)=7296\ \mathrm{kgf}$$

$$V_{s,req}=44{,}444-7296=37{,}148\ \mathrm{kgf}>24{,}237\ \mathrm{kgf}\quad\Rightarrow\quad s\le d/4=10.98\ \mathrm{cm}$$

$$s_{req}=\frac{291{,}320}{37{,}148}=7.84\ \mathrm{cm}\quad\Rightarrow\quad s=7.5\ \mathrm{cm}$$

and the section still passes 22.5.1.2: $0.75\left(7296+48{,}474\right)=41{,}828\ \mathrm{kgf}\ge33{,}333$ kgf.

Both variations check the 50 cm direction only. If the shear along the 30 cm side of the next section also acted, the loss of axial compression would lower $V_{c,y}$ as well, and the biaxial check would have to be repeated with these values of $N_u$.

## Biaxial Shear Check

The orthogonal frame also loads the captive column: in the same combination it carries $V_{u,y}=18.00$ tf along the 30 cm side, while $V_{u,x}=33.33$ tf acts along the 50 cm side as computed above.

$V_{u,y}$ is an assumed demand, chosen to show how the interaction works; it is not derived here. Over the free length of 1.2 m it implies end moments of about $18\left(1.2\right)/2=10.8\ \mathrm{tf\cdot m}$ about the weak axis, acting together with $M_{u,x}=20\ \mathrm{tf\cdot m}$ and $P_u=120$ tf. That combination of moments is a biaxial bending problem, solved in example 4 of notebook 09: it exceeds the flexural strength of this section ($\phi P_n=83.8$ tf against $P_u=120$ tf), and a 40 x 50 cm section with the same bars would be needed. This notebook checks shear only.

**Step 1: Strength along the 30 cm side.** The section now works with $b_w=50$ cm and the depth to the steel layer farthest from the compression face, $d=30-6.10=23.90$ cm. The perimeter tie also has two legs parallel to the 30 cm side, $A_v=1.58\ \mathrm{cm^2}$.

$$b_wd=50\left(23.90\right)=1195\ \mathrm{cm^2}$$

$$V_{c,y}=\left(8.87+13.33\right)\left(1195\right)=22.20\left(1195\right)=26{,}529\ \mathrm{kgf}\le1.33\left(16.73\right)\left(1195\right)=26{,}590\ \mathrm{kgf}\ \checkmark$$

$$A_vf_{yt}d=1.58\left(4200\right)\left(23.90\right)=158{,}600\ \mathrm{kgf\cdot cm}$$

**Step 2: Check with ties at 15 cm**

$$V_{s,y}=\frac{158{,}600}{15}=10{,}573\ \mathrm{kgf},\qquad \phi V_{n,y}=0.75\left(26{,}529+10{,}573\right)=27{,}826\ \mathrm{kgf}$$

$$\frac{V_{u,x}}{\phi V_{n,x}}=\frac{33{,}333}{36{,}494}=0.913,\qquad \frac{V_{u,y}}{\phi V_{n,y}}=\frac{18{,}000}{27{,}826}=0.647$$

Both ratios exceed 0.5, so the interaction cannot be neglected (22.5.1.10), and Eq. (22.5.1.11) gives

$$0.913+0.647=1.560>1.5\ \text{(not satisfied)}$$

Each direction checked alone would pass, with ratios below 1.0. That is exactly the unconservative case the commentary warns about.

**Step 3: Check with ties at 12.5 cm**

$$V_{s,x}=\frac{291{,}320}{12.5}=23{,}306\ \mathrm{kgf},\qquad \phi V_{n,x}=0.75\left(29{,}237+23{,}306\right)=39{,}407\ \mathrm{kgf}$$

$$V_{s,y}=\frac{158{,}600}{12.5}=12{,}688\ \mathrm{kgf},\qquad \phi V_{n,y}=0.75\left(26{,}529+12{,}688\right)=29{,}413\ \mathrm{kgf}$$

$$\frac{33{,}333}{39{,}407}+\frac{18{,}000}{29{,}413}=0.846+0.612=1.458\le1.5\ \checkmark$$

The interaction is satisfied, but strength is not the only requirement. The spacing limits of Table 10.7.6.5.2 apply in **each** direction, and the effective depth along the 30 cm side is much smaller:

$$V_{s,y}=12{,}688\le1.1\sqrt{f'_c}\,b_wd=1.1\left(16.73\right)\left(1195\right)=21{,}992\ \mathrm{kgf}\quad\Rightarrow\quad s\le\frac{d}{2}=\frac{23.90}{2}=11.95\ \mathrm{cm}$$

$$s=12.5\ \mathrm{cm}>11.95\ \mathrm{cm}\ \text{(not satisfied)}$$

Ties at 12.5 cm fail the spacing limit along the 30 cm side, even though they pass the strength interaction. The lesson: the spacing limits are checked in both directions, with the $d$ of each direction.

**Step 4: Check with ties at 10 cm**

$$V_{s,x}=\frac{291{,}320}{10}=29{,}132\ \mathrm{kgf},\qquad \phi V_{n,x}=0.75\left(29{,}237+29{,}132\right)=43{,}777\ \mathrm{kgf}$$

$$V_{s,y}=\frac{158{,}600}{10}=15{,}860\ \mathrm{kgf},\qquad \phi V_{n,y}=0.75\left(26{,}529+15{,}860\right)=31{,}792\ \mathrm{kgf}$$

$$\frac{33{,}333}{43{,}777}+\frac{18{,}000}{31{,}792}=0.761+0.566=1.328\le1.5\ \checkmark$$

The remaining checks, direction by direction:

- **Spacing along the 30 cm side** (Table 10.7.6.5.2): $V_{s,y}=15{,}860\le21{,}992$ kgf, so $s\le d/2=11.95$ cm, and $s=10$ cm. $\checkmark$
- **Spacing along the 50 cm side**: the required $V_{s,x}=15{,}207$ kgf is below 24,237 kgf, so $s\le d/2=21.95$ cm. This course compares the threshold with the required $V_s$. A more conservative practice compares it with the provided $V_s$; here $V_{s,x}=29{,}132>24{,}237$ kgf would then call for $s\le d/4=10.98$ cm, which is also met. $\checkmark$
- **Minimum reinforcement** (10.6.2.2), with $b_w=30$ cm for $x$ and $b_w=50$ cm for $y$:

$$A_{v,min,x}=\max\left[0.2\left(16.73\right)\frac{30\left(10\right)}{4200},\ 3.5\,\frac{30\left(10\right)}{4200}\right]=\max\left(0.239,\ 0.250\right)=0.250\ \mathrm{cm^2}\le1.58\ \mathrm{cm^2}\ \checkmark$$

$$A_{v,min,y}=\max\left[0.2\left(16.73\right)\frac{50\left(10\right)}{4200},\ 3.5\,\frac{50\left(10\right)}{4200}\right]=\max\left(0.398,\ 0.417\right)=0.417\ \mathrm{cm^2}\le1.58\ \mathrm{cm^2}\ \checkmark$$

- **Limit on the cross section along the 30 cm side** (22.5.1.2): $0.75\left(26{,}529+2.2\left(16.73\right)\left(1195\right)\right)=0.75\left(26{,}529+43{,}983\right)=52{,}884\ \mathrm{kgf}\ge18{,}000$ kgf. $\checkmark$

**Final design of case B: ties of 10 mm at 10 cm.** Compared with the ceiling on the demand found earlier, the strength along the 50 cm side, $\phi V_{n,x}=43.78$ tf, is also above the 40.72 tf that the column could be asked for if its ends reached $\phi M_n$, though not above the 62.6 tf that corresponds to $M_n$. Designing the shear for the flexural strength is what Chapter 18 does for seismic frames.

## Detailing the Example

**Tie diameter (25.7.2.2).** The longitudinal bars are 22 mm, smaller than No. 10 (32 mm), so a 10 mm tie is sufficient. $\checkmark$

**Lateral support (25.7.2.3).** The four corner bars sit in the corners of the tie, whose included angle is 90 degrees, not more than 135 degrees. The tie is anchored with standard tie hooks engaging a longitudinal bar (25.7.2.3.1, 25.3.2). The middle bar of each 30 cm face is not at a tie corner. It is an alternate bar, so 25.7.2.3(a) does not require it to be supported, and its clear distance to the corner bars along the tie is

$$\frac{30-2\left(6.10\right)}{2}-2.2=8.90-2.2=6.70\ \mathrm{cm}\le15\ \mathrm{cm}\ \checkmark\quad\text{(25.7.2.3(b))}$$

so no crosstie is needed. This is the upper left arrangement of Fig. R25.7.2.3a.

**Clear spacing of ties (25.7.2.1(a)).** With 19 mm (3/4 in.) maximum aggregate, $(4/3)d_{agg}=(4/3)\left(1.9\right)=2.53$ cm; the clear spacing of the ties of case B is $10-1.0=9.0$ cm. $\checkmark$

**First and last tie (10.7.6.2).** In case A the bottom tie goes not more than $s/2=10$ cm above the slab; in case B not more than $s/2=5$ cm. At the top, not more than $s/2$ below the lowest slab reinforcement, or not more than 7.5 cm below the lowest bars of the shallowest beam if beams frame into all four sides.

**Extent of the close spacing in case B.** The shear of 33.33 tf acts over the free length, but the ties at 10 cm are carried over the full height of the column: the infill walls may crack or be modified during the life of the building, and the free length is not a reliable constant. This is a design decision, not a requirement of the provisions used here; Chapter 18 makes the equivalent requirement explicit for columns of special moment frames.

**Splices.** If the 22 mm bars are lap spliced in compression above the slab, the lap length is the longer of

$$0.0073f_yd_b=0.0073\left(4200\right)\left(2.2\right)=67.5\ \mathrm{cm}\qquad\text{and}\qquad30\ \mathrm{cm}$$

so 70 cm in practice. If any combination puts these bars in tension, the tension splice class of Table 10.7.5.2.2 governs instead.

### A Spiral Column, Briefly

For comparison, consider a circular spiral column of diameter 45 cm with 4 cm of cover to the spiral, $f'_c=280$, and a 10 mm spiral with $f_{yt}=4200\ \mathrm{kgf/cm^2}$. The core diameter to the outside of the spiral is $D_{ch}=45-2\left(4\right)=37$ cm.

$$A_g=\frac{\pi\left(45\right)^2}{4}=1590.4\ \mathrm{cm^2},\qquad A_{ch}=\frac{\pi\left(37\right)^2}{4}=1075.2\ \mathrm{cm^2}$$

$$\rho_s\ge0.45\left(\frac{1590.4}{1075.2}-1\right)\frac{280}{4200}=0.45\left(0.4792\right)\left(0.06667\right)=0.01438$$

For a spiral of bar area $A_{sp}$, bar diameter $d_b$, and pitch $s$, the volume of one turn divided by the volume of core it encloses gives

$$\rho_s=\frac{A_{sp}\,\pi\left(D_{ch}-d_b\right)}{\left(\pi D_{ch}^2/4\right)s}=\frac{4A_{sp}\left(D_{ch}-d_b\right)}{D_{ch}^2\,s}$$

so

$$s\le\frac{4A_{sp}\left(D_{ch}-d_b\right)}{D_{ch}^2\,\rho_s}=\frac{4\left(0.79\right)\left(37-1.0\right)}{37^2\left(0.01438\right)}=\frac{113.8}{19.69}=5.78\ \mathrm{cm}$$

The common approximation $\rho_s\approx4A_{sp}/(D_{ch}s)$ takes the spiral at the outer diameter of the core and would give 5.94 cm, slightly unconservative. A pitch of 5.5 cm gives $\rho_s=113.8/\left[37^2\left(5.5\right)\right]=0.01511\ge0.01438$, and a clear spacing of $5.5-1.0=4.5$ cm, between 2.5 and 7.5 cm (25.7.3.1). $\checkmark$ The spiral is far denser than any tie spacing computed for shear in this notebook: its job is to confine the core, and its contribution to shear follows from Eq. (22.5.8.5.3) with $A_v$ equal to twice the bar area (22.5.8.5.6).

## Summary

A column bent by end moments carries a shear $V=(M_{top}+M_{bot})/\ell$, small under gravity loads and potentially large under lateral loads. Shear failure is brittle, and the history of the provisions is a history of failures: the Wilkins warehouse (1955) showed that axial tension can remove the concrete contribution, and the San Fernando earthquake (1971) showed tied and captive columns failing in shear. The truss analogy of Ritter and M&ouml;rsch, with a concrete term added, is still the model: $V_n=V_c+V_s$, $\phi=0.75$.

ACI CODE-318-25 computes $V_c$ from Table 22.5.5.1, where axial compression enters as $N_u/6A_g$, capped at $0.05f'_c$; $V_c$ is limited to $5\lambda\sqrt{f'_c}\,b_wd$ and, new in this edition, need not be taken below $\lambda\sqrt{f'_c}\,b_wd$ unless the member is in net tension. Also new, $d=0.8h$ is permitted for rectangular columns. The ties provide $V_s=A_vf_{yt}d/s$, with $A_{v,min}$ wherever $V_u>0.5\phi V_c$, spacing not more than $d/2$ (or $d/4$ for high shear), and a linear interaction when the column is sheared along both axes. In $\mathrm{kgf/cm^2}$ the coefficients are those of ACI 318-19 Appendix C: 0.53 and 2.1 in $V_c$, 1.33 for the upper limit, 0.27 for the lower bound, $\lambda_s=\sqrt{2/(1+0.04d)}$, 0.2 and 3.5 in $A_{v,min}$, 1.1 for the spacing threshold, and 2.2 for the section limit.

For the 30 x 50 cm column of notebook 07 under $P_u=120$ tf, $V_c=29.24$ tf, 2.5 times its value without axial load and at 99.8 percent of the upper limit. As a free column 3.0 m high ($V_u=13.33$ tf) it needs only minimum ties, and the spacing $d/2$ gives 10 mm ties at 20 cm. Made captive by infill walls, with 1.2 m of free length, the same column carries $V_u=33.33$ tf: ties at 20 cm fall 1.4 percent short, and 15 cm suffices in one direction. With $V_{u,y}=18$ tf along the other axis, 12.5 cm satisfies the biaxial interaction but not the spacing limit $d/2=11.95$ cm along the 30 cm side, and the final design is 10 mm ties at 10 cm. Without the axial compression, the same captive column would need ties at 8.5 cm, and in net tension of 30 tf at 7.5 cm. The section needs no crosstie (6.7 cm clear between bars, against 15 cm), and a spiral column of the same size would need a 10 mm spiral at 5.5 cm pitch to satisfy $\rho_s\ge0.45(A_g/A_{ch}-1)f'_c/f_{yt}$.

## From Slab Behavior to Slab Design

The previous notebook established why a slab panel behaves as a one-way or a two-way member, how ACI CODE-318-25 classifies the resulting construction forms, and, for two-way action, how curvature at a point resolves into $m_x$, $m_y$, and a twisting moment $m_{xy}$ governed by the plate equation $\partial^2m_x/\partial x^2+2\partial^2m_{xy}/\partial x\partial y+\partial^2m_y/\partial y^2=-q$. That equation is not solved directly in ordinary practice. What ACI CODE-318-25 offers instead is 8.2.1: a slab system may be designed by any procedure satisfying equilibrium and geometric compatibility, provided design strength at every section is at least equal to required strength and all serviceability requirements are satisfied. The direct design method and the equivalent frame method are the two standardized paths the Code names explicitly for that purpose; this notebook develops the direct design method (DDM) for two-way slabs together with the reinforcement requirements, common to both methods, that finish a slab design once the design moments are known.

### A necessary disclosure about where the Direct Design Method actually lives

Section 6.2.4.1 of ACI CODE-318-25 states that two-way slabs are permitted to be analyzed for gravity loads using (a) the direct design method for nonprestressed slabs or (b) the equivalent frame method for nonprestressed and prestressed slabs. Its commentary, R6.2.4.1, is unusually direct about what that permission actually means today: *"Code editions from 1971 to 2014 contained provisions for use of the direct design method and the equivalent frame method... These provisions for gravity load analysis of two-way slabs have been removed from the Code because they are considered to be only two of several analysis methods currently used for the design of two-way slabs. The direct design method and the equivalent frame method of the 2014 Code, however, may still be used for the analysis of two-way slabs for gravity loads."* In plain terms: ACI CODE-318-25 authorizes the DDM by reference, but the numerical machinery of the method itself, the static-moment formula, the longitudinal distribution percentages, the column-strip and middle-strip apportionment tables, is no longer printed inside the current Code text. It survives, essentially unchanged since its introduction in 1971, in ACI 318-14 (and identically in the editions immediately before it).

This notebook is explicit about that boundary everywhere it applies. Provisions that remain fully normative in ACI CODE-318-25 itself, minimum slab thickness (already covered in the previous notebook), reinforcement limits, and detailing, are cited directly against the current Code numbering (Chapter 8 and Chapter 24). The DDM mechanics developed below are cited against ACI 318-14, with the 318-25 permission noted once here rather than repeated at every equation. A designer using this material in practice should treat the DDM the same way this course does: as a Code-sanctioned, historically stable procedure that happens to be documented one edition back, not as an obsolete method.

## Direct Design Method: Scope and Limitations

The DDM converts a two-way slab panel into a small number of design moments through a fixed sequence of coefficients, without ever solving the governing plate equation. That shortcut is only safe within the geometric and loading envelope the coefficients were calibrated for. ACI 318-14 8.10.2 states that envelope as six conditions, all of which must hold simultaneously for the method to apply:

1. **There must be a minimum of three continuous spans in each direction.** The moment coefficients already build in an assumption about how much rotational restraint an interior support provides; that assumption only converges to something realistic once there are enough interior spans for a genuinely interior condition to exist. A two-span system has no truly interior span at all, only two end spans meeting at a shared support, so the interior-span coefficients would not describe anything physically present.
2. **Panels must be rectangular, with the ratio of the longer to the shorter span (center-to-center of supports) not exceeding 2.** This is the same aspect-ratio boundary developed by compatibility in the previous notebook: beyond that ratio, two-way redistribution is negligible and the panel should be treated as one-way instead, not forced through two-way coefficients that no longer describe its behavior.
3. **Successive span lengths, center-to-center of supports, in each direction must not differ by more than one-third of the longer span.** Sharply unequal adjacent spans redistribute moment in ways the fixed 65/35 and end-span percentages were never calibrated against.
4. **Columns may be offset from either axis between centerlines of successive columns by no more than 10 percent of the span (in the direction of the offset).** A larger offset turns the regular orthogonal frame the method assumes into something closer to an irregular frame, which requires the equivalent frame method or a general analysis instead.
5. **All loads must be due to gravity only, uniformly distributed over an entire panel, and the unfactored live load must not exceed twice the unfactored dead load.** The DDM has no mechanism for lateral loads, concentrated loads, or the kind of live-to-dead ratio that would call for a full pattern-loading study; those cases fall outside its calibration.
6. **For a panel with beams between supports on all sides, the relative stiffness of the beams in the two perpendicular directions must satisfy** $0.2\leq\dfrac{\alpha_{f1}\ell_2^2}{\alpha_{f2}\ell_1^2}\leq5.0$, **where** $\alpha_f=E_{cb}I_b/E_{cs}I_s$ **is the ratio of beam to slab flexural stiffness in the direction of the moment.** A panel whose beams are far stiffer in one direction than the other stops behaving like the regular two-way system the coefficients describe and starts behaving like a one-way system carried on the stiff beam line; the stiffness ratio window keeps the panel inside genuinely two-way territory.

A slab system that fails any one of these six conditions is not forbidden, it is simply outside the DDM's calibrated envelope, and ACI 318-14 (and 8.2.1 of the current Code) directs the designer to the equivalent frame method or to a general elastic, numerical, or yield-line analysis instead.

## Total Factored Static Moment

Within its envelope, the DDM starts every span the same way a simply supported beam analysis starts: with the total static moment the span must carry, computed independently of how that moment will later be split between negative and positive regions. For a uniformly loaded, simply supported span of length $\ell_n$ carrying a load $q_u$ per unit width over a design strip of width $\ell_2$, statics alone gives a total moment of $q_u\ell_2\ell_n^2/8$, exactly the familiar $wL^2/8$ result already used throughout this course, generalized from a line load on a beam to a distributed load on a strip of finite width. ACI 318-14 8.10.3.2 adopts this quantity directly as the total factored static moment for the span:

$$M_o=\frac{q_u\ell_2\ell_n^2}{8}$$

Here $\ell_2$ is the transverse width of the design strip (the distance between the panel centerlines on either side of the span under consideration), and $\ell_n$ is the clear span in the direction moments are being determined, measured face-to-face of supports (or face-to-face of the effective support area of 8.4.1.4 when capitals or brackets are present), subject to the condition $\ell_n\geq0.65\ell_1$. That lower bound exists because a very wide column or capital, taken literally, would shrink the clear span and therefore the computed $M_o$ well below what the frame actually experiences; capping how much support width is allowed to reduce $\ell_n$ keeps $M_o$ from being understated for stocky columns.

Physically, $M_o$ is not itself a design moment. No section of the slab is reinforced for $M_o$ directly. It is the same kind of intermediate quantity as the total static moment on a fixed-end beam: a bookkeeping total that guarantees, once it is split into negative and positive regions in the next step, that the sum of those regions is statically consistent with the load actually applied to the span. Because $q_u$, $\ell_2$, and $\ell_n$ are all expressed in whatever consistent length and force units the rest of the design uses, $M_o$ requires no unit-calibrated constant of the kind flagged elsewhere in this course: computed entirely in kgf and cm, it returns a moment in $\mathrm{kgf\cdot cm}$ without any conversion factor, exactly as it would in lb and in.

## Longitudinal Distribution of the Static Moment

$M_o$ is next split, along the length of the span, between one positive region at midspan and either one or two negative regions at the supports. ACI 318-14 8.10.4.1 fixes that split for an **interior span** without reference to any table:

$$M_u^-=0.65M_o\quad\text{(at each face of support)},\qquad M_u^+=0.35M_o\quad\text{(at midspan)}$$

An idealized elastic beam fixed rigidly at both ends and uniformly loaded splits its own static moment into $wL^2/12$ negative and $wL^2/24$ positive, a 2:1 ratio that sums, as it must, to $wL^2/8$. The DDM's 0.65/0.35 split is close to that idealized ratio but not identical to it, because a real interior slab-column connection is not a perfect fixed end: the column offers substantial but not infinite rotational restraint, so the negative moment concentrating at the support is slightly smaller, and the positive moment at midspan slightly larger, than pure fixity would predict.

It is worth being precise about how "0.65" and "0.35" recover the original $M_o$, because reading them as three independent fractions of the same total is a natural but incorrect way to check the arithmetic: $0.65+0.65+0.35=1.65$, not $1.00$. The identity these coefficients actually satisfy is the same one used throughout beam analysis for a span with two different end moments, the positive moment plus the *average* of the two end negative moments equals the total static moment, $M_u^++\left(M_{u,\text{left}}^-+M_{u,\text{right}}^-\right)/2=M_o$, not their unweighted sum. For the interior span, where both end moments are equal, that check gives $0.35M_o+\left(0.65M_o+0.65M_o\right)/2=0.35M_o+0.65M_o=1.00M_o$, exactly recovering $M_o$. The fixed-end beam comparison above is consistent with the same identity for the same reason: $wL^2/12$ already stands for one representative end moment, not both ends added together, so $wL^2/12+wL^2/24=wL^2/8$ needs no further averaging to check out.

An **end span** has no such symmetry: the exterior support is not a mirror image of the interior support one span over, so ACI 318-14 8.10.4.2 apportions the same $M_o$ (now computed with the end span's own $\ell_n$) among three critical sections, exterior negative, positive, and interior negative, according to how much rotational restraint the exterior edge itself provides:

| End-span condition | Exterior negative | Positive | Interior negative |
|---|---|---|---|
| (1) Exterior edge unrestrained | $0$ | $0.63M_o$ | $0.75M_o$ |
| (2) Slab with beams between all supports | $0.16M_o$ | $0.57M_o$ | $0.70M_o$ |
| (3) Slab without beams between interior supports, without an edge beam | $0.26M_o$ | $0.52M_o$ | $0.70M_o$ |
| (4) Slab without beams between interior supports, with an edge beam | $0.30M_o$ | $0.50M_o$ | $0.70M_o$ |
| (5) Exterior edge fully restrained | $0.65M_o$ | $0.35M_o$ | $0.65M_o$ |

The same identity used above to check the interior span, positive moment plus the average of the two end negative moments equals $M_o$, applies row by row here as well, now averaging the exterior and interior negative entries instead of two equal values: $(1)$ $0.63+(0+0.75)/2=1.005$; $(2)$ $0.57+(0.16+0.70)/2=1.00$; $(3)$ $0.52+(0.26+0.70)/2=1.00$; $(4)$ $0.50+(0.30+0.70)/2=1.00$; $(5)$ $0.35+(0.65+0.65)/2=1.00$ (all in units of $M_o$). Row (1) misses exact unity by 0.005 only because ACI 318-14 publishes these coefficients rounded to two decimal places, not because the underlying elastic analysis behind the table is inconsistent.

Reading the table as a single physical progression, rather than as five unrelated rows, makes it easier to remember than to look up. Condition (1), a slab edge resting on a masonry wall or a beam with negligible torsional stiffness, cannot develop any negative moment at all: with nothing to resist rotation at that edge, the slab behaves there as if simply supported, so the exterior negative moment is exactly zero and the positive moment, at $0.63M_o$, is close to the $0.625M_o$ a genuinely simply supported span would carry. Condition (5), an exterior edge built monolithically into a stiff wall or a torsionally rigid spandrel, approaches full fixity, and its row is, by design, identical to the interior-span split: $0.65M_o$ negative, $0.35M_o$ positive. Conditions (2) through (4) fill in the space between these two extremes in order of decreasing restraint, slab-and-beam systems on all sides restrain the edge more than a flat plate or flat slab with no edge beam at all, and adding an edge beam to an otherwise beamless system restrains it a little more still, so exterior negative moment climbs steadily from 0 to 0.65 $M_o$ across the five rows while positive moment falls in step from $0.63M_o$ to $0.35M_o$, and interior negative moment moves only modestly, since the interior face of an end span is already close to a genuinely interior condition regardless of what the exterior edge is doing.

## Lateral Distribution: Column Strips and Middle Strips

Every moment obtained above, $0.65M_o$, $0.35M_o$, or one of the end-span entries, is still a moment for the full transverse width $\ell_2$ of the design strip. It must still be apportioned, across that width, between the column strip centered on the support line, with a half-width on each side equal to the lesser of $0.25\ell_1$ and $0.25\ell_2$ (8.4.1.5, already introduced in the previous notebook), and the middle strip that occupies the remainder. ACI 318-14 8.10.5 fixes that apportionment as a percentage of each longitudinal moment assigned to the column strip, the rest going to the middle strip, and the percentage itself depends on three things: the aspect ratio $\ell_2/\ell_1$ of the panel, the relative stiffness $\alpha_{f1}\ell_2/\ell_1$ of any beam running in the direction of the moment ($\alpha_{f1}=0$ when there is no such beam), and, for the exterior negative moment only, a torsional stiffness ratio $\beta_t$ that measures how effectively an edge beam or wall resists twisting about its own axis.

Column strip and middle strip layout in plan, for the row of columns spanning in direction 1 (schematic, not an ACI figure).</figcaption></figure>

The column strip has a fixed width, set purely by geometry ($0.25\ell_1$ or $0.25\ell_2$ per side, whichever is smaller), independent of how the moment itself is distributed. For a square panel, where that fixed width happens to make the column strip and the middle strip equally wide, the figure makes a point already raised in this notebook easy to see directly: an equal-width split of the transverse dimension is not the same thing as an equal split of the moment. The percentage tables that follow decide how much of each moment, not how much width, goes to the shaded band.

**Interior negative moment** (ACI 318-14 Table 8.10.5.1):

| $\alpha_{f1}\ell_2/\ell_1$ | $\ell_2/\ell_1=0.5$ | $\ell_2/\ell_1=1.0$ | $\ell_2/\ell_1=2.0$ |
|---|---|---|---|
| $0$ | 75% | 75% | 75% |
| $\geq1.0$ | 90% | 75% | 45% |

with linear interpolation given by $\text{CS}\%=75+30\left(\alpha_{f1}\dfrac{\ell_2}{\ell_1}\right)\left(1-\dfrac{\ell_2}{\ell_1}\right)$.

**Positive moment** (ACI 318-14 Table 8.10.5.5):

| $\alpha_{f1}\ell_2/\ell_1$ | $\ell_2/\ell_1=0.5$ | $\ell_2/\ell_1=1.0$ | $\ell_2/\ell_1=2.0$ |
|---|---|---|---|
| $0$ | 60% | 60% | 60% |
| $\geq1.0$ | 90% | 75% | 45% |

For intermediate values of $\alpha_{f1}\ell_2/\ell_1$ between 0 and 1.0, ACI 318-14 intends straight-line interpolation between the tabulated $\alpha_{f1}\ell_2/\ell_1=0$ and $\alpha_{f1}\ell_2/\ell_1\geq1.0$ rows within the same $\ell_2/\ell_1$ column. No closed-form shortcut is given here: a compact formula for this specific table, circulating in non-ACI course material, was checked cell by cell while preparing this notebook and does not reproduce the table's own $\alpha_{f1}\ell_2/\ell_1\geq1.0$ row (it predicts 75%, 60%, and 45% at $\ell_2/\ell_1=0.5$, $1.0$, and $2.0$ where the table above reads 90%, 75%, and 45%), so it is omitted rather than repeated here unverified. The two interior-negative and exterior-negative interpolation formulas given elsewhere in this section were each checked the same way, cell by cell against their own tables, and both reproduce every tabulated value exactly.

**Exterior negative moment** (ACI 318-14 Table 8.10.5.2):

| $\alpha_{f1}\ell_2/\ell_1$ | $\beta_t$ | $\ell_2/\ell_1=0.5$ | $\ell_2/\ell_1=1.0$ | $\ell_2/\ell_1=2.0$ |
|---|---|---|---|---|
| $0$ | $0$ | 100% | 100% | 100% |
| $0$ | $\geq2.5$ | 75% | 75% | 75% |
| $\geq1.0$ | $0$ | 100% | 100% | 100% |
| $\geq1.0$ | $\geq2.5$ | 90% | 75% | 45% |

with $\text{CS}\%=100-10\beta_t+12\beta_t\left(\alpha_{f1}\dfrac{\ell_2}{\ell_1}\right)\left(1-\dfrac{\ell_2}{\ell_1}\right)$, where $\beta_t=E_{cb}C/(2E_{cs}I_s)$, $I_s$ is the second moment of area of the slab strip of width $\ell_2$, and $C$ is the torsional constant of the edge beam's cross-section obtained by summing $\left(1-0.63x/y\right)x^3y/3$ over the component rectangles of that section (the same thin-rectangle torsion approximation used throughout reinforced concrete practice; $x$ is the shorter side and $y$ the longer side of each component rectangle).

Two features of these tables are easy to misread if approached only as lookup tables rather than as physical statements. First, $\beta_t=0$ (no edge beam, or an edge resting on a wall with negligible torsional stiffness) still sends 100% of the exterior negative moment to the column strip, not less. A middle strip cannot pick up negative moment at a discontinuous edge by bending in its own plane; it can only receive that moment if an edge beam is stiff enough in torsion to twist and hand load sideways to it, which is exactly what a nonzero, and eventually large, $\beta_t$ represents. Second, the same $\alpha_{f1}\ell_2/\ell_1\geq1.0$ row that sends up to 90% of a moment to a nearly square column strip ($\ell_2/\ell_1=0.5$) sends only 45% to it when the panel is elongated ($\ell_2/\ell_1=2.0$): a fixed-width column strip, $0.25\ell_1$ or $0.25\ell_2$ on each side, is a shrinking fraction of an increasingly wide bay, so even a stiff beam concentrating moment on the support line leaves a growing majority of the panel's transverse width, and therefore a growing share of the total moment, to the middle strip.

Where beams run in the direction of the moment ($\alpha_{f1}\ell_2/\ell_1\geq1.0$), the beam itself is proportioned to carry a substantial share of the column-strip moment obtained above, increasing with $\alpha_{f1}\ell_2/\ell_1$, with the remaining column-strip moment carried by the slab within that strip. That beam/slab split inside the column strip belongs to beam-supported two-way systems specifically and is developed in detail in structural design references; it is not reproduced here because it falls outside the flat-plate and flat-slab emphasis this course has followed since the previous notebook, and because the exact interpolation was not independently verified against the 1971-1977 origin of the DDM tables during the preparation of this material and is better looked up directly than risked from memory.

## Minimum Reinforcement

Once the design moments are known, two independent minimum-reinforcement requirements govern every slab section regardless of what flexural demand actually computes to. ACI CODE-318-25 8.6.1.1 requires a minimum area of flexural reinforcement

$$A_{s,min}=0.0018A_g$$

placed near the tension face of the slab in the direction of the span under consideration, where $A_g$ is the gross cross-sectional area of the slab strip under consideration. The commentary to 8.6.1.1 notes explicitly that this is the same ratio required for shrinkage and temperature reinforcement in 24.4.3.2, so a slab that already carries its minimum flexural steel in both directions has, as a byproduct, already satisfied its shrinkage and temperature requirement in both directions as well.

Two things make this particular ratio easier to apply correctly than most ACI provisions calibrated to a specific unit system. First, $0.0018$ is a dimensionless ratio of reinforcement area to gross concrete area: it returns the same $A_{s,min}$ whether $A_g$ is expressed in $\mathrm{cm^2}$, $\mathrm{in^2}$, or any other consistent area unit, with no conversion constant of the kind flagged for the Table 7.3.1.1 thickness modifier in the previous notebook. Second, and worth stating explicitly because it is easy to assume otherwise from older editions of the Code, this ratio no longer depends on $f_y$. The commentary to 24.4.3.2 records that previous editions allowed a reduced ratio for reinforcement with $f_y>60{,}000$ psi, and that this reduction has been removed: *"the mechanics of cracking suggest that increased yield strength provides no benefit for the control of cracking."* A grade 4200 $\mathrm{kgf/cm^2}$ design and a grade 60,000 psi design therefore use exactly the same $0.0018$, with no grade-dependent adjustment to look up or convert.

The spacing limits that go with this reinforcement, unlike the ratio itself, are genuine length quantities written in US customary units and do need a direct conversion for use in centimeters. ACI CODE-318-25 24.4.3.3 caps the spacing of shrinkage and temperature reinforcement at the lesser of $5h$ and 18 in.; 8.7.2.2 caps the spacing of ordinary flexural reinforcement in a nonprestressed solid slab at the lesser of $2h$ and 18 in. at critical sections, and the lesser of $3h$ and 18 in. elsewhere. The $2h$, $3h$, and $5h$ terms are already unit-free multiples of the slab thickness and transfer directly into centimeters once $h$ is expressed in centimeters; the fixed 18 in. cap does not, and converts, at $1\text{ in.}=2.54\text{ cm}$, to

$$18\text{ in.}=45.72\text{ cm}\approx46\text{ cm}$$

This is not a purely academic distinction. For a $h=20$ cm slab, $2h=40$ cm is smaller than the 45.72 cm cap, so the $2h$ limit governs and the spacing check has nothing to do with the 18 in. figure at all. For a $h=30$ cm slab, $2h=60$ cm now exceeds 45.72 cm, so the fixed cap governs instead and the true limit is 45.72 cm, not 60 cm. A designer who remembers only "18 inches" and substitutes the number 18 directly as if it were already a spacing in centimeters would under-space the reinforcement in the first case and, more dangerously, believe a 60 cm spacing were acceptable in the second, when the Code cap actually holds it to 45.72 cm, a difference of nearly 24 percent.

## Corner Restraint and Termination of Reinforcement

### Corner Restraint

An unrestrained slab corner, one bounded only by simply supported edges with no torsional restraint, tends to lift off its supports under load rather than developing the corner moments a fully continuous plate would carry. Where that lifting tendency is instead restrained, at exterior corners supported by edge walls, or wherever an edge beam has $\alpha_f>1.0$, ACI CODE-318-25 8.7.3.1 requires top and bottom reinforcement at the corner designed for a factored moment per unit width equal to the maximum positive $M_u$ per unit width already computed for the panel: the corner is, in effect, treated as demanding on the order of the interior span itself. That reinforcement extends a distance equal to one-fifth of the longer span in each direction from the corner (8.7.3.1.2), and may be placed either as bars parallel to the diagonal from the corner in the top of the slab and perpendicular to the diagonal in the bottom, or, as an equally acceptable alternative, as two layers parallel to the slab edges in both the top and the bottom (8.7.3.1.3).

### Minimum Bar Extensions

For slabs without beams, ACI CODE-318-25 Figure 8.7.4.1.3 sets the minimum length and extension of every bar category, column-strip top and bottom, middle-strip top and bottom, at both an exterior support with no slab continuity and an interior support with continuity provided, as fractions of the clear span $\ell_n$.

ACI CODE-318-25 Fig. 8.7.4.1.3, minimum extensions for deformed reinforcement in two-way slabs without beams.</figcaption></figure>

Reading the figure alongside the two-way shear mechanism developed in the previous notebook explains why it is drawn the way it is. Ordinary top column-strip bars extend $0.30\ell_n$ past an interior support (or $0.20\ell_n$ for the remainder not needed at the critical moment section), enough to cover where the negative-moment diagram from a uniformly loaded continuous strip actually needs steel. But punching shear cracks, as the previous notebook showed, can initiate at angles as shallow as about 20 degrees to the slab plane, tracing a cone that reaches well beyond where a moment-diagram cutoff alone would place the bar end. For that reason 8.7.4.1.3(a) requires at least half of the column-strip top bars to extend not less than $5d$ past the face of support, explicitly to intercept that inclined crack rather than only the flexural demand. All column-strip bottom bars, by contrast, are required to be continuous or spliced through the support region (100% at the bottom row of the figure), consistent with the structural integrity requirement discussed next; the figure marks a 6 in. minimum embedment for bars that terminate at a discontinuous exterior support, which converts, at $1\text{ in.}=2.54\text{ cm}$, to $6\text{ in.}=15.24\text{ cm}$.

### Structural Integrity

The residual-strength logic introduced conceptually in the previous notebook becomes a specific detailing rule here. ACI CODE-318-25 8.7.4.2.1 requires all bottom deformed bars or wires within the column strip, in each direction, to be continuous or spliced with mechanical, welded, or Class B tension lap splices, located within the region the figure above marks as permitted for splices. Section 8.7.4.2.2 goes further for at least two of those column-strip bottom bars in each direction: they must pass within the region bounded by the column's own longitudinal reinforcement, and must be developed in tension using a bar stress of $1.25f_y$, not $f_y$, at exterior supports. That 25 percent increase over the ordinary development demand is deliberate: these two bars are the ones expected to still be functioning, hanging the slab from the adjacent support, after a punching failure has already occurred at the column immediately above them, a condition more demanding than the intact, undamaged state every other development-length calculation in this course has assumed so far.

## Correlation with ACI CODE-318-25

The boundary drawn at the start of this notebook is worth restating now that both halves of the design procedure have been developed. The direct design method itself, the total static moment, the longitudinal 65/35 and end-span distributions, and the column-strip and middle-strip apportionment tables, is authorized by ACI CODE-318-25 6.2.4.1 but is no longer printed inside the current Code; every numbered provision cited for it in this notebook belongs to ACI 318-14. Everything else developed here is fully normative in ACI CODE-318-25 as it stands today: minimum flexural reinforcement and its unit-free 0.0018 ratio (8.6.1.1), shrinkage and temperature reinforcement (24.4.3), spacing limits (8.7.2), corner restraint (8.7.3), bar extension and termination (8.7.4.1), and structural integrity (8.7.4.2).

One boundary flagged only in passing in the previous notebook deserves a second, sharper mention now that minimum reinforcement has been developed in full. Section 8.6.1.2 sets a second, larger $A_{s,min}$, triggered whenever the factored two-way shear stress at a slab-column connection exceeds a stated fraction of the concrete's two-way shear strength, specifically to prevent a brittle, sliding-driven punching failure in lightly reinforced slabs. That trigger condition depends on $v_c$, the permissible two-way shear stress, which, as the previous notebook already noted, lives in Chapter 22 rather than in Chapter 8. This notebook has developed ordinary flexural minimum reinforcement in full; the shear-triggered minimum of 8.6.1.2, like the punching check it protects against, is deferred to the next notebook, where $v_c$ is finally developed quantitatively.

## Transition to Punching Shear and the Equivalent Frame Method

Two pieces of the two-way slab design sequence remain open. The direct design method developed here only applies within the six-condition envelope of ACI 318-14 8.10.2; panels with irregular geometry, unequal spans beyond the one-third tolerance, or beam stiffness ratios outside the 0.2-to-5.0 window require the equivalent frame method instead, an analysis that idealizes the structure as a series of slab-beam and column elements connected through a torsional member representing the slab's own torsional flexibility. And the two-way shear condition introduced conceptually in the previous notebook, the eccentric shear stress $v_u=V_u/A_c\pm\gamma_vM_{sc}c/J_c$ acting around the critical section at $d/2$ from the column face, has so far been developed only qualitatively. ## Summary

The direct design method converts a two-way slab panel into design moments through three sequential steps: a total factored static moment $M_o=q_u\ell_2\ell_n^2/8$ for each span, a longitudinal split of that moment into negative and positive regions (0.65/0.35 for an interior span, a five-condition table spanning free to fully restrained for an end span), and a lateral apportionment of each of those moments between column strip and middle strip, governed by panel aspect ratio, relative beam stiffness $\alpha_{f1}\ell_2/\ell_1$, and, at exterior edges, torsional stiffness $\beta_t$. ACI CODE-318-25 authorizes this procedure by reference (6.2.4.1) without reproducing it, so every numerical coefficient above traces to ACI 318-14 rather than to the current Code text. What the current Code text does state directly, and what closes every slab design regardless of which analysis method produced the moments, is a unit-free 0.0018 minimum flexural reinforcement ratio shared with shrinkage and temperature steel, spacing limits that do require a genuine unit conversion (18 in. equals 45.72 cm), corner reinforcement for restrained corners, minimum bar extensions calibrated to intercept both the flexural demand and the inclined punching crack, and structural integrity reinforcement developed at an elevated $1.25f_y$ to give the slab residual strength after a local punching failure.

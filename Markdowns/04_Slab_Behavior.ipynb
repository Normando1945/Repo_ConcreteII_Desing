## Slabs in the Structural System

A slab is the structural element that first receives most of the gravity loads described in the previous notebook: occupancy loads, partitions, finishes, snow or rain accumulation, and its own self-weight act directly on the slab surface before any of that demand reaches a beam, a wall, or a column. In load-path terms, the slab is the origin of the vertical path introduced earlier: surface load enters the slab, the slab transfers it to its supporting beams, walls, or columns, and those members carry it downward to the foundation.

What changes with respect to every member studied so far is dimensionality. A beam is idealized as a line element: one span, one bending-moment diagram, one governing direction of curvature. A slab is a flat, thin, two-dimensional continuum whose thickness is small compared with its plan dimensions. A point inside a slab can curve simultaneously about two orthogonal axes, and the load reaching a given support depends on how the slab is bounded on all of its edges, not on a single span length. This notebook develops that difference physically before any code provision is introduced: why some slabs still behave essentially like a family of parallel beams (one-way action), why others genuinely redistribute load in two directions (two-way action), what typical construction forms exist for each case, and how ACI CODE-318-25 organizes its slab provisions around this same physical distinction.

## A Short History of Slab Construction

Long before reinforced concrete, flat spanning surfaces were the hardest structural problem to solve economically. Masonry and timber floors either spanned short distances between closely spaced joists or relied on vaulting and arching action, converting bending into compression by curving the surface itself. A thin, flat, horizontal plate loaded transversely has almost no capacity in plain masonry or unreinforced concrete, because it immediately develops tension on its underside, precisely the stress plain materials cannot resist. The slab as a genuinely flat spanning element only becomes structurally viable once steel reinforcement is available to carry that tension.

Reinforced concrete slabs entered practice at nearly the same moment as reinforced concrete framing itself, in the last years of the nineteenth century, closely tied to European monolithic beam-and-slab systems such as Hennebique's patented frame. A more radical step followed shortly after: eliminating the beams altogether. C. A. P. Turner's flat slab system, patented in the United States in the first decade of the twentieth century, carried floor loads directly from a solid slab into enlarged column heads, capitals, and drop panels, with no beams framing into the columns at all. This was not a beam analogy; it was, for the first time in reinforced concrete practice, a genuine plate resting directly on a grid of columns. Because no rational theory yet existed to justify these proportions, the system was defended largely through proprietary formulas and full-scale load tests rather than derived mechanics, and it provoked open technical disputes for years. Independently, in Europe, Robert Maillart refined closely related mushroom-column and slab-column structures, again validated primarily through testing.

A rational footing for the flat slab arrived only once classical elastic plate theory was applied to it. Approximate static-equilibrium procedures, most notably the one published by Nichols in 1914, gave engineers a way to check the total moment a flat slab panel had to resist without solving the governing plate differential equation directly. Westergaard's and Slater's later refinements of moment distribution between column and middle regions, followed decades later by Johansen's yield-line theory in the 1940s and 1950s, added both an elastic serviceability picture and a plastic capacity picture of two-way behavior. Twentieth-century codes absorbed this dual heritage directly: a coefficient-based procedure for slabs of sufficiently regular geometry, coexisting with a fully general mechanics-based path for anything else. That same duality organizes ACI CODE-318-25 today, as the closing section of this notebook makes explicit.

## Classification and Types of Slab Systems

### One-Way and Two-Way Action

Whether a slab panel behaves as a one-way or a two-way member is not a construction choice; it is a consequence of geometry and boundary support, and it can be seen from a simple compatibility argument before any code rule is introduced. Imagine a rectangular panel simply supported on all four edges, with clear spans $\ell_x$ (short direction) and $\ell_y$ (long direction), idealized as two crossing unit-width strips meeting at the panel center. Each strip must deflect the same amount at that shared point. Treating each strip as a simply supported beam carrying its share $w_x$ or $w_y$ of the total uniform load $w=w_x+w_y$, compatibility of central deflection gives

$$\frac{5w_x\ell_x^4}{384EI}=\frac{5w_y\ell_y^4}{384EI}$$

so that

$$\frac{w_x}{w_y}=\left(\frac{\ell_y}{\ell_x}\right)^4$$

Stiffness against transverse deflection grows with the fourth power of the inverse span, so the shorter strip attracts load far out of proportion to its width. At $\ell_y/\ell_x=1$ the two directions share the load equally; at $\ell_y/\ell_x=2$, the short direction alone already carries $2^4=16$ times what the long direction carries, on the order of 94 percent of the total load. Beyond that ratio, treating the panel as spanning only in the short direction stops being a simplification and starts being an accurate description of how the slab actually behaves. This is close to the boundary ACI CODE-318-25 commentary points to directly for slabs with beams on all four sides: for panels with a long-to-short span ratio greater than 2, the thickness expressions built on this two-way stiffness ratio can give unrealistic results, and the one-way slab rules should govern instead. One-way and two-way action are therefore two ends of the same physical continuum, separated by how effectively the stiffer, shorter direction absorbs the load before the longer direction has to participate.

### Types of One-Way Slab Systems

A one-way slab may be a solid slab of uniform thickness spanning between parallel beams or walls, in which every design strip behaves like the beams already studied, or a one-way joist (ribbed) system, in which the slab depth is subdivided into closely spaced ribs with the material between them removed. The rib arrangement keeps the concrete near the extreme tension and compression fibres, where it does the most flexural work, while removing self-weight from material near the physically inactive neutral-axis region. Both forms answer to the same governing equilibrium and compatibility relationships developed for beams; what changes is only the width and shape of the compression zone that those relationships must be applied to.

### Types of Two-Way Slab Systems

When a panel is supported so that both directions participate structurally, several construction forms follow from the same underlying two-way action, distinguished mainly by how the slab meets its columns:

- A flat plate is a solid slab of uniform thickness resting directly on columns, with no beams, drop panels, or column capitals. It offers the simplest formwork and the shallowest floor-to-floor height, at the price of concentrating an unusually demanding shear and moment transfer directly into a thin slab-column connection.
- A flat slab adds a drop panel, a column capital, or both, thickening or widening the slab locally around the column. This directly increases the concrete available to resist the two-way shear and negative moment that concentrate exactly where the flat plate is most vulnerable.
- A two-way slab with beams on all four sides of every panel places genuinely stiff line supports along the panel boundary, which is the condition classical elastic two-way plate solutions assume most closely; the beams themselves must then be designed for the reactions the slab delivers to them.
- A waffle (two-way joist) slab extends the one-way joist idea into both directions: a grid of ribs runs orthogonally over most of the panel, with a solid head retained around the columns where two-way shear and concentrated moment require full-depth concrete.

## One-Way Slab Behavior

### Design Strip and Load Path

A one-way slab is designed as a series of parallel strips of unit width, each idealized as a shallow, continuous beam. Every equilibrium, compatibility, and constitutive relationship developed for beams in the previous two notebooks, plane sections remaining plane, the same strain-compatibility argument for the neutral axis, the same Whitney-block idealization for nominal flexural strength, applies to a one-way slab strip without modification other than expressing moment, shear, and reinforcement per unit width instead of for a fixed web width $b_w$. Physically, this idealization is only exact if the strip does not need help from its neighbours to remain in equilibrium, which is why the aspect-ratio argument above is a precondition, not an afterthought, for treating a slab as one-way at all.

### Minimum Thickness and Continuity

$$\kappa=\frac{M}{EI}$$

remains the governing relationship between demand and curvature developed earlier in this course, so the same conclusion applies here: less curvature demand for the same span allows a thinner section without violating the deflection limits implicit in the code's serviceability screening. ACI CODE-318-25 Table 7.3.1.1 expresses this directly through span-to-depth ratios keyed to end continuity: a simply supported strip must satisfy $h\geq \ell/20$, a strip continuous at one end $h\geq \ell/24$, a strip continuous at both ends $h\geq \ell/28$, and a cantilever $h\geq \ell/10$. Continuity redistributes moment away from midspan into the supports and reduces peak curvature demand for a given span, which is exactly why continuous strips are allowed to be proportionally thinner, and why the isolated cantilever, with no redistribution available at all, requires by far the largest span-to-depth ratio of the four.

### Units: Converting the fy Modifier to kgf/cm2

Table 7.3.1.1 itself is unit-free: $\ell/20$, $\ell/24$, $\ell/28$, and $\ell/10$ are ratios of length to length, so $h$ comes out in whatever consistent length unit $\ell$ was measured in, centimeters included. The modifier in 7.3.1.1.1 is a different matter. ACI CODE-318-25 writes it strictly in US customary units, calibrated so that $f_y=60{,}000$ psi gives a modifier of exactly 1.0:

$$0.4+\frac{f_y}{100{,}000}\quad(f_y\text{ in psi})$$

ACI CODE-318-25 does not print this modifier in $\mathrm{kgf/cm^2}$ directly, but ACI 318-19 Appendix C, which tabulates the equivalent form of every non-homogeneous equation in the Code across SI-metric, mks-metric, and U.S. customary units, gives the mks-metric ($\mathrm{kgf/cm^2}$) form as

$$0.4+\frac{f_y}{7000}\quad(f_y\text{ in kgf/cm}^2)$$

The denominator is a clean 7000, not the 7031 that dividing 100,000 by the exact factor $0.070307$ would produce: ACI's own published conversion pairs grade 60 (60,000 psi) with the nominal designation $f_y=4200\text{ kgf/cm}^2$, the same commercially rounded grade used throughout local practice, rather than the mathematically exact $60{,}000\times0.070307=4218\text{ kgf/cm}^2$. Substituting $f_y=4200$ into the two forms confirms they are meant to agree: $0.4+4200/7000=1.00$ exactly, matching $0.4+60{,}000/100{,}000=1.00$. Someone who instead substitutes $f_y=4200\text{ kgf/cm}^2$ directly into the psi-based constant of 100,000, without converting units first, gets $0.4+4200/100{,}000=0.44$, a silent error that would understate the required minimum thickness by more than half.

### Transverse Reinforcement

The unit-width design strip is a modelling device, not a physical boundary: the concrete on either side of the strip is still continuous with it and still shrinks, still responds to temperature change, and still wants to crack transversely to the main span direction. ACI CODE-318-25 requires minimum reinforcement perpendicular to the main flexural steel for exactly this reason, unrelated to the flexural demand computed along the span. Physically, this transverse steel does two things at once: it controls shrinkage and temperature cracking that has nothing to do with the applied moment, and it ties adjacent design strips together so that the slab behaves as one continuous surface rather than as a set of independent planks.

## Two-Way Slab Behavior

### Plate Bending and Orthogonal Moments

Once a panel genuinely works in two directions, curvature at a point is no longer described by a single number. Two orthogonal bending curvatures, $\kappa_x$ and $\kappa_y$, together with a twisting curvature $\kappa_{xy}$, act simultaneously and produce two bending moments per unit width, $m_x$ and $m_y$, plus a twisting moment $m_{xy}$. Equilibrium of a differential plate element under a distributed transverse load $q$ replaces the familiar beam relationship $d^2M/dx^2=-w$ with its two-dimensional counterpart

$$\frac{\partial^2 m_x}{\partial x^2}+2\frac{\partial^2 m_{xy}}{\partial x\,\partial y}+\frac{\partial^2 m_y}{\partial y^2}=-q$$

This single equation is the two-way analogue of everything the beam equilibrium equation has been doing throughout this course; it simply distributes the same transverse load over two coupled bending fields instead of one. It assumes a thin, elastic, small-deflection plate, and it is not solved directly in ordinary design practice. Its role here is conceptual: it explains why a two-way slab cannot be reduced to two independent one-way strips without some loss of information, since the twisting term $m_{xy}$ carries load that a pair of crossing beams, considered separately, cannot represent.

### Column Strips and Middle Strips

Rather than solving the plate equation directly, ACI CODE-318-25's design methods convert the continuous field of $m_x$ and $m_y$ into discrete design values by first computing a total static moment for a panel and then apportioning it between two kinds of design strips. A column strip is a design strip centered on the column centerline, with a width on each side equal to the smaller of $\ell_1/4$ and $\ell_2/4$; a middle strip occupies the remaining width between two column strips. Physically, this apportionment concentrates reinforcement where curvature, and therefore moment, is genuinely largest: near the column line, where the slab must transfer both flexure and, as the next section shows, a demanding shear condition into a small support area. The exact fractions of static moment assigned to column strip and middle strip, and to the negative and positive regions of the span, belong to the direct design method developed in the next notebook; the concept established here is why any such apportionment is necessary at all.

### Minimum Thickness for Two-Way Systems

<figure style="margin:28px 0; text-align:center;"><img src="assets/04_slab_behavior/fig-02.png" alt="Minimum thickness of nonprestressed two-way slabs without interior beams" width="1100" style="max-width:100%; border-radius:8px;"><figcaption style="margin-top:10px; color:#475569; font-size:0.9em;">Figure. ACI CODE-318-25 Table 8.3.1.1, minimum thickness for flat plates and flat slabs without interior beams.</figcaption></figure>

ACI CODE-318-25 Table 8.3.1.1 sets minimum thickness for nonprestressed two-way slabs without interior beams as a function of clear span $\ell_n$, the tension-steel yield strength $f_y$, whether the panel is interior or exterior, whether an edge beam is present, and whether a drop panel is used. Every one of these variables changes deflection behaviour physically: an exterior panel without an edge beam loses the rotational restraint an interior panel enjoys from continuity on all sides; a drop panel adds local stiffness exactly where negative moment and shear concentrate; a higher $f_y$ typically corresponds to less reinforcement for the same moment, hence a less stiff cracked section and a larger required thickness to keep deflections in check. Where a two-way panel is instead supported on beams along all four edges, ACI directs the designer to a separate thickness expression governed by the relative stiffness of those beams to the slab, because the beams now share meaningfully in carrying the panel's curvature.

Because the table's $f_y$ axis is itself printed directly in psi (40,000, 60,000, and 80,000), rather than as a formula, the conversion has to be applied to those three values themselves. Following the same ACI 318-19 Appendix C correspondence used above for the nominal grade designations, rather than a literal $1\text{ psi}=0.070307\text{ kgf/cm}^2$ conversion of the printed psi numbers, these become the clean grade values 2800, 4200, and 5600 $\mathrm{kgf/cm^2}$. A design carried out in kgf/cm2 should interpolate, per the table's own footnote, against these converted grade values directly, not interpolate against the printed psi numbers using a locally reported $f_y$ in kgf/cm2 as if the two scales matched one for one.

## Two-Way Shear (Punching) Behavior

### Punching Failure Mechanism

<figure style="margin:28px 0; text-align:center;"><img src="assets/04_slab_behavior/fig-03.png" alt="Punching shear cracks in ordinary and thick slabs" width="900" style="max-width:100%; border-radius:8px;"><figcaption style="margin-top:10px; color:#475569; font-size:0.9em;">Figure. Punching shear failure surface in an ordinary slab and in a thick slab.</figcaption></figure>

Near a column, the entire reaction that the slab must deliver to that column concentrates through a slab thickness that is often only a few inches, over a loaded area no larger than the column cross-section itself. One-way (beam-type) shear failure develops along a roughly planar diagonal crack running across the full width of a member; two-way shear failure develops instead as an inclined, three-dimensional cracked surface that surrounds the column on every side, punching a truncated cone or pyramid out of the slab. ACI CODE-318-25 commentary illustrates this directly for an ordinary slab and for a thick slab: in both cases, a diagonal crack initiates at the column face and inclines outward and upward toward the top surface, but the crack surface in a thicker slab is comparatively shallower relative to the span, so top flexural reinforcement must extend further, out to about $5d$ from the column face rather than stopping at the $0.3\ell_n$ cutoff sufficient for an ordinary slab, to make sure that reinforcement actually crosses the real inclined crack rather than the idealized flexural cutoff point. This is the same physical lesson already met in this course for shear-related detailing: a cutoff decided from a bending-moment diagram alone is not automatically safe against an inclined crack that does not follow that diagram.

### Eccentric Shear Transfer

<figure style="margin:28px 0; text-align:center;"><img src="assets/04_slab_behavior/fig-04.png" alt="Assumed distribution of shear stress at interior and edge columns" width="950" style="max-width:100%; border-radius:8px;"><figcaption style="margin-top:10px; color:#475569; font-size:0.9em;">Figure. Eccentric shear stress distribution at interior and edge column connections.</figcaption></figure>

At an interior column, the column reaction is usually delivered symmetrically, and the shear stress around the critical section, taken at a distance $d/2$ from the column face, is close to uniform. Wherever the connection must also transfer an unbalanced moment $M_{sc}$, at every edge and corner column, and at interior columns with unequal adjacent spans or loading, part of that moment is resisted by flexure across the critical section and part is resisted by a shear stress that varies linearly around the perimeter, exactly the way a combined axial-and-bending stress $\sigma=P/A\pm Mc/I$ varies linearly across a cross-section, except that here it is a shear stress varying around a perimeter rather than a normal stress varying across an area. This is reflected conceptually in the design shear stress

$$v_u=\frac{V_u}{A_c}\pm\frac{\gamma_v M_{sc}\,c}{J_c}$$

where $A_c$ is the area of the critical section and $J_c$ is a property of that section analogous to a polar moment of inertia. An edge column has a critical section open on the outside face rather than closed on all four sides; that missing perimeter both reduces the available area $A_c$ for direct shear and shifts the centroid of the section, which is exactly why edge and corner columns are consistently the most demanding two-way shear condition in a flat-plate or flat-slab system. The full definitions of $\gamma_v$ and $J_c$, and the corresponding permissible stress $v_c$, are developed with the design method in the next notebook; what matters here is the physical picture of direct shear and moment-transfer shear stress acting together around the same critical perimeter.

### Structural Integrity After a Local Punching Failure

Because punching is a brittle failure, with comparatively little deformation warning before the cracked cone separates from the slab, ACI CODE-318-25 requires that a minimum number of bottom column-strip bars pass continuously through the column core in each direction, developed in tension at the column. Physically, this integrity reinforcement gives the slab a residual ability to hang from adjacent supports if a single column connection does fail, converting what would otherwise be an abrupt local collapse into a condition the structure has some chance of surviving. This is the same capacity-design logic already introduced for flexural members: strength at one location is necessary, but a structure that also survives the unexpected local loss of that strength is a different, more robust, design objective.

## Correlation with ACI CODE-318-25: Chapters 7 and 8

The classification developed in this notebook is not an independent framework laid over the code; it is how ACI CODE-318-25 itself organizes slab design. Chapter 7 governs solid nonprestressed and prestressed one-way slabs, and its own commentary repeatedly points back to the beam provisions already studied in this course: flexural design, one-way shear, and reinforcement detailing for a one-way slab are treated as the same problem already solved for beams, applied to a unit-width strip.

Chapter 8 governs two-way slab systems, defined broadly enough to include solid slabs, slabs cast on stay-in-place noncomposite steel deck, composite slabs built from separately placed elements connected to act as a unit, and two-way joist systems, but explicitly excluding slabs-on-ground that do not transmit load from the structure into the soil. Its general provision permits a two-way slab to be designed by any procedure that satisfies equilibrium and geometric compatibility everywhere, provided strength and serviceability are verified: the direct design method and the equivalent frame method are offered as standardized paths for regular geometries, while classical elastic-continuum solutions, discrete numerical methods, and yield-line analysis remain available in general. This is precisely the dual heritage traced in the historical section: a coefficient-based path for ordinary cases, and a fully general mechanics-based path behind it.

One boundary is easy to lose track of and worth stating explicitly: the two-way shear design values used to check the punching condition just developed, the permissible stress $v_c$ and its associated expressions, are not found in Chapter 8 itself but in Chapter 22, the shear chapter shared with beams and one-way slabs, and are only invoked by cross-reference from Chapter 8's design-strength requirements. A slab design that treats Chapter 8 as self-contained will miss this dependency.

## Transition to Slab Design

The behaviour developed here is the basis for the next notebook. One-way slab strips will be designed with exactly the flexural and shear procedures already coded for beams, applied per unit width. Two-way slabs will be designed with the direct design method: computing the total static moment for a panel, distributing it longitudinally between negative and positive regions, and apportioning it laterally between column strips and middle strips, followed by the equivalent frame method for geometries the direct design method cannot cover. Two-way shear will be checked quantitatively at the critical section, including the eccentric shear stress transfer introduced conceptually above, and minimum flexural and temperature reinforcement, together with structural integrity detailing, will close the design sequence.

## Summary

A slab's behavior is decided by geometry before it is decided by any code table: a short-to-long span ratio beyond about 2 makes two-way redistribution physically negligible, and one-way beam behavior becomes an accurate description rather than a simplification. Below that ratio, a slab genuinely bends about two axes at once, concentrates its most demanding flexural and shear conditions at the column line, and can fail in a brittle, three-dimensional punching mode that a beam never experiences. ACI CODE-318-25 organizes Chapters 7 and 8 around exactly this physical distinction, sharing the beam provisions already developed in this course for one-way action, and adding column-strip/middle-strip moment distribution, punching shear, and structural integrity requirements wherever two-way action governs.

## Columns in the Structural System

The load path traced in the loads notebook ends, for every floor of a building, in the same place: slabs deliver their reactions to beams or directly to columns, beams deliver theirs to columns, and columns carry the accumulated sum down to the foundation. A column is therefore the one member in a frame whose failure is never local. A beam that fails in flexure loses a span; a column that fails in compression can take with it every floor it supports above, which is exactly why the Code treats columns more conservatively than any member studied so far in this course.

A column is rarely loaded by axial force alone. Frame continuity, unbalanced gravity loads on adjacent spans, lateral loads, and simple construction tolerances all introduce bending moment together with the axial force, so the design question is never "how much axial load can this section carry" but "which combinations of axial force $P$ and bending moment $M$ can this section carry at the same time." ACI CODE-318-25 states this directly in 10.4.2.1: $P_u$ and $M_u$ occurring simultaneously for each applicable factored load combination shall be considered. The answer to that question, for a given cross section and set of materials, is a single curve in the $P$-$M$ plane: the interaction diagram. Building that curve by hand, from strain compatibility and equilibrium alone, is the purpose of this notebook.

The flexure notebooks already contain every physical ingredient needed. The interaction diagram is nothing more than the flexural analysis of 03 repeated for a section that also carries axial force, so that equilibrium no longer forces the compression resultant to equal the tension resultant.

## A Short History of the Reinforced Concrete Column

### From Masonry to Reinforced Concrete

For most of the history of construction, columns were masonry or stone, materials that work almost entirely in compression and whose design was governed by proportion and experience rather than calculation. The reinforced concrete column appears in the last decade of the nineteenth century, within the patented systems of the period: the system that Fran&ccedil;ois Hennebique patented in 1892 already combined longitudinal bars in columns with transverse ties, and it was licensed throughout Europe within a few years. The structural idea was immediately intuitive (steel bars to add strength and to hold the member together under accidental bending), but the theory needed to quantify it did not yet exist.

### Considere and the Spiral

Around 1902, the French engineer Armand Consid&egrave;re published the results of tests on what he called hooped concrete (*b&eacute;ton frett&eacute;*): concrete cylinders wrapped by closely spaced helical steel. He showed that a spiral does not simply add its own strength to the column; it confines the concrete core, restraining its lateral expansion under compression and raising both the strength and, more importantly, the deformation capacity of the core. A spirally reinforced column, once its outer shell spalls, keeps carrying load instead of failing abruptly. This single observation is the origin of the two families of columns that the Code still distinguishes today, tied and spiral, and of the higher strength reduction factor that spiral columns receive.

The mechanics of confinement were quantified at the University of Illinois by Richart, Brandtzaeg, and Brown (1928), who showed experimentally that the strength of concrete under a lateral confining pressure increases by roughly four times that pressure. The amount of spiral steel required by the Code (25.7.3) still follows the reasoning born from this research: provide enough spiral that the strength gained by the confined core at least replaces the strength lost when the unconfined shell spalls off.

### The ACI Column Investigation

During the 1920s, columns were designed by the working-stress method, assuming that concrete and steel shared the load in proportion to their elastic stiffness through the modular ratio $n=E_s/E_c$. A large experimental program sponsored by the American Concrete Institute between roughly 1927 and 1933, carried out at the University of Illinois under Richart and at Lehigh University under Slater and Lyse, showed that this assumption failed badly under sustained load: creep and shrinkage of the concrete transferred load progressively into the steel, so that the steel stress computed elastically had little to do with the real one. The same tests, however, showed something far more useful: whatever the load history, the ultimate capacity of an axially loaded column was reliably the sum of the concrete strength (close to $0.85f'_c$ acting on the net concrete area) and the yield force of the longitudinal steel. That observation is, almost word for word, the nominal axial strength $P_o$ of Eq. (22.4.2.2) used later in this notebook. Columns were thus the first reinforced concrete member whose design was, in substance, based on ultimate strength.

### From Combined Loading to the Interaction Diagram

The step from axial load to axial load plus moment came with the equivalent rectangular stress block. Charles S. Whitney (1937) proposed replacing the curved concrete stress distribution by a uniform stress acting over a reduced depth, and applied it explicitly to members under "flexure or combined flexure and direct compression", the case of every real column. Eivind Hognestad's study of combined bending and axial load at the University of Illinois (1951) tested eccentrically loaded columns systematically, proposed the concrete stress-strain model still used in research, and supported the limiting compressive strain of 0.003 that the Code adopts in 22.2.2.1. The rectangular block was calibrated against tests in its present form by Mattock, Kriz, and Hognestad (1961), the reference the Code commentary still cites in R22.2.2.4.1.

Strength design entered the ACI Code as an appendix in ACI 318-56, was placed on equal footing with working stress design in ACI 318-63, and became the primary method in ACI 318-71. From that point on, the interaction diagram, the locus of all $(P_n, M_n)$ pairs a section can develop, became the central design tool for columns.

### The Strength Reduction Factor, Revised Three Times

The last part of this history is recent and still moving, and it matters directly for the calculation later in this notebook:

- **ACI 318-02** adopted the unified design provisions, which assign $\phi$ according to the net tensile strain $\varepsilon_t$ in the extreme tension reinforcement rather than according to member type (Mast 1992). Columns and beams became two regions of the same strain-based rule.
- **ACI 318-19** redefined the tension-controlled limit from the fixed value $\varepsilon_t=0.005$ to $\varepsilon_t=\varepsilon_{ty}+0.003$, so that the same ductility margin applies to higher-grade reinforcement (R21.2.2).
- **ACI CODE-318-25** added 21.2.2.3, which caps $\phi$ for members under combined moment and axial compression with $0.1f'_cA_g\le P_n\le P_{n,bal}$ by a linear interpolation on the axial force itself. The commentary states openly that this can produce lower axial and flexural design strengths than the 2019 edition. The worked example below shows by how much.

## Classification and Types of Columns

### By Transverse Reinforcement

ACI CODE-318-25 recognizes two families of transverse reinforcement for compression members, and the distinction runs through Chapters 10, 21, 22, and 25:

- **Tied columns.** Longitudinal bars held in position by individual closed ties at a regular spacing (22.4.2.4, 10.7.6.2, 25.7.2). Ties prevent buckling of the longitudinal bars and hold the cage together, but provide only modest confinement of the core. Rectangular and square columns, by far the most common in building practice, are almost always tied.
- **Spiral columns.** Longitudinal bars arranged in a circle and enclosed by a continuous helix at a small pitch (22.4.2.5, 10.7.6.3, 25.7.3). The spiral confines the core, so that after the shell spalls the column continues to carry load with considerable deformation capacity. This is why spiral columns receive $\phi=0.75$ instead of 0.65 when compression controls, and $P_{n,max}=0.85P_o$ instead of $0.80P_o$.

Composite columns, with a structural steel shape or tube acting together with concrete, are also covered by the Code but are outside the scope of this course.

### By Slenderness

A column is **short** when the additional moments caused by its own lateral deflection under axial load (second-order or $P$-$\Delta$ effects) are small enough to be neglected, so that its strength is governed entirely by the strength of its cross section. A column is **slender** when those additional moments are significant and must be added to the first-order moments before checking the section. ACI CODE-318-25 6.2.5.1 sets the slenderness limits below which these effects may be neglected, and 6.6.4 provides the moment magnification method for the slender case.

This notebook deals only with the cross section, that is, with short columns or with the section check that follows any slenderness analysis. Slenderness, biaxial bending, and transverse reinforcement detailing are the subject of the next column notebook.

### By Loading

A column whose factored moments act about only one principal axis of its section is in **uniaxial** bending, and its strength is described by a single two-dimensional interaction diagram. A corner column of a building frame, or any column with significant moments about both axes, is in **biaxial** bending, and its strength becomes a three-dimensional interaction surface. Uniaxial bending is treated here; biaxial bending is deferred.

## Behavior Under Axial Load

### Nominal Axial Strength

Under concentric axial compression, concrete and steel strain together. Reinforcing steel reaches its yield strain $\varepsilon_{ty}=f_y/E_s\approx0.0021$ well before concrete reaches its crushing strain, so at nominal strength the steel has yielded and the concrete carries its maximum stress, taken as $0.85f'_c$. That stress acts on the net concrete area, since the bars occupy part of the gross section. Superposition then gives the nominal axial strength of a nonprestressed member, ACI CODE-318-25 Eq. (22.4.2.2):

$$P_o=0.85f'_c\left(A_g-A_{st}\right)+f_yA_{st}$$

where $A_g$ is the gross area of the section and $A_{st}$ is the total area of longitudinal reinforcement. The factor 0.85 reflects the difference between concrete strength in a cylinder test and in a real column cast vertically, the same observation the ACI column investigation of the 1930s placed at the base of column design. The equation is dimensionally homogeneous and applies unchanged with stresses in $\mathrm{kgf/cm^2}$ and areas in $\mathrm{cm^2}$.

### Accidental Eccentricity and the Cap on Axial Strength

A perfectly concentric load does not exist in a real building. Construction tolerances, imperfect alignment of the column through the floors, and unavoidable unbalanced moments always introduce some eccentricity. Older editions of the Code required a minimum design eccentricity; the current Code accounts for it instead by capping the nominal axial strength (22.4.2.1, Table 22.4.2.1):

$$P_{n,max}=0.80P_o\quad\text{(tied columns)},\qquad P_{n,max}=0.85P_o\quad\text{(spiral columns)}$$

According to R22.4.2.1, these percentages approximate the axial strength at eccentricity-to-depth ratios of about 0.10 for tied and 0.05 for spiral columns. The same provision limits $f_y$ to a maximum of 80,000 psi for this calculation, which corresponds to the nominal designation $f_y=5600\ \mathrm{kgf/cm^2}$ in the technical-metric system (the same 70-$\mathrm{kgf/cm^2}$-per-1000-psi pattern of grade designations already used in the slab notebooks). Above that stress, the concrete is likely to crush before the steel can develop it.

### Limits on Longitudinal Reinforcement

ACI CODE-318-25 10.6.1.1 requires the longitudinal reinforcement of a nonprestressed column to satisfy

$$0.01A_g\le A_{st}\le0.08A_g$$

The lower limit provides resistance to bending that the analysis may not capture, and limits the effect of creep and shrinkage, which, as the column investigation showed, shift load from concrete to steel under sustained load until a lightly reinforced column could yield its bars at service. The upper limit is a practical constructability limit: above 8 percent the bars cannot be placed and the concrete cannot be consolidated properly, especially at lap splices, where the steel area doubles locally. In practice, ratios above about 4 percent are rarely economical.

## Design Assumptions for Combined Flexure and Axial Load

ACI CODE-318-25 22.4.1.1 requires nominal flexural and axial strength to be calculated with the same assumptions of 22.2 already used for beams in the flexure notebooks. They are restated here because every equation of the interaction diagram follows directly from them:

1. **Equilibrium** shall be satisfied at each section (22.2.1.1), and **compatibility** follows from assuming strain in concrete and in nonprestressed reinforcement proportional to the distance from the neutral axis (22.2.1.2): plane sections remain plane.
2. **The maximum usable strain at the extreme concrete compression fiber is 0.003** (22.2.2.1).
3. **The tensile strength of concrete is neglected** (22.2.2.2).
4. **Equivalent rectangular stress block** (22.2.2.4.1): a uniform stress of $0.85f'_c$ acts over a depth $a=\beta_1c$ from the extreme compression fiber, where $c$ is the depth of the neutral axis.
5. **Reinforcement is elastic-perfectly plastic** (20.2.2.1): $f_s=E_s\varepsilon_s$ while $|\varepsilon_s|\le\varepsilon_{ty}$, and $f_s=f_y$ (with the sign of the strain) beyond it.

### Units: the Values That Need Conversion

Most of these assumptions are unit-free (strains, ratios). Three numbers are not, and are handled as follows for the technical-metric system used in this course:

- **$\beta_1$, Table 22.2.2.4.3.** The transition expression is calibrated to $f'_c$ in psi. Its official mks-metric form, taken from ACI 318-19 Appendix C (Equivalence between SI-metric, mks-metric, and U.S. customary units of nonhomogenous equations in the Code), is

$$\beta_1=0.85-\frac{0.05\left(f'_c-280\right)}{70}\qquad\left(280<f'_c<560\ \mathrm{kgf/cm^2}\right)$$

with $\beta_1=0.85$ for $175\le f'_c\le280\ \mathrm{kgf/cm^2}$ and $\beta_1=0.65$ for $f'_c\ge560\ \mathrm{kgf/cm^2}$ (the nominal equivalents of the 2500, 4000, and 8000 psi limits of the U.S. customary table; note that the Appendix C expression gives exactly 0.85 at 280 and exactly 0.65 at 560). Substituting $f'_c$ in $\mathrm{kgf/cm^2}$ directly into the psi expression $0.85-0.05(f'_c-4000)/1000$ would give $\beta_1=1.04$ for $f'_c=280\ \mathrm{kgf/cm^2}$, a stress block deeper than the neutral axis and a physically meaningless result.

- **$E_s$, 20.2.2.2.** The Code permits $E_s=29{,}000{,}000$ psi. This is a material property rather than a nonhomogeneous equation, so it does not appear in Appendix C; its exact conversion is about $2{,}039{,}000\ \mathrm{kgf/cm^2}$. This course uses the rounded value $E_s=2{,}000{,}000\ \mathrm{kgf/cm^2}$, consistent with the yield strain $\varepsilon_y=0.0021$ for $f_y=4200\ \mathrm{kgf/cm^2}$ already used in the flexural strength notebook.

- **$\varepsilon_{ty}$, 21.2.2.1.** For deformed reinforcement, $\varepsilon_{ty}=f_y/E_s$, and for Grade 60 (the $4200\ \mathrm{kgf/cm^2}$ designation) the Code permits $\varepsilon_{ty}=0.002$. This notebook uses $\varepsilon_{ty}=4200/2{,}000{,}000=0.0021$ throughout; the difference is small and moves the balanced point by about half a centimeter of neutral axis depth.

## The Interaction Diagram

### One Strain Profile, One Point

Fix the extreme compression fiber at the crushing strain $\varepsilon_{cu}=0.003$ and let the neutral axis depth $c$ vary. Every value of $c$ defines one linear strain profile through the section, and therefore one set of steel strains, one set of stresses, one set of internal forces, and finally one pair $(P_n, M_n)$ that the section can carry at nominal strength. Sweeping $c$ from very large values (the whole section in compression) to very small ones (almost the whole section in tension) traces the complete interaction diagram.

Left to right: stress block and internal forces, strain profile, and the nominal (blue) and design (orange) diagrams traced point by point. It pauses at the four hand-computed points.</figcaption></figure>

For a rectangular section of width $b$ and total depth $h$, with a layer of compression steel $A'_s$ at depth $d'$ and a layer of tension steel $A_s$ at depth $d$ (both measured from the extreme compression fiber), the procedure for a chosen $c$ is:

**Strains, by compatibility (similar triangles)**

$$\varepsilon'_s=0.003\,\frac{c-d'}{c},\qquad\varepsilon_t=\varepsilon_s=0.003\,\frac{d-c}{c}$$

**Stresses, by the elastic-perfectly plastic law**

$$f'_s=\min\left(E_s\varepsilon'_s,\ f_y\right),\qquad f_s=\min\left(E_s\varepsilon_s,\ f_y\right)$$

**Forces**

$$C_c=0.85f'_c\,a\,b\quad\text{with }a=\beta_1c,\qquad C_s=A'_s\left(f'_s-0.85f'_c\right),\qquad T=A_sf_s$$

The term $-0.85f'_c$ in $C_s$ removes the concrete displaced by the compression bars, which the stress block $C_c$ counts as if it were concrete. It is the same correction that appears as $(A_g-A_{st})$ in $P_o$, and it keeps the diagram consistent at its top end. It applies only while the bars lie inside the stress block ($d'<a$).

**Equilibrium (compression positive)**

$$P_n=C_c+C_s-T$$

**Moment about the plastic centroid**

$$M_n=C_c\left(\bar y-\frac{a}{2}\right)+C_s\left(\bar y-d'\right)+T\left(d-\bar y\right)$$

The moment must be taken about the point through which $P_o$ acts, the plastic centroid, so that the point of pure compression has zero moment. For a section with symmetric reinforcement, as in this notebook, the plastic centroid coincides with the geometric centroid, $\bar y=h/2$.

### Three Regions and the Strength Reduction Factor

The net tensile strain $\varepsilon_t$ in the extreme tension reinforcement classifies every point of the diagram (Table 21.2.2, Fig. R21.2.2a):

- **Compression-controlled**, $\varepsilon_t\le\varepsilon_{ty}$: the tension steel has not yielded when the concrete crushes. Failure is sudden, with little warning. $\phi=0.65$ for tied columns, 0.75 for spiral columns.
- **Transition**, $\varepsilon_{ty}<\varepsilon_t<\varepsilon_{ty}+0.003$: $\phi$ varies linearly with $\varepsilon_t$, for tied columns $\phi=0.65+0.25\left(\varepsilon_t-\varepsilon_{ty}\right)/0.003$.
- **Tension-controlled**, $\varepsilon_t\ge\varepsilon_{ty}+0.003$: the tension steel yields well before crushing, giving the ductile, beam-like behavior of the flexure notebooks. $\phi=0.90$.

The point where $\varepsilon_t=\varepsilon_{ty}$ exactly, that is, where the tension steel yields at the same instant the concrete crushes, is the **balanced point** $(P_{n,bal}, M_{n,bal})$. It separates the compression-controlled branch of the diagram (above it) from the transition and tension-controlled branches (below it), and it is usually close to the point of maximum moment.

### The New Axial-Load Cap of ACI CODE-318-25

ACI CODE-318-25 21.2.2.3 adds one more limit: for nonprestressed members under combined moment and axial compression with

$$0.1f'_cA_g\le P_n\le P_{n,bal}$$

$\phi$ computed from Table 21.2.2 shall not exceed the value obtained by linear interpolation between 0.90 at $0.1f'_cA_g$ and $\phi_{cc}$ (0.65 for tied columns) at $P_{n,bal}$:

$$\phi\le0.90-\left(0.90-\phi_{cc}\right)\frac{P_n-0.1f'_cA_g}{P_{n,bal}-0.1f'_cA_g}$$

<figure style="margin:28px 0; text-align:center;"><img src="assets/07_column_design/fig-03.png" alt="Variation of phi with nominal axial strength according to 21.2.2.3" width="700" style="max-width:100%; border-radius:8px;"><figcaption style="margin-top:10px; color:#475569; font-size:0.9em;">Figure. ACI CODE-318-25 Fig. R21.2.2c, variation of the strength reduction factor with the nominal axial strength in accordance with 21.2.2.3.</figcaption></figure>

The commentary (R21.2.2) explains the motivation: for flanged sections or asymmetric reinforcement, the strain-based $\phi$ alone can let the design axial strength in the transition and tension-controlled regions exceed $\phi_{cc}P_{n,bal}$, an unconservative result (Lequesne and Pincheira 2014). The provision is written for all nonprestressed members in that axial-load range, however, and it applies to the symmetric section of this notebook as well. When $P_{n,bal}<0.1f'_cA_g$, the provision does not apply.

### Design Strength

Every nominal point is reduced by the single $\phi$ that corresponds to it, applied to both coordinates, and the design axial strength is capped at $\phi P_{n,max}$. ACI CODE-318-25 10.5.1.1 then requires, for each factored load combination, $\phi P_n\ge P_u$ and $\phi M_n\ge M_u$ with interaction between them considered, which in practice means: the point $(M_u, P_u)$ must fall inside the design interaction diagram.

## Worked Example: Section and Data

The section analyzed by hand in the rest of this notebook is a rectangular tied column bending about the axis parallel to its short side:

| Property | Value |
|---|---|
| Width $b$ | 30 cm |
| Depth in the direction of bending $h$ | 50 cm |
| Concrete | $f'_c=280\ \mathrm{kgf/cm^2}$ |
| Reinforcement | $f_y=4200\ \mathrm{kgf/cm^2}$, $E_s=2{,}000{,}000\ \mathrm{kgf/cm^2}$ |
| Longitudinal bars | 3 bars of 22 mm on each face perpendicular to the bending direction (6 bars total) |
| Ties | 10 mm |
| Clear cover | 4 cm (the usual metric rounding of the 1-1/2 in. of Table 20.5.1.3.1 for columns not exposed to weather) |

**Derived quantities**

Area of one 22 mm bar: $A_b=3.80\ \mathrm{cm^2}$.

Steel area per face:

$$A_s=A'_s=3\left(3.80\right)=11.40\ \mathrm{cm^2}$$

Total longitudinal steel and gross area:

$$A_{st}=2\left(11.40\right)=22.80\ \mathrm{cm^2},\qquad A_g=30\left(50\right)=1500\ \mathrm{cm^2}$$

Reinforcement ratio, checked against 10.6.1.1:

$$\rho_g=\frac{A_{st}}{A_g}=\frac{22.80}{1500}=0.0152\qquad 0.01\le0.0152\le0.08\ \checkmark$$

Depth of each steel layer, measured from the extreme compression fiber (cover plus tie plus half a bar):

$$d'=4.0+1.0+\frac{2.2}{2}=6.10\ \mathrm{cm},\qquad d=h-d'=50-6.10=43.90\ \mathrm{cm}$$

Material constants:

$$\beta_1=0.85\ \left(f'_c=280\ \mathrm{kgf/cm^2}\right),\qquad 0.85f'_c=0.85\left(280\right)=238\ \mathrm{kgf/cm^2},\qquad \varepsilon_{ty}=\frac{4200}{2{,}000{,}000}=0.0021$$

Plastic centroid (symmetric reinforcement): $\bar y=h/2=25\ \mathrm{cm}$. Lever arms of the steel forces about it:

$$\bar y-d'=25-6.10=18.90\ \mathrm{cm},\qquad d-\bar y=43.90-25=18.90\ \mathrm{cm}$$

The four points computed by hand are: (1) pure compression, (2) the balanced point, (3) the tension-controlled limit, and (4) pure tension, which closes the diagram at its lower end. The three strain profiles that define points 1 to 3 are shown below.

## Point 1: Pure Compression

With the whole section uniformly strained to 0.003, every bar has yielded in compression and the concrete carries $0.85f'_c$ over its net area. There is no moment about the plastic centroid.

**Net concrete area**

$$A_g-A_{st}=1500-22.80=1477.20\ \mathrm{cm^2}$$

**Nominal axial strength**, Eq. (22.4.2.2):

$$P_o=0.85f'_c\left(A_g-A_{st}\right)+f_yA_{st}$$

$$P_o=0.85\left(280\right)\left(1477.20\right)+4200\left(22.80\right)$$

$$P_o=351{,}574+95{,}760=447{,}334\ \mathrm{kgf}\approx447.3\ \mathrm{tf},\qquad M_n=0$$

Concrete carries about 79 percent of $P_o$ and the steel the remaining 21 percent, even though the steel occupies only 1.5 percent of the section.

**Maximum nominal axial strength**, Table 22.4.2.1(a), tied column:

$$P_{n,max}=0.80P_o=0.80\left(447{,}334\right)=357{,}867\ \mathrm{kgf}\approx357.9\ \mathrm{tf}$$

**Design value.** Pure compression is compression-controlled ($\varepsilon_t$ is a compressive strain), so $\phi=0.65$:

$$\phi P_{n,max}=0.65\left(357{,}867\right)=232{,}614\ \mathrm{kgf}\approx232.6\ \mathrm{tf}$$

This is the horizontal plateau at the top of the design diagram. No factored axial load may exceed it, whatever the moment.

## Point 2: The Balanced Point

At the balanced point the tension steel reaches $\varepsilon_{ty}$ at the same instant the concrete reaches 0.003.

**Neutral axis depth, by similar triangles**

$$\frac{c_b}{d}=\frac{\varepsilon_{cu}}{\varepsilon_{cu}+\varepsilon_{ty}}\quad\Rightarrow\quad c_b=\frac{0.003}{0.003+0.0021}\left(43.90\right)=0.5882\left(43.90\right)=25.82\ \mathrm{cm}$$

**Depth of the stress block**

$$a=\beta_1c_b=0.85\left(25.82\right)=21.95\ \mathrm{cm}$$

**Strain and stress in the compression steel**

$$\varepsilon'_s=0.003\,\frac{c_b-d'}{c_b}=0.003\,\frac{25.82-6.10}{25.82}=0.00229$$

Since $0.00229>\varepsilon_{ty}=0.0021$, the compression steel has yielded: $f'_s=4200\ \mathrm{kgf/cm^2}$. The bars lie inside the stress block ($d'=6.10<a=21.95$ cm), so the displaced concrete is subtracted.

**Strain and stress in the tension steel.** By definition of the balanced point, $\varepsilon_t=\varepsilon_{ty}=0.0021$ and $f_s=4200\ \mathrm{kgf/cm^2}$.

**Forces**

$$C_c=0.85f'_c\,a\,b=0.85\left(280\right)\left(21.95\right)\left(30\right)=156{,}723\ \mathrm{kgf}$$

$$C_s=A'_s\left(f'_s-0.85f'_c\right)=11.40\left(4200-238\right)=11.40\left(3962\right)=45{,}167\ \mathrm{kgf}$$

$$T=A_sf_s=11.40\left(4200\right)=47{,}880\ \mathrm{kgf}$$

**Axial force**

$$P_{n,bal}=C_c+C_s-T=156{,}723+45{,}167-47{,}880=154{,}010\ \mathrm{kgf}\approx154.0\ \mathrm{tf}$$

**Moment about the plastic centroid**

$$M_{n,bal}=C_c\left(\bar y-\frac{a}{2}\right)+C_s\left(\bar y-d'\right)+T\left(d-\bar y\right)$$

$$M_{n,bal}=156{,}723\left(25-\frac{21.95}{2}\right)+45{,}167\left(18.90\right)+47{,}880\left(18.90\right)$$

$$M_{n,bal}=156{,}723\left(14.025\right)+853{,}656+904{,}932=2{,}198{,}040+853{,}656+904{,}932$$

$$M_{n,bal}=3{,}956{,}628\ \mathrm{kgf\cdot cm}\approx39.57\ \mathrm{tf\cdot m}$$

The corresponding balanced eccentricity is $e_b=M_{n,bal}/P_{n,bal}=3{,}956{,}628/154{,}010=25.69$ cm. Any load applied with a smaller eccentricity puts the section in the compression-controlled branch; any load with a larger eccentricity, in the transition or tension-controlled branch.

**Design value.** At $\varepsilon_t=\varepsilon_{ty}$ the section is at the compression-controlled limit, $\phi=0.65$ (and 21.2.2.3 gives the same value at $P_n=P_{n,bal}$):

$$\phi P_{n,bal}=0.65\left(154{,}010\right)=100{,}107\ \mathrm{kgf}\approx100.1\ \mathrm{tf}$$

$$\phi M_{n,bal}=0.65\left(3{,}956{,}628\right)=2{,}571{,}808\ \mathrm{kgf\cdot cm}\approx25.72\ \mathrm{tf\cdot m}$$

## Point 3: The Tension-Controlled Limit

This point is the lower end of the transition region: the tension steel strain reaches the tension-controlled limit of Table 21.2.2.

**Target strain**

$$\varepsilon_t=\varepsilon_{ty}+0.003=0.0021+0.003=0.0051$$

**Neutral axis depth**

$$c=\frac{0.003}{0.003+0.0051}\left(43.90\right)=0.3704\left(43.90\right)=16.26\ \mathrm{cm}$$

**Depth of the stress block**

$$a=\beta_1c=0.85\left(16.26\right)=13.82\ \mathrm{cm}$$

**Strain and stress in the compression steel**

$$\varepsilon'_s=0.003\,\frac{c-d'}{c}=0.003\,\frac{16.26-6.10}{16.26}=0.0018745$$

This time $0.0018745<\varepsilon_{ty}=0.0021$: **the compression steel has not yielded**, and its stress comes from the elastic branch:

$$f'_s=E_s\varepsilon'_s=2{,}000{,}000\left(0.0018745\right)=3749\ \mathrm{kgf/cm^2}$$

Assuming $f'_s=f_y$ here, as is tempting after point 2, overstates the compression steel force by about 13 percent ($3962/3511=1.13$). The yield check is never optional.

**Forces**

$$C_c=0.85\left(280\right)\left(13.82\right)\left(30\right)=98{,}675\ \mathrm{kgf}$$

$$C_s=11.40\left(3749-238\right)=11.40\left(3511\right)=40{,}025\ \mathrm{kgf}$$

$$T=11.40\left(4200\right)=47{,}880\ \mathrm{kgf}$$

**Axial force**

$$P_n=98{,}675+40{,}025-47{,}880=90{,}820\ \mathrm{kgf}\approx90.8\ \mathrm{tf}$$

**Moment about the plastic centroid**

$$M_n=98{,}675\left(25-\frac{13.82}{2}\right)+40{,}025\left(18.90\right)+47{,}880\left(18.90\right)$$

$$M_n=98{,}675\left(18.09\right)+756{,}473+904{,}932=1{,}785{,}031+756{,}473+904{,}932$$

$$M_n=3{,}446{,}436\ \mathrm{kgf\cdot cm}\approx34.46\ \mathrm{tf\cdot m}$$

with eccentricity $e=3{,}446{,}436/90{,}820=37.95$ cm.

### The Strength Reduction Factor: Table 21.2.2 Is Not the Last Word

By Table 21.2.2 alone, $\varepsilon_t=\varepsilon_{ty}+0.003$ makes this point tension-controlled, $\phi=0.90$. But ACI CODE-318-25 21.2.2.3 must now be checked:

$$0.1f'_cA_g=0.1\left(280\right)\left(1500\right)=42{,}000\ \mathrm{kgf}$$

$$42{,}000\le P_n=90{,}820\le P_{n,bal}=154{,}010\ \mathrm{kgf}\quad\Rightarrow\quad\text{21.2.2.3 applies}$$

$$\phi\le0.90-\left(0.90-0.65\right)\frac{90{,}820-42{,}000}{154{,}010-42{,}000}=0.90-0.25\left(\frac{48{,}820}{112{,}010}\right)$$

$$\phi\le0.90-0.25\left(0.4359\right)=0.90-0.1090=0.791$$

The interpolated limit governs: $\phi=0.791$.

**Design value**

$$\phi P_n=0.791\left(90{,}820\right)=71{,}839\ \mathrm{kgf}\approx71.8\ \mathrm{tf}$$

$$\phi M_n=0.791\left(3{,}446{,}436\right)=2{,}726{,}131\ \mathrm{kgf\cdot cm}\approx27.26\ \mathrm{tf\cdot m}$$

Under ACI 318-19, the same point would have been taken with $\phi=0.90$: $\phi M_n=31.02$ $\mathrm{tf\cdot m}$. The 2025 provision reduces the design moment at this point by about 12 percent, exactly the effect the commentary warns about. A designer working from a 2019-era spreadsheet or textbook would overestimate this column.

## Point 4: Pure Tension and the Assembled Diagram

### Pure Tension

When the whole section is in tension, the concrete is cracked through and contributes nothing (22.2.2.2); every bar yields in tension. With no prestressing steel, Eq. (22.4.3.1) reduces to

$$P_{nt}=-f_yA_{st}=-4200\left(22.80\right)=-95{,}760\ \mathrm{kgf}\approx-95.8\ \mathrm{tf},\qquad M_n=0$$

A member in pure tension is tension-controlled, $\phi=0.90$:

$$\phi P_{nt}=0.90\left(-95{,}760\right)=-86{,}184\ \mathrm{kgf}\approx-86.2\ \mathrm{tf}$$

### Reading the Diagram

- **The top plateau** of the design curve is $\phi P_{n,max}=232.6$ tf. Between about $M=0$ and 12 $\mathrm{tf\cdot m}$ the column is limited by axial load alone, the accidental-eccentricity cap of Table 22.4.2.1.
- **Above the balanced point**, adding axial load reduces the moment the section can carry: the section is compression-controlled and fails by crushing.
- **Below the balanced point**, adding moderate axial load *increases* the moment capacity, because compression delays the yielding of the tension steel. A column with a small axial load is not "safer" in bending than one with a moderate load, and a load combination with minimum gravity load and maximum lateral moment (such as $0.9D$ plus lateral load) can govern the design.
- **The pure-flexure point**, where the curves cross $P=0$, is the beam strength of the section. From the same sweep, $M_n\approx19.4$ $\mathrm{tf\cdot m}$ at $c\approx6.8$ cm, well into the tension-controlled region, $\phi M_n\approx17.5$ $\mathrm{tf\cdot m}$. Computing it by hand requires solving the quadratic in $c$ produced by an elastic compression steel, the same procedure as the doubly reinforced beams of the flexural strength notebook.
- **The kink in the design curve near points 2 and 3** is the signature of 21.2.2.3: between $P_n=42$ and 154 tf, $\phi$ is governed by the axial load instead of the strain, and the design curve bends back toward lower moments compared with the 2019 rule.
- **A factored load pair $(M_u, P_u)$** is acceptable when it plots inside the orange curve (10.5.1.1). For example, $P_u=120$ tf with $M_u=20$ $\mathrm{tf\cdot m}$ lies inside; the same $P_u$ with $M_u=27$ $\mathrm{tf\cdot m}$ lies outside, even though the nominal curve would accept it with ample margin.

## Summary

A column carries axial force and bending moment simultaneously, and its strength is the set of all $(P_n, M_n)$ pairs its cross section can develop: the interaction diagram. Its history runs from the confinement tests of Consid&egrave;re and Richart, through the ACI column investigation that made the column the first member designed in substance by ultimate strength, to Whitney's stress block and Hognestad's combined-loading tests that made the diagram computable. Each point of the diagram follows from one strain profile with 0.003 at the extreme compression fiber: compatibility gives the steel strains, the elastic-plastic law gives the stresses (with a yield check every time), equilibrium gives $P_n=C_c+C_s-T$, and moments about the plastic centroid give $M_n$. For the 30 x 50 cm example column, pure compression gives $P_o=447.3$ tf, capped at $P_{n,max}=0.80P_o=357.9$ tf; the balanced point gives 154.0 tf and 39.57 $\mathrm{tf\cdot m}$; the tension-controlled limit gives 90.8 tf and 34.46 $\mathrm{tf\cdot m}$ with an elastic compression steel; and pure tension gives -95.8 tf. Design strengths use $\phi=0.65$ in the compression-controlled region, 0.90 in the tension-controlled region, and a linear transition between them, now further capped by ACI CODE-318-25 21.2.2.3 whenever $0.1f'_cA_g\le P_n\le P_{n,bal}$, a new provision that lowers $\phi$ at the tension-controlled limit of the example from 0.90 to 0.791. The only unit conversion needed is $\beta_1$, whose technical-metric form comes from ACI 318-19 Appendix C.

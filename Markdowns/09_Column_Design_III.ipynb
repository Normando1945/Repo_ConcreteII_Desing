## From the Section to the Member

Columns I (notebook 07) treated the column as a cross section under axial load and flexure, and Columns II (notebook 08) designed the same section for shear. Its interaction diagram answers one question: which pairs $(P_n, M_n)$ can the section develop at nominal strength. It answers it under two silent assumptions, and this notebook removes both.

The first assumption is that the moment acting on the section is the one computed by the structural analysis. That is true only if the analysis was made on the deformed shape of the structure. A first-order analysis, the kind used for almost every frame, writes equilibrium on the undeformed geometry. But a column under axial load $P$ that bends by a lateral deflection $\Delta$ carries an additional moment $P\Delta$, and that moment produces more deflection, which produces more moment. In a short, stocky column this loop dies out after a negligible increase. In a slender column it does not, and the moment the section must resist can be substantially larger than the first-order value. ACI CODE-318-25 calls these **second-order effects** or **slenderness effects**, and Chapter 6 gives the rules to decide when they matter and how to include them.

The second assumption is that the column bends about one principal axis only. Corner columns, columns in irregular frames, and any column under the minimum eccentricity in both directions bend about both axes at the same time. The interaction diagram then becomes an interaction surface, and the neutral axis is no longer parallel to a side of the section.

Both problems are solved here on the same 30 x 50 cm tied column of notebook 07, so that the interaction diagram already computed by hand becomes the strength against which the magnified and the biaxial demands are checked. The transverse reinforcement of this column, ties and their spacing, was designed for shear in notebook 08 and is not revisited here.

## A Short History of Column Stability

### Euler and the Elastic Column

In 1744, in an appendix on elastic curves to his treatise on the calculus of variations, Leonhard Euler found the load at which a perfectly straight, pin-ended elastic column stops being straight:

$$P_c=\frac{\pi^2EI}{\ell^2}$$

Below $P_c$ the straight shape is the only equilibrium; at $P_c$ a bent shape becomes possible. Euler's load contains no material strength at all, only stiffness and length, and that is the essential idea of slenderness: a long column can fail at a stress well below the crushing strength of its material. The effective length $k\ell$ that the Code uses today is Euler's length generalized to other end conditions: $k=1$ for pinned ends, 0.5 for fixed ends without sway, and more than 1 when the ends can translate.

### From Buckling to Amplification

Euler's formula describes an ideal column that is perfectly straight and loaded exactly on its axis. Real columns have end moments, initial crookedness, and inelastic materials, and for them buckling is not a sudden branching but a growth of deflection and moment that starts with the first increment of load. The tangent-modulus work of Engesser and Consid&egrave;re (1889) and Shanley's inelastic column theory (1947) explained the inelastic case for metals. For design, the decisive observation came from the elastic beam-column: the moment at mid-height of a column with end moments is approximately the first-order moment multiplied by $1/(1-P/P_c)$. A single factor, depending only on how close the load is to the buckling load, captures the whole $P$-$\Delta$ loop.

### The Moment Magnifier Enters the Code

Until the 1960s, ACI codes handled long columns with an empirical reduction factor applied to the strength of a short column. The research of MacGregor, Breen, and Pfrang (1970) replaced it with the moment magnifier method: compute $P_c$ with a stiffness $EI$ that accounts for cracking and creep, magnify the first-order moment, and design the section for the magnified moment. The method was adopted in the 1971 Code and, with refinements, it is still 6.6.4 of ACI CODE-318-25. The same study is the origin of the slenderness limits of 6.2.5.1, which the commentary describes as the slenderness at which the magnifier would increase the moment by about 5 percent.

### Biaxial Bending

For biaxial bending, an exact solution by strain compatibility was out of reach of hand calculation for most of the twentieth century. Boris Bresler (1960) proposed two approximations that are still taught: the reciprocal load method, which estimates the axial strength under biaxial eccentricity from three uniaxial strengths, and the load contour method, later developed into a design procedure by Parme, Nieves, and Gouwens (1966). Today the full surface is computed numerically, but the approximations remain the standard hand check, and ACI CODE-318-25 commentary points to the review of Furlong, Hsu, and Mirza (2004) for slenderness under biaxial bending.

### A Sign Convention Changed in 2014

One detail matters when reading older books: until the 2011 Code, the ratio of end moments $M_1/M_2$ was positive for single curvature. Since the 2014 Code it is **negative for single curvature and positive for double curvature** (commentary R6.2.5.1), and the equations for the slenderness limit and for $C_m$ changed sign accordingly. This notebook uses the current convention throughout.

## Slenderness and Second-Order Effects

### Member Curvature ($P$-$\delta$) and Story Sway ($P$-$\Delta$)

Two kinds of second-order effects are distinguished, because they arise from different deflections and are treated by different parts of the Code:

- **$P$-$\delta$, member curvature.** The column deflects between its ends, relative to the chord that joins them. The extra moment $P\delta$ is largest near mid-height. It exists in every column, braced or not.
- **$P$-$\Delta$, story sway.** In a frame that is free to translate laterally, the top of the column moves relative to the bottom by the story drift $\Delta$. The extra moment $P\Delta$ acts at the column ends and adds to the moments from lateral loads.

### Nonsway and Sway

Unless slenderness may be neglected under 6.2.5.1, ACI CODE-318-25 6.6.4.1 requires columns and stories to be designated **nonsway** (braced) or **sway** (unbraced). 6.2.5.1 itself gives a direct rule: if the bracing elements that resist lateral movement of a story have a total stiffness of at least 12 times the gross lateral stiffness of the columns in the direction considered, the columns of that story may be considered braced against sidesway. A story is nonsway when its lateral stiffness, usually provided by structural walls, is large enough that sway does not change the column strength appreciably. 6.6.4.3 permits a story to be treated as nonsway if either (a) second-order effects increase the column end moments by no more than 5 percent of the first-order end moments, or (b) the stability index does not exceed 0.05:

$$Q=\frac{\sum P_u\,\Delta_o}{V_{us}\,\ell_c}\le0.05\qquad\text{(6.6.4.4.1)}$$

where $\sum P_u$ and $V_{us}$ are the total factored vertical load and the factored story shear, $\Delta_o$ is the first-order relative lateral deflection between the top and bottom of the story due to $V_{us}$, and $\ell_c$ is the column length measured center to center of the joints. $Q$ is the ratio of the second-order moment $\sum P_u\Delta_o$ to the first-order story moment $V_{us}\ell_c$; when it is below 5 percent, sway amplification can be ignored.

### When Can Slenderness Be Neglected?

6.2.5.1 lets the designer skip second-order effects when the slenderness ratio is small enough:

$$\text{(a) sway:}\quad\frac{k\ell_u}{r}\le22\qquad\qquad\text{(b) nonsway:}\quad\frac{k\ell_u}{r}\le34+12\left(\frac{M_1}{M_2}\right)\quad\text{and}\quad\frac{k\ell_u}{r}\le40$$

with $M_1/M_2$ negative for single curvature and positive for double curvature. In these limits:

- $\ell_u$ is the unsupported length, the clear distance between floor slabs, beams, or other members that provide lateral support in the direction considered.
- $r$ is the radius of gyration, which 6.2.5.2 permits to be taken as $\sqrt{I_g/A_g}$, as 0.30 times the dimension in the direction considered for rectangular columns, or as 0.25 times the diameter for circular ones.
- $k$ is the effective length factor. For nonsway members, 6.6.4.4.3 permits $k=1.0$; a smaller value can be justified with the alignment charts of Fig. R6.2.5.1, which give $k$ from the relative stiffness $\Psi$ of the columns and beams at each end.

Two features of the limits deserve attention. In a nonsway frame, double curvature ($M_1/M_2>0$) raises the limit, because the maximum moment then stays at the ends, where deflection cannot add to it; single curvature lowers it to as little as 22. And the limit depends on the direction considered: a rectangular column has two values of $r$, and the smaller dimension is usually the critical one.

### The Overall Procedure, and an Upper Bound

When slenderness cannot be neglected, 6.2.5.3 requires the column, the restraining beams, and the supports to be designed for moments that include second-order effects, by the moment magnification method of 6.6.4, by an elastic second-order analysis (6.7), or by an inelastic second-order analysis (6.8). Whatever the method, the moment including second-order effects shall not exceed **1.4 times** the first-order moment. Beyond that ratio the structure is too flexible, and the Code requires it to be stiffened rather than the column to be reinforced.

## The Moment Magnification Method: Nonsway Frames

### Critical Buckling Load

The magnifier compares the factored axial load with Euler's load of the column, 6.6.4.4.2:

$$P_c=\frac{\pi^2\left(EI\right)_{eff}}{\left(k\ell_u\right)^2}$$

The stiffness cannot be the gross elastic stiffness $E_cI_g$: at the load levels of interest the concrete is cracked, its stress-strain response is nonlinear, and sustained load adds creep. 6.6.4.4.4 gives three expressions:

$$\text{(a)}\ \left(EI\right)_{eff}=\frac{0.4E_cI_g}{1+\beta_{dns}}\qquad\text{(b)}\ \left(EI\right)_{eff}=\frac{0.2E_cI_g+E_sI_{se}}{1+\beta_{dns}}\qquad\text{(c)}\ \left(EI\right)_{eff}=\frac{E_cI}{1+\beta_{dns}}$$

where $\beta_{dns}$ is the ratio of the maximum factored sustained axial load to the maximum factored axial load of the same load combination, $I_{se}$ is the moment of inertia of the reinforcement about the centroidal axis of the section, and $I$ in (c) comes from Table 6.6.3.1.1(b). Expression (a) is the simplest and does not require the reinforcement to be known, which makes it the usual choice in preliminary design; (b) is more accurate when the reinforcement is known. The commentary notes that with $\beta_{dns}=0.6$, (a) becomes $0.25E_cI_g$.

### The Magnifier

The design moment of a column in a nonsway frame is the larger end moment $M_2$ magnified, 6.6.4.5.1 and 6.6.4.5.2:

$$M_c=\delta M_2\qquad\qquad\delta=\frac{C_m}{1-\dfrac{P_u}{0.75P_c}}\ge1.0$$

The factor 0.75 is a stiffness reduction factor that accounts for the variability of $(EI)_{eff}$ in a single column (R6.6.4.5.2). As $P_u$ approaches $0.75P_c$, the denominator tends to zero and the magnifier grows without bound: this is the buckling load, approached through amplification rather than reached by sudden branching.

### The Factor Cm

The magnifier was derived for a column bent by equal and opposite end moments, that is, uniform moment in single curvature, where the maximum first-order moment and the maximum deflection coincide at mid-height. For other moment diagrams, $C_m$ converts the actual end moments into an equivalent uniform moment $C_mM_2$, 6.6.4.5.3:

$$C_m=0.6-0.4\,\frac{M_1}{M_2}\qquad\text{(columns without transverse loads between supports)}$$

with $M_1$ the end moment of smaller absolute value. Uniform single curvature ($M_1/M_2=-1$) gives $C_m=1.0$; one end moment equal to zero gives 0.6; symmetric double curvature ($M_1/M_2=+1$) gives 0.2. For columns with transverse loads between supports, $C_m=1.0$. Because $\delta$ cannot be less than 1.0, a small $C_m$ simply means that the end moment $M_2$ itself governs: the maximum moment stays at the end of the column, where member curvature adds nothing.

### Minimum Moment

A column with very small computed moments is still designed for an accidental eccentricity, 6.6.4.5.4: $M_2$ shall be at least

$$M_{2,min}=P_u\left(1.5+0.03h\right)\qquad\text{(}h\text{ and the eccentricity in cm)}$$

about each axis separately, not about both at once. When $M_{2,min}$ governs, $C_m$ is taken as 1.0 or computed from the actual end moments.

### Units: the Values That Need Conversion

Two quantities in this notebook are not dimensionally homogeneous in the Code and take their technical-metric form from ACI 318-19 Appendix C:

| Quantity | U.S. customary (in., psi) | Technical metric (cm, $\mathrm{kgf/cm^2}$) |
|---|---|---|
| Modulus of elasticity of normalweight concrete, 19.2.2.1 | $E_c=57{,}000\sqrt{f'_c}$ | $E_c=15{,}100\sqrt{f'_c}$ |
| Minimum moment, 6.6.4.5.4 | $M_{2,min}=P_u\left(0.6+0.03h\right)$ | $M_{2,min}=P_u\left(1.5+0.03h\right)$ |

The constant 15,100 is not a linear conversion of 57,000, for the same reason as the $\sqrt{f'_c}$ coefficients of the two-way shear notebook: the square root changes the units of the constant. All other expressions used here, $Q$, $P_c$, $\delta$, $C_m$, and the slenderness ratios, are dimensionally homogeneous.

## Sway Frames, Briefly

In a sway story the end moments are split into the part caused by loads that produce no appreciable sway, $M_{ns}$ (usually gravity), and the part caused by loads that do, $M_s$ (usually lateral). Only the sway part is magnified, 6.6.4.6.1:

$$M_1=M_{1ns}+\delta_sM_{1s}\qquad\qquad M_2=M_{2ns}+\delta_sM_{2s}$$

6.6.4.6.2 offers three ways to compute $\delta_s$:

$$\text{(a)}\ \delta_s=\frac{1}{1-Q}\ge1\qquad\text{(b)}\ \delta_s=\frac{1}{1-\dfrac{\sum P_u}{0.75\sum P_c}}\ge1\qquad\text{(c) elastic second-order analysis}$$

where $\sum P_u$ runs over all the columns of the story and $\sum P_c$ over all its sway-resisting columns, because a story sways as a unit: a stiff column cannot buckle sideways without dragging the others with it. Each $P_c$ is computed with $k$ for sway members, which shall be at least 1.0 (6.6.4.4.3), and with the $(EI)_{eff}$ of 6.6.4.4.4 using $\beta_{ds}$ in place of $\beta_{dns}$ (6.6.4.6.2). $\beta_{ds}$ is the ratio of the sustained to the total factored story shear, usually zero for wind or earthquake. If $\delta_s$ exceeds 1.5, only (b) or (c) are permitted. After the end moments are magnified for sway, second-order effects along the length of the column must still be considered (6.6.4.6.4): it is permitted to do so with the nonsway procedure of 6.6.4.5, with $C_m$ computed from the end moments $M_1$ and $M_2$ of 6.6.4.6.1.

## Worked Example 1: A Slender Column in a Braced Frame

### Data

The column is the 30 x 50 cm tied column of notebook 07: $f'_c=280\ \mathrm{kgf/cm^2}$, $f_y=4200\ \mathrm{kgf/cm^2}$, three 22 mm bars on each 30 cm face, bending about the strong axis ($h=50$ cm). It stands in the ground story of a building with a high lobby:

| Item | Value |
|---|---|
| Unsupported length $\ell_u$ (both directions, no intermediate bracing) | 6.0 m |
| Story height center to center of joints $\ell_c$ | 6.5 m |
| Factored axial load $P_u$ | 120 tf |
| Factored sustained axial load in the same combination | 72 tf |
| First-order end moments about the strong axis, single curvature | $M_2=16$ $\mathrm{tf\cdot m}$ (top), $M_1=10$ $\mathrm{tf\cdot m}$ (bottom) |
| Story: $\sum P_u$, $V_{us}$, $\Delta_o$ from the first-order analysis | 1500 tf, 75 tf, 1.4 cm |

$P_u=120$ tf is the load of the design check of notebook 07, so the design strength at this axial load is already known: $\phi M_n=24.43$ $\mathrm{tf\cdot m}$ about the strong axis.

### Step 1: Nonsway or Sway?

$$Q=\frac{\sum P_u\,\Delta_o}{V_{us}\,\ell_c}=\frac{1500\left(1.4\right)}{75\left(650\right)}=\frac{2100}{48{,}750}=0.0431\le0.05\ \checkmark$$

The story may be analyzed as nonsway (6.6.4.3(b)).

### Step 2: Can Slenderness Be Neglected?

$$r=0.30h=0.30\left(50\right)=15\ \mathrm{cm},\qquad k=1.0\ \text{(6.6.4.4.3)},\qquad\frac{k\ell_u}{r}=\frac{1.0\left(600\right)}{15}=40.0$$

$$\frac{M_1}{M_2}=-\frac{10}{16}=-0.625\ \text{(single curvature)},\qquad34+12\left(-0.625\right)=26.5$$

$$\frac{k\ell_u}{r}=40.0>26.5\quad\Rightarrow\quad\text{slenderness effects must be considered}$$

### Step 3: Effective Stiffness and Critical Load

$$E_c=15{,}100\sqrt{f'_c}=15{,}100\sqrt{280}=15{,}100\left(16.733\right)=252{,}671\ \mathrm{kgf/cm^2}$$

$$I_g=\frac{bh^3}{12}=\frac{30\left(50\right)^3}{12}=312{,}500\ \mathrm{cm^4},\qquad\beta_{dns}=\frac{72}{120}=0.60$$

$$\left(EI\right)_{eff}=\frac{0.4E_cI_g}{1+\beta_{dns}}=\frac{0.4\left(252{,}671\right)\left(312{,}500\right)}{1+0.60}=\frac{3.1584\times10^{10}}{1.60}=1.9740\times10^{10}\ \mathrm{kgf\cdot cm^2}$$

$$P_c=\frac{\pi^2\left(EI\right)_{eff}}{\left(k\ell_u\right)^2}=\frac{9.8696\left(1.9740\times10^{10}\right)}{\left(600\right)^2}=541{,}181\ \mathrm{kgf}\approx541.2\ \mathrm{tf}$$

As a check on the choice of expression, (b) with $I_{se}=2\left(11.40\right)\left(18.90\right)^2=8144\ \mathrm{cm^4}$ gives $\left(0.2E_cI_g+E_sI_{se}\right)/1.60=\left(1.5792+1.6289\right)\times10^{10}/1.60=2.0050\times10^{10}\ \mathrm{kgf\cdot cm^2}$, only 1.6 percent stiffer. For this section, with a moderate steel ratio, the simple expression is adequate.

### Step 4: Cm and the Magnifier

$$C_m=0.6-0.4\,\frac{M_1}{M_2}=0.6-0.4\left(-0.625\right)=0.85$$

$$\frac{P_u}{0.75P_c}=\frac{120.0}{0.75\left(541.2\right)}=\frac{120.0}{405.9}=0.2956$$

$$\delta=\frac{C_m}{1-\dfrac{P_u}{0.75P_c}}=\frac{0.85}{1-0.2956}=\frac{0.85}{0.7044}=1.207\ge1.0$$

### Step 5: Minimum Moment

$$M_{2,min}=P_u\left(1.5+0.03h\right)=120\left(1.5+0.03\left(50\right)\right)=120\left(3.0\right)=360\ \mathrm{tf\cdot cm}=3.60\ \mathrm{tf\cdot m}<M_2=16\ \mathrm{tf\cdot m}$$

The computed moment governs.

### Step 6: Magnified Moment and Checks

$$M_c=\delta M_2=1.207\left(16\right)=19.31\ \mathrm{tf\cdot m}$$

Second-order limit of 6.2.5.3:

$$M_c=19.31\le1.4M_2=1.4\left(16\right)=22.40\ \mathrm{tf\cdot m}\ \checkmark$$

Strength, against the design strength at $P_u=120$ tf computed by hand in notebook 07:

$$M_c=19.31\le\phi M_n=24.43\ \mathrm{tf\cdot m}\ \checkmark\qquad\text{demand/capacity}=\frac{19.31}{24.43}=0.79$$

Slenderness has raised the moment by 21 percent and the demand/capacity ratio from $16/24.43=0.65$ to 0.79. The column is adequate about its strong axis.

## The Other Axis, and How Fast the Magnifier Grows

### Step 7: The Weak Axis Under the Minimum Moment

The structural analysis gives no moments about the weak axis, but 6.6.4.5.4 still requires the minimum moment about each axis separately, and the weak axis has the smaller radius of gyration. With no computed end moments, $C_m=1.0$, and for the slenderness limit the conservative single-curvature value $M_1/M_2=-1$ is taken, which gives a limit of 22.

$$r_y=0.30\left(30\right)=9\ \mathrm{cm},\qquad\frac{k\ell_u}{r_y}=\frac{1.0\left(600\right)}{9}=66.7>22$$

$$I_{gy}=\frac{50\left(30\right)^3}{12}=112{,}500\ \mathrm{cm^4},\qquad\left(EI\right)_{eff,y}=\frac{0.4\left(252{,}671\right)\left(112{,}500\right)}{1.60}=7.1064\times10^{9}\ \mathrm{kgf\cdot cm^2}$$

$$P_{c,y}=\frac{9.8696\left(7.1064\times10^{9}\right)}{\left(600\right)^2}=194{,}825\ \mathrm{kgf}\approx194.8\ \mathrm{tf}$$

$$M_{2,min}=120\left(1.5+0.03\left(30\right)\right)=120\left(2.4\right)=288\ \mathrm{tf\cdot cm}=2.88\ \mathrm{tf\cdot m}$$

$$\delta_y=\frac{1.0}{1-\dfrac{120.0}{0.75\left(194.8\right)}}=\frac{1.0}{1-0.8212}=5.593,\qquad M_{c,y}=5.593\left(2.88\right)=16.11\ \mathrm{tf\cdot m}$$

The magnified moment is 5.6 times the first-order one, far beyond the 1.4 limit of 6.2.5.3, and also beyond the design strength about the weak axis at $P_u=120$ tf, $\phi M_{ny}=11.37$ $\mathrm{tf\cdot m}$ (from the weak-axis design diagram plotted in the figure above, computed with the procedure of notebook 07 with $b$ and $h$ interchanged; notebook 07 itself draws only the strong-axis diagram). **The column fails about the axis that carries no computed moment.** At $P_u/(0.75P_c)=0.82$ it is close to buckling sideways, and no amount of longitudinal steel can fix a stability problem efficiently.

The remedy is structural: a tie beam at mid-height in the weak direction halves the unsupported length about that axis.

$$\frac{k\ell_u}{r_y}=\frac{1.0\left(300\right)}{9}=33.3>22,\qquad P_{c,y}=194.8\left(\frac{600}{300}\right)^2=779.3\ \mathrm{tf}$$

$$\delta_y=\frac{1.0}{1-\dfrac{120.0}{0.75\left(779.3\right)}}=\frac{1.0}{1-0.2053}=1.258,\qquad M_{c,y}=1.258\left(2.88\right)=3.62\ \mathrm{tf\cdot m}$$

About the weak axis the only first-order moment is the minimum moment, so $M_{2,min}$ plays the role of $M_2$ in the 1.4 check of 6.2.5.3. The Code does not state this explicitly; it is the consistent reading, since 6.6.4.5.4 makes $M_{2,min}$ the moment that is magnified.

$$M_{c,y}=3.62\le1.4\left(2.88\right)=4.03\ \checkmark\qquad M_{c,y}=3.62\le\phi M_{ny}=11.37\ \mathrm{tf\cdot m}\ \checkmark$$

The lesson is general: slenderness is governed by the smaller dimension and the longer unbraced length, not by the direction of the computed moments.

## Biaxial Bending

### From a Curve to a Surface

When a column carries moments $M_{ux}$ and $M_{uy}$ about both principal axes, the axial load acts at a point with eccentricities $e_y=M_{ux}/P_u$ and $e_x=M_{uy}/P_u$, off both axes. The neutral axis is then inclined, and its angle is not known in advance: it is whatever angle makes the resultant of the internal forces pass through the load point. The strength of the section is no longer a curve but a surface in the space $(M_{nx}, M_{ny}, P_n)$; the uniaxial diagrams of notebook 07 are two of its meridian sections. A horizontal cut of the surface at a given axial force is a **load contour**: all the moment pairs the section can carry at that axial force.

ACI CODE-318-25 contains no simplified method for biaxial bending. The strength follows from the same assumptions of 22.2 used for the uniaxial diagram, applied with an inclined neutral axis, which is routine for software but long by hand. Approximate methods from the technical literature are used for hand checks; the commentary (R6.2.5.3) refers to Furlong et al. (2004) for a review.

### The Exact Solution: An Inclined Neutral Axis

The assumptions of 22.2 say nothing about the direction of bending: plane sections remain plane, the concrete crushes at 0.003, the stress block has depth $a=\beta_1c$. They apply unchanged when the neutral axis is inclined. With the origin at a corner of the section and the most compressed corner at the opposite one, a single evaluation goes as follows:

1. Choose the angle $\theta$ of the neutral axis with the $x$ axis and its depth $c$, measured perpendicular to it from the most compressed corner.
2. The stress block is the part of the section within a distance $a=\beta_1c$ of that corner, measured in the same direction. Its boundary is a line parallel to the neutral axis, and depending on $\theta$ and $a$ the block is a triangle, a trapezoid, or the whole section minus a triangle. Compute its area $A$ and centroid $(\bar x,\bar y)$: $C_c=0.85f'_cA$.
3. Each bar $i$ has a depth $d_i$ measured in the same direction, a strain $\varepsilon_i=0.003\left(c-d_i\right)/c$, a stress $f_i=E_s\varepsilon_i$ limited to $\pm f_y$, reduced by $0.85f'_c$ when the bar lies inside the block in compression, and a force $F_i=A_if_i$.
4. With the centroid of the section at $(x_c,y_c)$:

$$P_n=C_c+\sum F_i,\qquad M_{nx}=C_c\left(\bar y-y_c\right)+\sum F_i\left(y_i-y_c\right),\qquad M_{ny}=C_c\left(\bar x-x_c\right)+\sum F_i\left(x_i-x_c\right)$$

5. The pair $(\theta,c)$ is the solution for a load with eccentricities $e_x$, $e_y$ when $M_{nx}/P_n=e_y$ and $M_{ny}/P_n=e_x$.

Two unknowns and two nonlinear equations: an iteration that software solves in milliseconds and a person would solve in an afternoon. One evaluation, however, is pure arithmetic, and it is the way to check what a program reports. Example 2 does exactly that. Note also that the neutral axis is generally **not** perpendicular to the line from the centroid to the load: the section bends more easily about its weak axis, so the curvature about that axis, and with it the tilt of the neutral axis, is larger than the ratio of the two moments alone would suggest.

### Bresler's Reciprocal Load Method

Bresler observed that the surface $1/P_n$ as a function of the two eccentricities is smooth and nearly flat over the range of practical interest. It can then be approximated by the plane through three points that are easy to compute: $1/P_o$ at zero eccentricity, $1/P_{nx}$ with eccentricity $e_y$ only, and $1/P_{ny}$ with eccentricity $e_x$ only. The plane gives:

$$\frac{1}{P_n}\approx\frac{1}{P_{nx}}+\frac{1}{P_{ny}}-\frac{1}{P_o}$$

where $P_{nx}$ is the nominal axial strength when the load acts at eccentricity $e_y$ alone (bending about the $x$ axis), and $P_{ny}$ the strength at eccentricity $e_x$ alone. Three points of practice:

- The method is meant for sections where axial compression governs. The usual rule is not to use it below $P_n=0.1f'_cA_g$, where the surface $1/P_n$ is far from a plane. Figure 5 adds a finer picture for this column: above the balanced load the estimate is on the safe side, by a margin that grows from almost zero near the balanced load to 6 or 7 percent at high axial load; between $0.1f'_cA_g$ and the balanced load it can be up to about 8 percent above it, on the unsafe side; below $0.1f'_cA_g$ it is simply inaccurate. In the transition zone a margin, or a second method, is prudent.
- $P_{nx}$ and $P_{ny}$ are found on the uniaxial diagrams at a fixed eccentricity, that is, where a ray from the origin with slope $1/e$ cuts the diagram. By hand, this needs a short iteration on $c$.
- The usual design check is $\phi P_n\ge P_u$, with one value of $\phi$ for the three terms; it is valid when the points involved fall in the same region of Table 21.2.2, which must be verified.

### The Load Contour Method

The second approximation works on a horizontal cut of the surface. At a given axial force, the load contour is approximated by

$$\left(\frac{M_{nx}}{M_{nox}}\right)^{\alpha}+\left(\frac{M_{ny}}{M_{noy}}\right)^{\alpha}=1$$

where $M_{nox}$ and $M_{noy}$ are the uniaxial nominal moment strengths at that axial force, read from the two uniaxial diagrams, and $\alpha$ depends on the section, its reinforcement, the materials, and the axial force. As a design check, at the factored axial load $P_u$:

$$\left(\frac{M_{ux}}{\phi M_{nox}}\right)^{\alpha}+\left(\frac{M_{uy}}{\phi M_{noy}}\right)^{\alpha}\le1$$

- $\alpha=1$ is a straight line between the two uniaxial strengths. Because real contours are convex, it is conservative for ordinary sections, often by a wide margin.
- Bresler (1960) reported exponents roughly between 1.15 and 1.55 for the sections he studied, and $\alpha=1.5$ is a common choice for rectangular columns.
- Parme, Nieves, and Gouwens (1966), in the method published by the Portland Cement Association, write the exponent as $\alpha=\log0.5/\log\beta$, where $\beta$ is the fraction of both uniaxial strengths at which the contour crosses the diagonal $M_{nx}/M_{nox}=M_{ny}/M_{noy}$. Typical values of $\beta$ lie between about 0.55 and 0.70; $\beta=0.65$, that is $\alpha=1.61$, is a common design value.

For the example column the exact contours give $\beta$ between 0.625 and 0.683 (Figure 6), that is $\alpha$ between 1.48 and 1.82: $\alpha=1.5$ is accurate near the balanced load and conservative elsewhere.

The cross is the demand of example 3, normalized by the design strengths.</figcaption></figure>

One caution about $\phi$. At low axial load each uniaxial point is often tension-controlled, with $\phi=0.90$. Under biaxial bending, however, the extreme tension steel is a single corner bar, closer to the neutral axis, and its strain can be smaller than in either uniaxial case. Example 3 shows $\phi$ falling from 0.90 to about 0.81. Unless the strain of the corner bar is checked, $\phi$ taken from the uniaxial diagrams may be unconservative.

### Choosing a Method

| Axial force (estimated as $P_u/\phi$) | Hand method | Remarks |
|---|---|---|
| Above the balanced load | Bresler's reciprocal load | Simple and slightly conservative |
| Between $0.1f'_cA_g$ and the balanced load | Bresler with a margin, or load contour with $\alpha=1.5$ | Bresler may be a few percent unconservative |
| Below $0.1f'_cA_g$ | Load contour: $\alpha=1$ (conservative) or $\alpha$ from $\beta$ | Check the strain of the corner bar for $\phi$ |
| Any, to check software | One exact evaluation by hand | The verification of example 2 |

Some textbooks also present methods that replace the biaxial eccentricity by an equivalent, larger uniaxial one. They are not used in this course.

## Worked Example 3: A Lightly Loaded Corner Column

### Data

The corner column of example 2, in the top story, where the axial load is small: same section, $\ell_u=3.0$ m, braced, double curvature about both axes with $M_1/M_2=+0.5$, so slenderness may again be neglected. The governing combination gives

$$P_u=25\ \mathrm{tf},\qquad M_{ux}=14\ \mathrm{tf\cdot m},\qquad M_{uy}=6\ \mathrm{tf\cdot m}$$

### Step 1: Which Method?

With $\phi=0.90$ the nominal axial force is about $25/0.90=27.8$ tf, below $0.1f'_cA_g=0.1\left(280\right)\left(1500\right)=42{,}000$ kgf $=42.0$ tf. Bresler's method is outside its range. For the record, with $e_y=14/25=0.56$ m and $e_x=6/25=0.24$ m the uniaxial strengths at fixed eccentricity are $P_{nx}=51.1$ tf and $P_{ny}=63.2$ tf, and the reciprocal formula would give $P_n=30.2$ tf against an exact 34.6 tf, 13 percent off. The load contour method is used instead, at the fixed axial force $P_u$.

### Step 2: Strong-Axis Strength at $P_u$

Assume tension control, $\phi=0.90$, so $P_n=25{,}000/0.90=27{,}778$ kgf. With the compression steel elastic and inside the block, and the tension steel yielded, equilibrium multiplied by $c$ gives a quadratic, exactly as for the pure-flexure point of notebook 07 (with $E_s\varepsilon_{cu}=2{,}000{,}000\left(0.003\right)=6000\ \mathrm{kgf/cm^2}$):

$$0.85f'_cb\beta_1c^2+\left[A'_s\left(6000-0.85f'_c\right)-A_sf_y-P_n\right]c-A'_s\left(6000\right)d'=0$$

$$6069c^2+\left[11.40\left(5762\right)-11.40\left(4200\right)-27{,}778\right]c-11.40\left(6000\right)\left(6.10\right)=0$$

$$6069c^2-9971.2c-417{,}240=0\quad\Rightarrow\quad c=\frac{9971.2+\sqrt{9971.2^2+4\left(6069\right)\left(417{,}240\right)}}{2\left(6069\right)}=9.15\ \mathrm{cm}$$

$$a=0.85\left(9.15\right)=7.78\ \mathrm{cm},\qquad \varepsilon'_s=0.003\,\frac{9.15-6.10}{9.15}=0.00100\ \Rightarrow\ f'_s=2000\ \mathrm{kgf/cm^2}\ \text{(elastic)}$$

$$C_c=238\left(7.78\right)\left(30\right)=55{,}549,\qquad F'_s=11.40\left(2000-238\right)=20{,}087,\qquad T=11.40\left(4200\right)=47{,}880\ \mathrm{kgf}$$

$$P_n=55{,}549+20{,}087-47{,}880=27{,}756\ \mathrm{kgf}\approx27{,}778\ \checkmark\ \text{(rounding of }c\text{)}$$

$$M_{nox}=55{,}549\left(25-\frac{7.78}{2}\right)+20{,}087\left(18.90\right)+47{,}880\left(18.90\right)=2{,}457{,}215\ \mathrm{kgf\cdot cm}$$

$$\varepsilon_t=0.003\,\frac{43.90-9.15}{9.15}=0.01139\ge\varepsilon_{ty}+0.003=0.0051\ \Rightarrow\ \phi=0.90,\qquad \phi M_{nox}=0.90\left(24.57\right)=22.11\ \mathrm{tf\cdot m}$$

Below $0.1f'_cA_g$, 21.2.2.3 does not apply.

### Step 3: Weak-Axis Strength at $P_u$

The section is 50 cm wide and 30 cm deep, with three layers of $7.60\ \mathrm{cm^2}$ at depths 6.10, 15.00, and 23.90 cm. Assume the first layer elastic in compression and inside the block, the other two yielded in tension:

$$0.85\left(280\right)\left(50\right)\left(0.85\right)c^2+\left[7.60\left(5762\right)-2\left(7.60\right)\left(4200\right)-27{,}778\right]c-7.60\left(6000\right)\left(6.10\right)=0$$

$$10{,}115c^2-47{,}826.8c-278{,}160=0\quad\Rightarrow\quad c=8.12\ \mathrm{cm},\qquad a=6.90\ \mathrm{cm}$$

$$\varepsilon_1=0.003\,\frac{8.12-6.10}{8.12}=0.000746\ \Rightarrow\ f_1=1492,\quad F_1=7.60\left(1492-238\right)=9530\ \mathrm{kgf}$$

$$\varepsilon_2=0.003\,\frac{15.00-8.12}{8.12}=0.00254>\varepsilon_{ty},\qquad \varepsilon_3=0.003\,\frac{23.90-8.12}{8.12}=0.00583>\varepsilon_{ty}\ \Rightarrow\ T_2=T_3=7.60\left(4200\right)=31{,}920\ \mathrm{kgf}$$

$$C_c=238\left(6.90\right)\left(50\right)=82{,}110,\qquad P_n=82{,}110+9530-31{,}920-31{,}920=27{,}800\ \mathrm{kgf}\ \checkmark$$

$$M_{noy}=82{,}110\left(15-\frac{6.90}{2}\right)+9530\left(8.90\right)+31{,}920\left(0\right)+31{,}920\left(8.90\right)=1{,}317{,}276\ \mathrm{kgf\cdot cm}$$

$$\varepsilon_t=0.00583\ge0.0051\ \Rightarrow\ \phi=0.90,\qquad \phi M_{noy}=0.90\left(13.17\right)=11.86\ \mathrm{tf\cdot m}$$

### Step 4: Load Contour Check

$$\frac{M_{ux}}{\phi M_{nox}}=\frac{14}{22.11}=0.633,\qquad \frac{M_{uy}}{\phi M_{noy}}=\frac{6}{11.86}=0.506$$

| Exponent | $0.633^{\alpha}+0.506^{\alpha}$ | Result |
|---|---|---|
| $\alpha=1.0$ | 1.139 | not satisfied |
| $\alpha=1.5$ | $0.504+0.360=0.864$ | satisfied |
| $\alpha=1.61$ ($\beta=0.65$) | 0.813 | satisfied |

The straight line rejects the column; the two curved contours accept it. Which is right?

### Step 5: The Exact Answer, and the Value of $\phi$

An exact analysis at $P_n=27.8$ tf on the ray $M_{ny}/M_{nx}=6/14$ (same procedure as Step 6 of example 2) gives $\theta=60.8^\circ$ and a strain in the extreme tension bar, corner bar 6, of only $\varepsilon_t=0.00408$: the biaxial point is in the transition region, not tension-controlled, and $\phi=0.65+0.25\left(0.00408-0.0021\right)/0.003=0.815$, not 0.90. Since $P_n=P_u/\phi$ changes with $\phi$, the analysis is repeated until both agree: $\phi=0.807$, $P_n=31.0$ tf, and the exact design strengths along the load ray are $\phi M_{nx}=15.38$ and $\phi M_{ny}=6.59\ \mathrm{tf\cdot m}$:

$$\frac{M_{ux}}{\phi M_{nx}}=\frac{14}{15.38}=0.91\le1\ \checkmark$$

The column is adequate. If the same $\phi=0.807$ is used for the uniaxial strengths at $P_n=31.0$ tf ($\phi M_{nox}=20.29$, $\phi M_{noy}=10.84\ \mathrm{tf\cdot m}$), the contour with $\alpha=1.5$ gives $0.985\le1$, still satisfied and much closer to the exact margin.

Three lessons. The straight line, $\alpha=1$, rejected an adequate column. The exponent 1.5 gave the right answer. And the $\phi$ read from the uniaxial diagrams, 0.90, was about 10 percent too high for the biaxial state: in a corner column near the limit of tension control, check the strain of the corner bar, or use the transition value of $\phi$.

## Summary

A column is a member, not only a cross section. Under axial load its deflections add moment, $P\delta$ along its length and $P\Delta$ at its ends when the story sways, and these second-order effects can make the moment the section must resist substantially larger than the first-order value. The idea goes back to Euler's buckling load; the design method of ACI CODE-318-25, the moment magnifier, comes from the research of MacGregor, Breen, and Pfrang and has been in the Code since 1971.

The procedure is: classify the story as nonsway or sway (6.6.4.3, $Q\le0.05$); check whether the slenderness ratio $k\ell_u/r$ is below the limits of 6.2.5.1; otherwise compute the critical load with a reduced stiffness $(EI)_{eff}$, the factor $C_m$ that converts the moment diagram into an equivalent uniform moment, and the magnifier $\delta=C_m/(1-P_u/0.75P_c)\ge1$; design for $M_c=\delta M_2$, never less than $M_{2,min}$ about each axis, and never more than 1.4 times the first-order moment. For the 6.0 m example column the strong-axis moment grew from 16.0 to 19.31 $\mathrm{tf\cdot m}$, still inside the diagram of notebook 07; but the weak axis, loaded only by the minimum moment, failed with a magnifier of 5.6 until a tie beam halved its unsupported length. The sensitivity table showed the 1.4 limit governing before strength at 7.0 m.

Under biaxial bending the interaction diagram becomes a surface, and the Code gives no simplified method. The exact strength follows from the assumptions of 22.2 with an inclined neutral axis; one evaluation is arithmetic, and checking the software solution of the corner column by hand reproduced $P_n=217.6$ tf with the neutral axis at $59.4^\circ$. Bresler's reciprocal load method, $1/P_n\approx1/P_{nx}+1/P_{ny}-1/P_o$, gave $\phi P_n=137.2$ tf against $P_u=100$ tf, 3 percent below the exact strength; it is slightly conservative above the balanced load, can be a few percent unconservative between $0.1f'_cA_g$ and the balanced load, and is not used below $0.1f'_cA_g$. There the load contour method, $\left(M_{ux}/\phi M_{nox}\right)^{\alpha}+\left(M_{uy}/\phi M_{noy}\right)^{\alpha}\le1$, takes over: for the lightly loaded corner column $\alpha=1$ rejected an adequate column, $\alpha=1.5$ accepted it correctly, and the exact analysis showed that $\phi$ drops from 0.90 to 0.81 because the corner bar strains less than in uniaxial bending. Finally, the combination left open by notebook 08 fails ($\phi P_n=83.8$ tf against 120 tf); widening the section to 40 x 50 cm with the same bars solves it, while adding steel to the 30 x 50 cm section does not. The technical-metric forms needed in this notebook are $E_c=15{,}100\sqrt{f'_c}$ and $M_{2,min}=P_u(1.5+0.03h)$, from ACI 318-19 Appendix C.

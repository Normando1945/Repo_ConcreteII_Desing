## Loads are the beginning of the structural argument

A structural member is not designed for a number written at the top of a calculation sheet. It is designed for actions that originate in gravity, use, climate, soil, and ground motion, then travel through a specific load path. Before selecting reinforcement, the engineer must know **what is acting**, **where it acts**, **how long it acts**, and **how it reaches the resisting system**.

This notebook develops that argument from the source of the action to the factored combinations used for strength design. It deliberately separates three ideas that are often blurred in early calculations: a load model, its effect on a member, and the resistance required to control that effect.

### Learning outcomes

After this notebook, you should be able to:

1. Classify the principal gravity and lateral actions in a building.
2. Construct a transparent dead-load takeoff from material properties and geometry.
3. Select the appropriate minimum live load from the intended occupancy.
4. Read a load combination as a model of simultaneous actions, rather than as a formula to memorize.
5. Identify the governing combination for a particular demand and recognize when uplift, overturning, or stability controls.

The code is not a menu of coefficients. It is a calibrated statement about uncertainty, coincidence, consequences, and acceptable reliability.

## 1. From an action to a structural demand

A useful first model is:

<div style="margin:22px 0; padding:16px 18px; border-left:4px solid #2563eb; background:#f8fafc; color:#1e293b; text-align:center; font-size:1.08em; line-height:1.8;">
  <b>Source of action</b> &rarr; <b>Applied load</b> &rarr; <b>Load path</b> &rarr; <b>Member action</b> &rarr; <b>Internal demand</b>
</div>

For example, people and furniture create a distributed floor load; the slab delivers it to beams or walls; those supports deliver reactions to columns or bearing walls; and the vertical system ultimately transfers the accumulated force to foundations and soil. At each transition, tributary area, geometry, continuity, and stiffness influence the force that arrives.

The same distinction is essential for lateral actions. Wind and earthquake are not simply horizontal arrows: they create inertial or pressure effects, diaphragm forces, collector actions, shears, overturning moments, axial-force changes, and foundation reactions. A correct member calculation with an incomplete load path is still an incomplete design.

### A compact taxonomy of building actions

| Family | Typical sources | Why the distinction matters |
|---|---|---|
| **Permanent gravity actions** | Self-weight, fixed finishes, partitions with fixed locations, permanently attached equipment | Their magnitude is usually estimated from geometry and unit weight; they remain for much of the structures life. |
| **Variable gravity actions** | Occupants, movable furniture, storage, maintenance, roof access, rain or snow | Their intensity, location, and simultaneous occurrence vary with use and time. |
| **Environmental actions** | Wind, earthquake, temperature, hail, flooding | They can reverse direction, create local peaks, or govern stability instead of ordinary gravity strength. |
| **Imposed deformation and time-dependent actions** | Creep, shrinkage, temperature change, differential settlement, prestress effects | They may create force only when deformation is restrained; their treatment depends on compatibility and stiffness. |

The classification is not academic housekeeping. It controls the load factor, the combination rule, the spatial arrangement to be checked, and sometimes the analysis model itself.

## 2. Normative language and load symbols

The course uses the notation that appears in the governing loading standard and in the concrete design code. In Ecuador, the applicable NEC provisions and the authority having jurisdiction govern the project; ASCE 7 is a central companion for the definition and combination of design actions. The symbol is only a label -- the design judgment comes from defining its magnitude, tributary area, direction, and combination correctly.

| Symbol | Action | Practical reading |
|---|---|---|
| $D$ | Dead load | Permanent weight of structural and nonstructural construction. |
| $L$ | Live load | Occupancy-related movable load on floors and accessible areas. |
| $L_r$ | Roof live load | Transient roof use, maintenance, and related occupancy effects. |
| $R$ | Rain load | Ponding or accumulated water on roof systems. |
| $S$ | Snow load | Environmental roof load where climate requires it. |
| $W$ | Wind load | Pressure and suction acting on the main wind-force-resisting system and components. |
| $E$ | Seismic effect | Effect associated with earthquake action, including horizontal and vertical components where required. |
| $F$ | Fluid load | Hydrostatic or hydrodynamic action from liquids or contained materials. |
| $H$ | Lateral earth pressure | Soil pressure, surcharge effects, and related retained-ground actions. |
| $T$ | Self-straining action | Temperature, creep, shrinkage, settlement, or similar imposed deformation effects. |
| $N$ | Notional load | Small hypothetical lateral action used in stability assessment when prescribed. |

A symbol should never be entered into a combination until its physical definition is documented. In particular, $E$, $W$, $H$, and $T$ are not single universal numbers; they depend on the system, site, direction, restraint, and analysis assumptions.

## 3. Factored load combinations: a reliability model

Strength design asks whether the available design resistance is at least as large as the largest credible factored effect:

$$\phi R_n \ge U$$

where $R_n$ is nominal resistance, $\phi$ is the resistance-reduction factor, and $U$ is the load effect from a prescribed combination. The factors are not merely conservatism sprinkled on a calculation: they recognize variability in loads, material properties, dimensions, analysis, and consequences.

For the basic group of combinations used in this course, the critical cases include:

| Check | Load combination | Structural question it represents |
|---|---|---|
| 1 | $1.4D$ | Can the permanent system carry a high estimate of its own weight? |
| 2 | $1.2D+1.6L+(0.5L_r or 0.3S or 0.5R)$ | Does ordinary occupancy, together with a companion roof action, govern? |
| 3 | $1.2D+(1.6L_r or 1.0S or 1.6R) + (L or 0.5W)$ | Does the roof action govern, with appropriate concurrent occupancy or wind? |
| 4 | $1.2D+1.0W+L+(0.5L_r or 0.3S or 0.5R)$ | What is the critical wind-plus-gravity demand? |
| 5 | $0.9D+1.0W$ | Is there enough stabilizing dead load under wind uplift or overturning? |
| 6 | $1.2D+E_v+E_h+L+0.15S$ | What is the seismic-plus-gravity strength demand? |
| 7 | $0.9D-E_v+E_h$ | Does reduced gravity combined with seismic action threaten stability or anchorage? |

The exact expressions and seismic definitions must match the governing edition and project jurisdiction. In classroom work, write the adopted code edition beside the calculation -- not in a footnote discovered after the fact.

### Reading combinations instead of memorizing them

Three habits prevent most combination errors:

- **Do not add actions that cannot coexist.** The reduced companion factors reflect that maxima rarely occur simultaneously.
- **Check both signs of lateral action.** Reversing wind or seismic direction can change tension into compression, uplift into bearing, and a harmless reaction into a critical anchor force.
- **Seek the governing effect, not the largest total load.** A beam moment, a column axial force, a wall shear, an anchor tension, and a footing pressure can each be governed by different combinations.

A good calculation table therefore lists every candidate combination, the effect being compared, and the controlling row. A bare statement such as combination 4 governs is not a design record; it is a conclusion without its argument.

## 4. Dead loads: build the takeoff, do not guess it

Dead load is the weight of construction that is permanently incorporated into the building. It includes the structural frame, slabs, roofs, masonry, cladding, finishes, ceilings, fixed services, built-in equipment, and permanent landscaping where applicable. The calculation begins with either mass density <i>&rho;</i> or unit weight <i>&gamma;</i>:

<div style="margin:18px 0; text-align:center; font-family:Georgia,serif; font-size:1.15em; color:#1e293b;">&gamma; = &rho;g &nbsp;&nbsp;&middot;&nbsp;&nbsp; w = &gamma;t &nbsp;&nbsp;&middot;&nbsp;&nbsp; W = &gamma;V</div>

For a layer over an area, <i>w</i> is an area load; for a wall, the same material property is converted through thickness and height into a line load; and for a discrete component it becomes a concentrated load. Unit consistency is a first-order design check, not decoration around the answer.

**Recommended takeoff sequence:** identify each layer, state thickness or geometry, choose the unit weight, calculate its load, state where it acts, and then assign it to the correct tributary system. This is how a model stays auditable when the architecture changes.

### Hail and roof accumulation

Roof design is sensitive to local climate and geometry. The course reference identifies hail checks for regions above 1500 m above sea level, with values that depend on roof slope and an additional edge-zone check at eaves. These actions should not be treated as an arbitrary uniform surcharge: drainage paths, parapets, roof falls, blocked drains, deflection, and ponding stability can determine the controlling demand.

For every roof, record the adopted rainfall, hail, snow, and roof-live-load assumptions separately. They are not interchangeable names for one generic roof load.

## 5. Live loads: occupancy translated into structural demand

Live load represents the maximum expected occupancy-related load for the intended use of the area. It is variable in magnitude and location, so the structural model must consider the pattern that creates the largest relevant effect. A continuous beam may need a loaded/unloaded bay pattern for negative moment; a column must collect tributary load over several floors; a slab may be governed by a concentrated maintenance action even when its average uniform load is modest.

The minimum value is selected by **use**, not by the owners informal description of the room. A corridor serving assembly occupancy, a storage room, a residential bedroom, a parking deck, and a mechanical room are structurally different promises. If the use is uncertain, resolve it with the architect or specify a conservative design basis -- do not hide the uncertainty inside a favorite number.

### Live-load reduction is a statistical rule, not a default button

Large tributary areas or multiple supported floors may permit code-based live-load reduction because the probability that every square metre simultaneously reaches its maximum is small. That does **not** mean reduction is automatic. It depends on occupancy, member type, tributary area, number of floors, code limits, and sometimes a concentrated-load check that cannot be reduced.

Before applying a reduction, explicitly answer: What area does this member support? Which occupancy governs? Is the member exempt? Does another load effect -- such as vibration, deflection, local punching, storage, or a roof condition -- remain controlling?

## 6. Lateral actions complete the model

The most important lateral actions for ordinary building design are wind and earthquake. Their full characterization is developed in later notebooks; here the essential point is that they enter the same structural argument as gravity loads. A diaphragm must collect and distribute inertial or wind forces, vertical elements must resist shear and overturning, collectors and connections must transfer forces between systems, and foundations must close the equilibrium.

In much of Ecuador, seismic demand commonly controls the lateral-force-resisting system. That does not make wind optional. Wind can govern cladding, roof uplift, members with low seismic participation, construction stages, serviceability, and the stability combinations with reduced dead load. Design the whole system, not only the force you expect to win the final comparison.

### Where load errors hide

| Weak practice | Why it fails | Better design habit |
|---|---|---|
| Applying a uniform floor load without showing tributary width | The force transfer is invisible and cannot be checked. | Draw the tributary boundaries and convert area load to line or point load explicitly. |
| Using a generic wall weight for every partition | Masonry type, thickness, height, openings, and finishes vary. | Calculate each wall family or carry a clearly stated conservative allowance. |
| Checking only the largest downward combination | Uplift, overturning, anchor tension, and soil pressure can govern under reduced $D$. | Include both gravity-dominant and stability combinations. |
| Treating an architectural change as nonstructural | A finish, planter, facade, or equipment change can alter member and foundation demands. | Maintain a load schedule linked to drawings and revise it with the design. |
| Copying a table value without the code edition | Values and exceptions can change; jurisdiction may differ. | State the adopted standard, edition, occupancy, and project assumption beside the load. |

## 7. A repeatable load-model workflow

1. **Read the architecture and intended use.** Identify occupancies, spans, walls, roof geometry, equipment, facade, and possible future change.
2. **Trace the load path.** Locate diaphragms, supports, collectors, vertical elements, foundations, and discontinuities.
3. **Build the service-load schedule.** Separate $D$, $L$, roof actions, environmental actions, and imposed deformations; show units and locations.
4. **Convert loads through tributary geometry.** Move systematically from area loads to line loads, reactions, and member forces.
5. **Apply every required combination.** Preserve sign, direction, and companion-action rules.
6. **Compare the relevant effects.** Moment, shear, axial force, deflection, drift, bearing, uplift, and connection forces may each have a different controlling case.
7. **Record assumptions.** A load schedule is part of the engineering model and should survive peer review, revision, and construction.

This workflow is deliberately slower than pasting a coefficient into software. It is also the reason the final model can be trusted.

### Closing perspective

Loads are not a preliminary nuisance before the real reinforced-concrete design begins. They are the first physical claim made by the project. If the load model is incomplete, the analysis can be elegant and the reinforcement calculation can be exact -- yet the building has still not been designed for the problem it will actually face.

The next notebooks use these actions to develop member demands and resistance. Keep the chain visible: **source -> load -> path -> effect -> resistance -> detail**.

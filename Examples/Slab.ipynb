### Libraries

```python
import numpy as np
import pandas as pd
```

### Data

```python
lx = 6
ly = 6

w_total = 3.5
```

### One-Way and Two-Way Action

```python
wxx = []
wyy = []
lxx = []
lyy = []

for lx in np.arange(ly,2*ly,0.1):
    lxx.append(lx)
    lyy.append(ly)
    wy = w_total / (1 + (ly/lx)**4)
    wx = w_total - wy
    wxx.append(wx)
    wyy.append(wy)

resul_df = pd.DataFrame({
    'lx': lxx,
    'ly': lyy,
    'wx': wxx,
    'wy': wyy
})
resul_df.head(5)
```

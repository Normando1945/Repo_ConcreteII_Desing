### Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from repo__CORE_concreteII_desing import *
```

### Data

```python
fc = 280
fy = 4200
lx = 700
ly = 300
bw = 120
dvarF = 1.2
# One end continuous
H = np.ceil(np.max([lx, ly]) / 24 / 5) * 5
MuI = 6.29 * 1e5
MuS = 11.16 * 1e5
fi = 0.9
fiC = 0.75
rec= 2.5
Vu = 7610
tc = 40
tl = 40
alfas = 40
```

### One-Way and Two-Way Action

```python
wy = (lx/ly)**(4)
porc = wy / (np.ceil(wy) + 1) * 100

if porc > 75:
    mensaje = {'one way slab'}
else:
    mensaje = {'two way slab'}

print('='*120)
print(f'Slab behavior = {mensaje}')
print(f'H min = {H:.2f} cm')
print(f'%  = {porc:.2f}')
print('='*120)
```

### Calculation of As and Asmin

```python
capacityFI = SimpleFlexuralCapacity(fc = fc, fy = fy, bw = bw, H = H, rec = rec, fi = fi,diF = dvarF,Mu = MuI)
capacityFS = SimpleFlexuralCapacity(fc = fc, fy = fy, bw = bw, H = H, rec = rec, fi = fi,diF = dvarF,Mu = MuS)

[AsexcI, AsminI, S_excI, S_minI, numvar_excI, numvar_minI, fiMnI, MuI] = capacityFI.as_asmin()
[AsexcS, AsminS, S_excS, S_minS, numvar_excS, numvar_minS, fiMnS, MuS] = capacityFS.as_asmin()
```

#### Shear Capacity

```python
capacityV = SimpleShearSlab(fc = fc, fy = fy, bw = bw, H = H, rec = rec, fi = fiC, diC = 1.0, Vu = Vu, lamda= 1, tc = tc, tl = tl, alfas= alfas)
capacityV.Vc_Slab()
```
